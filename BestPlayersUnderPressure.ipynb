```python
# Join playing_eleven with competition information
player_match_analysis = (
    playing_eleven_copy
    .merge(
        matches_copy[["match_id", "comp_id"]],
        on="match_id",
        how="left"
    )
    .merge(
        competitions_copy[["comp_id", "comp_name", "match_format"]],
        on="comp_id",
        how="left"
    )
)

# Players listed per match and impact player counts
match_player_counts = (
    player_match_analysis
    .groupby(["match_id", "comp_name", "match_format"])
    .agg(
        total_listed=("player_id", "nunique"),
        played_count=("is_played", "sum"),
        impact_players=("is_impact", "sum"),
        impact_off_players=("is_impact_off", "sum")
    )
    .reset_index()
)

print("Player count distribution by format:")
print(
    match_player_counts
    .groupby(["comp_name", "match_format"])["total_listed"]
    .describe()
    .round(2)
)

print("\nMatches with more than 11 listed players:")
print(
    match_player_counts[
        match_player_counts["total_listed"] > 11
    ].head(20).to_string(index=False)
)
```

```python
# Filter IPL competitions
ipl_analysis = match_player_counts[
    match_player_counts["comp_name"].str.contains(
        "IPL", case=False, na=False
    )
].copy()

# Season-wise summary
print(
    ipl_analysis.groupby("comp_name").agg(
        total_matches=("match_id", "nunique"),
        min_players=("total_listed", "min"),
        max_players=("total_listed", "max"),
        avg_players=("total_listed", "mean"),
        min_played=("played_count", "min"),
        max_played=("played_count", "max"),
        total_impact_players=("impact_players", "sum"),
        total_impact_off=("impact_off_players", "sum")
    ).round(2).to_string()
)
```

```python
print(
    player_match_analysis[
        ["is_played", "is_impact", "is_impact_off"]
    ].value_counts(dropna=False).sort_index()
)
```

```python
# Collect all player IDs appearing in ball-by-ball data
ball_player_ids = set(
    ball_by_ball_copy[
        ["striker_id", "non_striker_id", "bowler_id"]
    ].stack().dropna().unique()
)

player_match_analysis["appears_in_ball_data"] = (
    player_match_analysis["player_id"].isin(ball_player_ids)
)

print(
    pd.crosstab(
        player_match_analysis["is_played"],
        player_match_analysis["appears_in_ball_data"],
        margins=True
    )
)
```

```python
# 1. Check duplicate match IDs in matches
print("MATCHES TABLE")
print("Total rows:", len(matches_copy))
print("Unique match IDs:", matches_copy["match_id"].nunique())
print("Duplicate match IDs:", matches_copy["match_id"].duplicated().sum())

# 2. Check duplicate competition IDs
print("\nCOMPETITIONS TABLE")
print("Total rows:", len(competitions_copy))
print("Unique competition IDs:", competitions_copy["comp_id"].nunique())
print("Duplicate competition IDs:", competitions_copy["comp_id"].duplicated().sum())

# 3. Check duplicate player records within a match and team
print("\nPLAYING ELEVEN TABLE")
print("Total rows:", len(playing_eleven_copy))
print(
    "Duplicate match/team/player records:",
    playing_eleven_copy.duplicated(
        subset=["match_id", "team_id", "player_id"]
    ).sum()
)

# 4. Check whether any match ID maps to multiple competitions
print("\nMATCH IDs MAPPED TO MULTIPLE COMPETITIONS")

match_comp_counts = (
    matches_copy.groupby("match_id")["comp_id"].nunique()
)

print("Match IDs with multiple competition IDs:",
      (match_comp_counts > 1).sum())

# 5. Compare row counts before and after joining
print("\nJOIN VALIDATION")
print("Original playing_eleven rows:", len(playing_eleven_copy))
print("Joined player_match_analysis rows:", len(player_match_analysis))
```

```python
# STEP A: Inspect duplicate player records
duplicate_players = playing_eleven_copy[
    playing_eleven_copy.duplicated(
        subset=["match_id", "team_id", "player_id"],
        keep=False
    )
].sort_values(["match_id", "team_id", "player_id"])

print("DUPLICATE PLAYER RECORDS")
print(
    duplicate_players[
        [
            "match_id", "team_id", "player_id",
            "batting_order", "is_played",
            "is_impact", "is_impact_off"
        ]
    ].to_string(index=False)
)

# STEP B: Find IPL matches with the highest player counts
print("\nIPL MATCHES WITH HIGHEST PLAYER COUNTS")

print(
    ipl_analysis
    .sort_values("total_listed", ascending=False)
    [["match_id", "comp_name", "total_listed", "played_count"]]
    .head(10)
    .to_string(index=False)
)

# STEP C: Inspect team-wise player counts in those matches
top_match_ids = (
    ipl_analysis
    .nlargest(5, "total_listed")["match_id"]
    .tolist()
)

print("\nTEAM-WISE PLAYER COUNTS")

for match_id in top_match_ids:
    print(f"\nMatch ID: {match_id}")

    match_data = player_match_analysis[
        player_match_analysis["match_id"] == match_id
    ]

    print(
        match_data.groupby("team_id").agg(
            total_rows=("player_id", "size"),
            unique_players=("player_id", "nunique"),
            played=("is_played", "sum"),
            impact=("is_impact", "sum"),
            impact_off=("is_impact_off", "sum")
        ).to_string()
    )
```

```python
match_id = 31243

# Match information
print("MATCH INFORMATION")
print(
    matches_copy.loc[
        matches_copy["match_id"] == match_id
    ].to_string(index=False)
)

# All player records for this match
print("\nPLAYER RECORDS")

match_players = player_match_analysis[
    player_match_analysis["match_id"] == match_id
].copy()

print(
    match_players[
        [
            "team_id", "player_id", "batting_order",
            "is_played", "is_captain", "is_wk",
            "is_impact", "is_impact_off"
        ]
    ].sort_values(["team_id", "batting_order"]).to_string(index=False)
)

# Actual players appearing in this match's ball-by-ball data
print("\nBALL-BY-BALL PARTICIPATION")

match_balls = ball_by_ball_copy[
    ball_by_ball_copy["match_id"] == match_id
]

for team_id, team_players in match_players.groupby("team_id"):
    actual_players = set(
        match_balls.loc[
            match_balls["batting_team_id"] == team_id,
            ["striker_id", "non_striker_id"]
        ].values.flatten()
    )

    actual_players.update(
        match_balls.loc[
            match_balls["bowling_team_id"] == team_id,
            "bowler_id"
        ].dropna().tolist()
    )

    actual_players.discard(None)
    actual_players.discard(float("nan"))

    print(f"\nTeam ID: {team_id}")
    print("Listed unique players:", team_players["player_id"].nunique())
    print("Players appearing in ball-by-ball:", len(actual_players))
    print("Players marked is_played:", team_players["is_played"].sum())
```

```python
duplicate_summary = (
    duplicate_players
    .groupby(["match_id", "team_id", "player_id"])
    .agg(
        record_count=("batting_order", "size"),
        unique_batting_orders=("batting_order", "nunique"),
        batting_orders=(
            "batting_order",
            lambda x: sorted(x.dropna().unique().tolist())
        ),
        played_values=("is_played", "nunique"),
        impact_values=("is_impact", "nunique"),
        impact_off_values=("is_impact_off", "nunique")
    )
    .reset_index()
)

print("Duplicate player keys:", len(duplicate_summary))
print("\nRecords per player:")
print(duplicate_summary["record_count"].value_counts().sort_index())

print("\nBatting-order counts:")
print(duplicate_summary["unique_batting_orders"].value_counts().sort_index())

print("\nFlag inconsistencies:")
print(
    duplicate_summary[
        (duplicate_summary["played_values"] > 1) |
        (duplicate_summary["impact_values"] > 1) |
        (duplicate_summary["impact_off_values"] > 1)
    ]
)
```

```python
# Step 1: Check competition names related to IPL
ipl_competitions = competitions_copy[
    competitions_copy["comp_name"].str.contains(
        "IPL|Premier League",
        case=False,
        na=False
    )
]

print("Possible IPL competitions:")
display(ipl_competitions)

# Step 2: Check impact flag values before filtering
playing_eleven_with_comp = (
    playing_eleven_analysis
    .merge(
        matches_copy[["match_id", "comp_id"]],
        on="match_id",
        how="left"
    )
    .merge(
        competitions_copy[["comp_id", "comp_name", "season"]],
        on="comp_id",
        how="left"
    )
)

print("\nImpact flag distributions:")
print(playing_eleven_with_comp["is_impact"].value_counts(dropna=False))
print(playing_eleven_with_comp["is_impact_off"].value_counts(dropna=False))

# Step 3: Check whether any impact records exist at all
impact_mask = (
    (playing_eleven_with_comp["is_impact"] == 1) |
    (playing_eleven_with_comp["is_impact_off"] == 1)
)

print("\nTotal impact records across all competitions:", impact_mask.sum())

print("\nCompetition names containing impact records:")
display(
    playing_eleven_with_comp.loc[
        impact_mask,
        ["comp_id", "comp_name", "season"]
    ].drop_duplicates().sort_values(["season", "comp_name"])
)
```

```python
ipl_impact_records = playing_eleven_with_comp[
    (
        playing_eleven_with_comp["comp_name"].str.match(
            r"IPL \d{4}",
            case=False,
            na=False
        )
    ) &
    (
        (playing_eleven_with_comp["is_impact"] == 1) |
        (playing_eleven_with_comp["is_impact_off"] == 1)
    )
].sort_values(["season", "match_id", "team_id"])

print("Total IPL impact records:", len(ipl_impact_records))

print("\nRecords by season:")
display(
    ipl_impact_records.groupby("season").agg(
        impact_in=("is_impact", "sum"),
        impact_out=("is_impact_off", "sum"),
        total_records=("player_id", "count")
    ).reset_index()
)

display(ipl_impact_records)
```

```python
impact_match_team_audit = (
    ipl_impact_records
    .groupby(
        ["season", "match_id", "team_id"],
        as_index=False
    )
    .agg(
        incoming=("is_impact", "sum"),
        outgoing=("is_impact_off", "sum"),
        flagged_records=("player_id", "count"),
        unique_players=("player_id", "nunique")
    )
)

# Find teams where incoming/outgoing counts do not match
impact_exceptions = impact_match_team_audit[
    (impact_match_team_audit["incoming"] != 1) |
    (impact_match_team_audit["outgoing"] != 1) |
    (impact_match_team_audit["flagged_records"] != 2) |
    (impact_match_team_audit["unique_players"] != 2)
]

print("Total match-team combinations:", len(impact_match_team_audit))
print("Exceptions:", len(impact_exceptions))

display(impact_exceptions)
```

```python
# Unique players appearing in ball-by-ball data
ball_participants = pd.concat([
    ball_by_ball_copy[["match_id", "striker_id"]]
        .rename(columns={"striker_id": "player_id"}),

    ball_by_ball_copy[["match_id", "non_striker_id"]]
        .rename(columns={"non_striker_id": "player_id"}),

    ball_by_ball_copy[["match_id", "bowler_id"]]
        .rename(columns={"bowler_id": "player_id"})
]).dropna().drop_duplicates()

ball_participants["player_id"] = (
    ball_participants["player_id"].astype(int)
)

# Check participation of each impact player
impact_participation_audit = ipl_impact_records.merge(
    ball_participants.assign(in_ball_data=True),
    on=["match_id", "player_id"],
    how="left"
)

impact_participation_audit["in_ball_data"] = (
    impact_participation_audit["in_ball_data"].fillna(False)
)

print("Impact records:", len(impact_participation_audit))

display(
    impact_participation_audit.groupby(
        ["is_impact", "is_impact_off", "in_ball_data"]
    ).size().reset_index(name="records")
)
```

```python
# Get impact players absent from ball-by-ball data
missing_impact_participation = impact_participation_audit[
    ~impact_participation_audit["in_ball_data"]
].copy()

# Add player names
missing_impact_participation = missing_impact_participation.merge(
    competition_mapping_copy[
        ["comp_id", "player_id", "player_name"]
    ].drop_duplicates(),
    on=["comp_id", "player_id"],
    how="left"
)

# Add match details
missing_impact_participation = missing_impact_participation.merge(
    matches_copy[
        ["match_id", "match_date", "home_team", "away_team"]
    ],
    on="match_id",
    how="left"
)

# Add impact role
missing_impact_participation["impact_role"] = (
    missing_impact_participation["is_impact"]
    .map({1: "Incoming", 0: "Outgoing"})
)

columns_to_display = [
    "season",
    "match_id",
    "match_date",
    "home_team",
    "away_team",
    "team_id",
    "player_id",
    "player_name",
    "impact_role",
    "is_played",
    "batting_order"
]

display(
    missing_impact_participation[columns_to_display]
    .sort_values(["season", "match_id", "team_id"])
)
```

```python
# Build a match-by-match verification report for the 14 exceptions

other_exceptions_report = other_audit[
    other_audit["extras_2_runs"] > 0
].copy()

other_exceptions_report = (
    other_exceptions_report
    .merge(
        matches_copy[
            [
                "match_id",
                "match_date",
                "comp_id",
                "home_team",
                "away_team",
                "venue_country"
            ]
        ],
        on="match_id",
        how="left"
    )
    .merge(
        competitions_copy[
            ["comp_id", "comp_name", "match_format"]
        ],
        on="comp_id",
        how="left"
    )
)

# Sort records in match and delivery order
other_exceptions_report = other_exceptions_report.sort_values(
    ["match_date", "match_id", "innings", "overid", "over_ball_num"]
)

# Create a readable over number (overid is zero-based)
other_exceptions_report["over_number"] = (
    other_exceptions_report["overid"] + 1
)

# Prepare manual verification columns
verification_columns = [
    "match_id",
    "match_date",
    "comp_name",
    "match_format",
    "home_team",
    "away_team",
    "venue_country",
    "innings",
    "over_number",
    "over_ball_num",
    "bowler_name",
    "striker_name",
    "non_striker_name",
    "runs",
    "extras1_runs",
    "extras_2_runs",
    "delivery_style",
    "dismissal_type",
    "dismissed_player_id"
]

other_exceptions_report = other_exceptions_report[
    verification_columns
].copy()

print("Total exceptional deliveries:", len(other_exceptions_report))
print("Unique matches:", other_exceptions_report["match_id"].nunique())

# Display one match at a time
for match_id, match_data in other_exceptions_report.groupby(
    "match_id", sort=False
):
    first_row = match_data.iloc[0]

    print("\n" + "=" * 100)
    print(
        f"Match ID: {match_id} | "
        f"Date: {first_row['match_date']} | "
        f"{first_row['home_team']} vs {first_row['away_team']}"
    )
    print(
        f"Competition: {first_row['comp_name']} | "
        f"Format: {first_row['match_format']} | "
        f"Venue country: {first_row['venue_country']}"
    )
    print("=" * 100)

    display(
        match_data[
            [
                "innings",
                "over_number",
                "over_ball_num",
                "bowler_name",
                "striker_name",
                "non_striker_name",
                "runs",
                "extras1_runs",
                "extras_2_runs",
                "delivery_style",
                "dismissal_type"
            ]
        ].reset_index(drop=True)
    )
```

```python

import pandas as pd

# Always create a separate analytical copy
ball_by_ball_analysis = ball_by_ball_copy.copy(deep=True)

# Calculate scoring values without modifying original columns
ball_by_ball_analysis["batter_runs"] = pd.to_numeric(
    ball_by_ball_analysis["runs"], errors="coerce"
)

ball_by_ball_analysis["total_extras"] = (
    pd.to_numeric(ball_by_ball_analysis["extras1_runs"], errors="coerce")
    + pd.to_numeric(ball_by_ball_analysis["extras_2_runs"], errors="coerce")
)

ball_by_ball_analysis["team_runs"] = (
    ball_by_ball_analysis["batter_runs"]
    + ball_by_ball_analysis["total_extras"]
)

# Initialize derived classifications
ball_by_ball_analysis["extras_category"] = "Unknown"
ball_by_ball_analysis["delivery_legality"] = "Unknown"
ball_by_ball_analysis["is_legal_delivery"] = pd.Series(
    pd.NA, index=ball_by_ball_analysis.index, dtype="boolean"
)

ball_by_ball_analysis["classification_confidence"] = "Unknown"
ball_by_ball_analysis["classification_source"] = "Unresolved"
ball_by_ball_analysis["classification_note"] = None

# Classifications directly supported by the source category
source_category_mapping = {
    "wide": "Wide",
    "byes": "Bye",
    "legbyes": "Leg-bye",
    "Overstep": "No-ball",
    "Penalty": "Penalty"
}

for source_category, derived_category in source_category_mapping.items():
    mask = ball_by_ball_analysis["extras_type"].eq(source_category)

    ball_by_ball_analysis.loc[
        mask, "extras_category"
    ] = derived_category

    ball_by_ball_analysis.loc[
        mask, "classification_confidence"
    ] = "Source-labelled"

    ball_by_ball_analysis.loc[
        mask, "classification_source"
    ] = "Original extras_type"

# Confirmed illegal delivery classifications
wide_mask = ball_by_ball_analysis["extras_type"].eq("wide")
no_ball_mask = ball_by_ball_analysis["extras_type"].eq("Overstep")

ball_by_ball_analysis.loc[
    wide_mask, "delivery_legality"
] = "Wide"

ball_by_ball_analysis.loc[
    no_ball_mask, "delivery_legality"
] = "No-ball"

ball_by_ball_analysis.loc[
    wide_mask | no_ball_mask, "is_legal_delivery"
] = False

# Manually verified missing-type leg-byes
manual_legbye_records = [
    (17555, 1, 5, 4),
    (36097, 2, 8, 6)
]

for match_id, innings, overid, ball_num in manual_legbye_records:
    mask = (
        ball_by_ball_analysis["match_id"].eq(match_id)
        & ball_by_ball_analysis["innings"].eq(innings)
        & ball_by_ball_analysis["overid"].eq(overid)
        & ball_by_ball_analysis["over_ball_num"].eq(ball_num)
    )

    ball_by_ball_analysis.loc[
        mask, "extras_category"
    ] = "Leg-bye"

    ball_by_ball_analysis.loc[
        mask, "classification_confidence"
    ] = "Manually verified"

    ball_by_ball_analysis.loc[
        mask, "classification_source"
    ] = "User manual verification"

    ball_by_ball_analysis.loc[
        mask, "classification_note"
    ] = "Manually verified leg-bye"

# Manually verified penalty delivery
penalty_mask = (
    ball_by_ball_analysis["match_id"].eq(16753)
    & ball_by_ball_analysis["innings"].eq(2)
    & ball_by_ball_analysis["overid"].eq(6)
    & ball_by_ball_analysis["over_ball_num"].eq(4)
)

ball_by_ball_analysis.loc[
    penalty_mask, "extras_category"
] = "Penalty"

ball_by_ball_analysis.loc[
    penalty_mask, "classification_confidence"
] = "Manually verified"

ball_by_ball_analysis.loc[
    penalty_mask, "classification_source"
] = "User manual verification"

ball_by_ball_analysis.loc[
    penalty_mask, "classification_note"
] = "5 penalty runs: ball touched wicketkeeper's helmet"

# Validation output
print("Original rows:", len(ball_by_ball_copy))
print("Derived rows:", len(ball_by_ball_analysis))

print(
    "Original data unchanged:",
    ball_by_ball_copy.equals(
        ball_by_ball_analysis[ball_by_ball_copy.columns]
    )
)

print("\nExtras category distribution:")
display(
    ball_by_ball_analysis["extras_category"]
    .value_counts(dropna=False)
    .to_frame("records")
)

print("\nDelivery legality distribution:")
display(
    ball_by_ball_analysis["delivery_legality"]
    .value_counts(dropna=False)
    .to_frame("records")
)

print("\nClassification confidence:")
display(
    ball_by_ball_analysis["classification_confidence"]
    .value_counts(dropna=False)
    .to_frame("records")
)

print("\nManually verified records found:")
print(
    ball_by_ball_analysis[
        ball_by_ball_analysis["classification_confidence"]
        == "Manually verified"
    ][
        [
            "match_id", "innings", "overid", "over_ball_num",
            "extras_type", "extras_category",
            "classification_note"
        ]
    ].to_string(index=False)
)
```

```python

# Identify ordinary deliveries with no recorded extras
no_extras_mask = (
    ball_by_ball_analysis["extras_type"].isna()
    & ball_by_ball_analysis["total_extras"].eq(0)
)

ball_by_ball_analysis.loc[
    no_extras_mask, "extras_category"
] = "None"

ball_by_ball_analysis.loc[
    no_extras_mask, "classification_confidence"
] = "Not applicable"

ball_by_ball_analysis.loc[
    no_extras_mask, "classification_source"
] = "No extras recorded"

# Recheck classification
print("Extras category distribution:")
display(
    ball_by_ball_analysis["extras_category"]
    .value_counts()
    .to_frame("records")
)

print("\nUnresolved records:")
display(
    ball_by_ball_analysis[
        ball_by_ball_analysis["extras_category"] == "Unknown"
    ]["extras_type"].value_counts(dropna=False)
)
```

```python
verified_other_no_ball_match_ids = [
    13614, 14127, 14159, 20188,
    20234, 20243, 21583, 36200,
    37093, 37278, 40038, 40695,
    48467, 55078
]

# Locate the 14 previously verified deliveries
verified_mask = (
    ball_by_ball_analysis["match_id"].isin(
        verified_other_no_ball_match_ids
    )
    & ball_by_ball_analysis["extras_type"].eq("other")
    & ball_by_ball_analysis["extras_2_runs"].gt(0)
)

verified_records = ball_by_ball_analysis.loc[verified_mask]

print("Verified records found:", len(verified_records))
print("Unique matches:", verified_records["match_id"].nunique())

# Safety checks: do not classify if the records don't match expectations
assert len(verified_records) == 14, "Expected exactly 14 records"
assert verified_records["match_id"].nunique() == 14
assert verified_records["runs"].eq(0).all()
assert verified_records["extras1_runs"].eq(1).all()

# Apply manually verified delivery legality
ball_by_ball_analysis.loc[
    verified_mask, "delivery_legality"
] = "No-ball"

ball_by_ball_analysis.loc[
    verified_mask, "is_legal_delivery"
] = False

ball_by_ball_analysis.loc[
    verified_mask, "classification_confidence"
] = "Manually verified"

ball_by_ball_analysis.loc[
    verified_mask, "classification_source"
] = "User-provided match commentary"

ball_by_ball_analysis.loc[
    verified_mask, "classification_note"
] = (
    "Manually verified no-ball; "
    "additional runs recorded as byes or leg-byes"
)

print("\nUpdated delivery legality:")
display(
    ball_by_ball_analysis["delivery_legality"]
    .value_counts()
    .to_frame("records")
)

print("\nVerified records:")
display(
    ball_by_ball_analysis.loc[
        verified_mask,
        [
            "match_id", "runs", "extras1_runs",
            "extras_2_runs", "extras_category",
            "delivery_legality"
        ]
    ].sort_values("match_id")
)
```

```python
remaining_other = ball_by_ball_analysis[
    (ball_by_ball_analysis["extras_type"] == "other")
    & (~verified_mask)
].copy()

print("Remaining other records:", len(remaining_other))

display(
    remaining_other.groupby(
        ["runs", "extras1_runs", "extras_2_runs"],
        dropna=False
    ).size().reset_index(name="records")
)

print("\nWicket distribution:")
display(remaining_other["dismissal_type"].value_counts(dropna=False))

print("\nDelivery style distribution:")
display(remaining_other["delivery_style"].value_counts(dropna=False))
```

```python

# FINAL DATASET VALIDATION

print("========== DATASET VALIDATION ==========")

# 1. Row count
print("\n1. Row Count")
print("Raw:", len(ball_by_ball_copy))
print("Analysis:", len(ball_by_ball_analysis))
print("Match:", len(ball_by_ball_copy) == len(ball_by_ball_analysis))

# 2. Duplicate delivery keys
duplicate_deliveries = ball_by_ball_analysis.duplicated(
    subset=["match_id", "innings", "innings_ball_num"]
).sum()

print("\n2. Duplicate Deliveries:", duplicate_deliveries)

# 3. Missing scoring values
scoring_columns = [
    "batter_runs",
    "total_extras",
    "team_runs"
]

print("\n3. Missing Scoring Values")
display(
    ball_by_ball_analysis[scoring_columns].isna().sum()
)

# 4. Negative scoring values
print("\n4. Negative Scoring Values")
display(
    (
        ball_by_ball_analysis[scoring_columns] < 0
    ).sum().to_frame("negative_records")
)

# 5. Verify scoring formula
expected_team_runs = (
    ball_by_ball_analysis["batter_runs"]
    + ball_by_ball_analysis["total_extras"]
)

scoring_mismatches = (
    ball_by_ball_analysis["team_runs"] != expected_team_runs
).sum()

print("\n5. Scoring Formula Mismatches:", scoring_mismatches)

# 6. Verify known illegal deliveries
print("\n6. Known Illegal Deliveries")

display(
    ball_by_ball_analysis.groupby(
        "delivery_legality"
    ).size().to_frame("records")
)

# 7. Remaining unresolved 'other' records
print("\n7. Unresolved Other Records")

unresolved_other = ball_by_ball_analysis[
    (ball_by_ball_analysis["extras_type"] == "other")
    & (ball_by_ball_analysis["delivery_legality"] == "Unknown")
]

print("Unresolved other records:", len(unresolved_other))

# 8. Raw data integrity
print("\n8. Raw Data Integrity")

print(
    "Original columns unchanged:",
    ball_by_ball_copy.equals(
        ball_by_ball_analysis[ball_by_ball_copy.columns]
    )
)
```

```python
format_thresholds = {
    "T20": {"low": 159, "high": 207},
    "T20I": {"low": 143, "high": 191}
}

first_innings_analysis["scoring_environment"] = (
    first_innings_analysis.apply(
        lambda row: (
            "Low"
            if row["total_runs"] < format_thresholds[row["match_format"]]["low"]
            else "High"
            if row["total_runs"] > format_thresholds[row["match_format"]]["high"]
            else "Normal"
        ),
        axis=1
    )
)

display(
    first_innings_analysis[
        ["match_id", "match_format", "total_runs", "scoring_environment"]
    ].head(10)
)
```

```python
# Create a separate delivery-level pressure dataset
standard_pressure_deliveries = standard_deliveries.copy(deep=True)

# Ensure chronological delivery order
standard_pressure_deliveries = (
    standard_pressure_deliveries
    .sort_values(
        ["match_id", "innings", "innings_ball_num"]
    )
    .reset_index(drop=True)
)

group_keys = ["match_id", "innings"]

# Runs scored before the current delivery
standard_pressure_deliveries["runs_before_delivery"] = (
    standard_pressure_deliveries
    .groupby(group_keys)["team_runs"]
    .cumsum()
    - standard_pressure_deliveries["team_runs"]
)

# Wickets lost before the current delivery
standard_pressure_deliveries["wickets_lost_before_delivery"] = (
    standard_pressure_deliveries
    .groupby(group_keys)["is_wicket"]
    .cumsum()
    - standard_pressure_deliveries["is_wicket"]
)

# Wickets remaining before the current delivery
standard_pressure_deliveries["wickets_remaining"] = (
    10 - standard_pressure_deliveries["wickets_lost_before_delivery"]
)

# Legal balls before the current delivery
legal_delivery_flag = (
    standard_pressure_deliveries["is_legal_delivery"]
    .eq(True)
    .fillna(False)
    .astype(int)
)

standard_pressure_deliveries["legal_balls_before_delivery"] = (
    legal_delivery_flag
    .groupby([
        standard_pressure_deliveries["match_id"],
        standard_pressure_deliveries["innings"]
    ])
    .cumsum()
    - legal_delivery_flag
)

# Get first-innings totals
first_innings_totals = (
    standard_innings_analysis[
        standard_innings_analysis["innings"] == 1
    ][["match_id", "total_runs"]]
    .rename(columns={"total_runs": "first_innings_total"})
)

# Attach first-innings totals
standard_pressure_deliveries = standard_pressure_deliveries.merge(
    first_innings_totals,
    on="match_id",
    how="left",
    validate="many_to_one"
)

# Nominal second-innings target
standard_pressure_deliveries["nominal_target"] = (
    standard_pressure_deliveries["first_innings_total"] + 1
)

print("Pressure delivery records:", len(standard_pressure_deliveries))
print("Unique matches:", standard_pressure_deliveries["match_id"].nunique())

display(
    standard_pressure_deliveries[
        [
            "match_id",
            "innings",
            "overid",
            "over_ball_num",
            "striker_name",
            "bowler_name",
            "team_runs",
            "runs_before_delivery",
            "wickets_lost_before_delivery",
            "wickets_remaining",
            "legal_balls_before_delivery",
            "first_innings_total",
            "nominal_target"
        ]
    ].head(15)
)
```

```python
# Legal balls bowled before the current delivery
legal_delivery_flag = (
    standard_pressure_deliveries["is_legal_delivery"]
    .eq(True)
    .fillna(False)
    .astype(int)
)

standard_pressure_deliveries["legal_balls_before_delivery"] = (
    legal_delivery_flag
    .groupby([
        standard_pressure_deliveries["match_id"],
        standard_pressure_deliveries["innings"]
    ])
    .cumsum()
    - legal_delivery_flag
)
print(
    standard_pressure_deliveries[
        [
            "runs_before_delivery",
            "wickets_lost_before_delivery",
            "wickets_remaining",
            "legal_balls_before_delivery"
        ]
    ].head(10)
)
```

```python
# Validate pressure calculations
print("Total pressure deliveries:", len(standard_pressure_deliveries))
print("Unique matches:", standard_pressure_deliveries["match_id"].nunique())

print(
    "Negative runs before delivery:",
    (standard_pressure_deliveries["runs_before_delivery"] < 0).sum()
)

print(
    "Invalid wickets remaining:",
    (
        (standard_pressure_deliveries["wickets_remaining"] < 0) |
        (standard_pressure_deliveries["wickets_remaining"] > 10)
    ).sum()
)

print(
    "Negative legal balls:",
    (standard_pressure_deliveries["legal_balls_before_delivery"] < 0).sum()
)

# Verify first delivery of every innings
first_delivery_check = (
    standard_pressure_deliveries
    .groupby(["match_id", "innings"], as_index=False)
    .first()
)

print(
    "First deliveries with non-zero runs:",
    (first_delivery_check["runs_before_delivery"] != 0).sum()
)

print(
    "First deliveries with non-zero legal balls:",
    (first_delivery_check["legal_balls_before_delivery"] != 0).sum()
)
```

```python
standard_pressure_deliveries["wickets_pressure"] = pd.cut(
    standard_pressure_deliveries["wickets_remaining"],
    bins=[0, 3, 6, 10],
    labels=["High", "Moderate", "Lower"],
    include_lowest=True
)

display(
    standard_pressure_deliveries["wickets_pressure"]
    .value_counts()
    .sort_index()
)
```

```python
standard_pressure_deliveries["current_over"] = (
    standard_pressure_deliveries["legal_balls_before_delivery"] // 6
) + 1

standard_pressure_deliveries["is_death_over"] = (
    standard_pressure_deliveries["current_over"] >= 16
)

print(
    standard_pressure_deliveries["is_death_over"]
    .value_counts()
)
```

```python
standard_pressure_deliveries["is_chasing"] = (
    standard_pressure_deliveries["innings"] == 2
)

print(
    standard_pressure_deliveries["is_chasing"]
    .value_counts()
)
```

```python
# Get first-innings totals
first_innings_totals = (
    standard_innings_analysis[
        standard_innings_analysis["innings"] == 1
    ][["match_id", "total_runs"]]
    .rename(columns={"total_runs": "first_innings_total"})
)

# Remove previously attached columns if they exist
standard_pressure_deliveries = standard_pressure_deliveries.drop(
    columns=["first_innings_total", "nominal_target"],
    errors="ignore"
)

# Attach first-innings totals
standard_pressure_deliveries = standard_pressure_deliveries.merge(
    first_innings_totals,
    on="match_id",
    how="left",
    validate="many_to_one"
)

# Calculate nominal target
standard_pressure_deliveries["nominal_target"] = (
    standard_pressure_deliveries["first_innings_total"] + 1
)

print("Missing first-innings totals:",
      standard_pressure_deliveries["first_innings_total"].isna().sum())

print("Missing nominal targets:",
      standard_pressure_deliveries["nominal_target"].isna().sum())

display(
    standard_pressure_deliveries[
        ["match_id", "innings", "first_innings_total", "nominal_target"]
    ].head(10)
)
```

```python
# Calculate runs and balls remaining
standard_pressure_deliveries["runs_required"] = (
    standard_pressure_deliveries["nominal_target"]
    - standard_pressure_deliveries["runs_before_delivery"]
)

standard_pressure_deliveries["balls_remaining"] = (
    120 - standard_pressure_deliveries["legal_balls_before_delivery"]
)

# Valid second-innings chase deliveries
valid_chase_mask = (
    standard_pressure_deliveries["is_chasing"]
    & standard_pressure_deliveries["match_id"].ne(18532)
    & standard_pressure_deliveries["balls_remaining"].gt(0)
    & standard_pressure_deliveries["runs_required"].gt(0)
)

standard_pressure_deliveries["required_run_rate"] = float("nan")

standard_pressure_deliveries.loc[
    valid_chase_mask, "required_run_rate"
] = (
    standard_pressure_deliveries.loc[valid_chase_mask, "runs_required"]
    * 6
    / standard_pressure_deliveries.loc[valid_chase_mask, "balls_remaining"]
).round(2)

print("Valid chase deliveries:", valid_chase_mask.sum())

print(
    "Missing required run rate in valid chases:",
    standard_pressure_deliveries.loc[
        valid_chase_mask, "required_run_rate"
    ].isna().sum()
)

display(
    standard_pressure_deliveries.loc[
        valid_chase_mask,
        [
            "match_id",
            "runs_before_delivery",
            "runs_required",
            "balls_remaining",
            "required_run_rate"
        ]
    ].head(10)
)
```

```python
standard_pressure_deliveries["required_run_rate_pressure"] = (
    standard_pressure_deliveries["required_run_rate"] >= 8
)

print(
    standard_pressure_deliveries.loc[
        standard_pressure_deliveries["is_chasing"],
        "required_run_rate_pressure"
    ].value_counts(dropna=False)
)

print(
    "High required run-rate deliveries:",
    (
        standard_pressure_deliveries["required_run_rate_pressure"]
        & standard_pressure_deliveries["is_chasing"]
    ).sum()
)
```

```python
# Preserve unavailable required run rates as NA
standard_pressure_deliveries["required_run_rate_pressure"] = (
    standard_pressure_deliveries["required_run_rate"]
    .ge(8)
    .astype("boolean")
)

# Count confirmed pressure factors
standard_pressure_deliveries["batting_pressure_factor_count"] = (
    standard_pressure_deliveries["required_run_rate_pressure"]
    .fillna(False).astype(int)
    + standard_pressure_deliveries["wickets_pressure"].eq("High").astype(int)
    + standard_pressure_deliveries["is_death_over"].astype(int)
)

standard_pressure_deliveries["is_high_batting_pressure"] = (
    standard_pressure_deliveries["batting_pressure_factor_count"] >= 2
)

print("Required run-rate pressure:")
print(
    standard_pressure_deliveries["required_run_rate_pressure"]
    .value_counts(dropna=False)
)

print("\nBatting pressure factor distribution:")
print(
    standard_pressure_deliveries["batting_pressure_factor_count"]
    .value_counts()
    .sort_index()
)

print(
    "\nHigh batting pressure deliveries:",
    standard_pressure_deliveries["is_high_batting_pressure"].sum()
)
```

```python
high_batting_pressure_deliveries = standard_pressure_deliveries[
    standard_pressure_deliveries["is_high_batting_pressure"]
].copy()

high_pressure_batting = (
    high_batting_pressure_deliveries
    .groupby(
        ["match_id", "innings", "batting_team_id",
         "striker_id", "striker_name"],
        as_index=False
    )
    .agg(
        runs=("batter_runs", "sum"),
        balls_faced=("is_legal_delivery", lambda x: x.eq(True).sum()),
        fours=("is_four", "sum"),
        sixes=("is_six", "sum")
    )
)

high_pressure_batting["strike_rate"] = (
    high_pressure_batting["runs"]
    .div(high_pressure_batting["balls_faced"])
    .mul(100)
    .where(high_pressure_batting["balls_faced"] > 0, 0)
    .round(2)
)

print("High-pressure batting records:", len(high_pressure_batting))
print("Unique batters:", high_pressure_batting["striker_id"].nunique())
print("Total runs:", high_pressure_batting["runs"].sum())
print("Total balls faced:", high_pressure_batting["balls_faced"].sum())

display(high_pressure_batting.head(10))
```

```python
batter_pressure_rankings = (
    high_pressure_batting
    .groupby(
        ["striker_id", "striker_name"],
        as_index=False
    )
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),
        total_runs=("runs", "sum"),
        total_balls=("balls_faced", "sum"),
        total_fours=("fours", "sum"),
        total_sixes=("sixes", "sum")
    )
)

batter_pressure_rankings["strike_rate"] = (
    batter_pressure_rankings["total_runs"]
    / batter_pressure_rankings["total_balls"]
    * 100
).where(
    batter_pressure_rankings["total_balls"] > 0, 0
).round(2)

batter_pressure_rankings = batter_pressure_rankings.sort_values(
    ["total_runs", "strike_rate"],
    ascending=[False, False]
).reset_index(drop=True)

print("Unique batters:", len(batter_pressure_rankings))

display(batter_pressure_rankings.head(20))
```

```python
standard_batting_full = (
    standard_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        runs=("batter_runs", "sum"),
        balls_faced=("is_legal_delivery", lambda x: x.eq(True).sum()),
        fours=("is_four", "sum"),
        sixes=("is_six", "sum")
    )
)

standard_batting_full["strike_rate"] = (
    standard_batting_full["runs"]
    / standard_batting_full["balls_faced"] * 100
).where(
    standard_batting_full["balls_faced"] > 0, 0
).round(2)

print("Complete batting records:", len(standard_batting_full))
print("Unique batters:", standard_batting_full["striker_id"].nunique())
print("Unique matches:", standard_batting_full["match_id"].nunique())

display(standard_batting_full.head(10))
```

```python
high_pressure_deliveries = standard_pressure_deliveries[
    standard_pressure_deliveries["is_high_batting_pressure"]
].copy()

high_pressure_batting = (
    high_pressure_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        pressure_runs=("batter_runs", "sum"),
        pressure_balls=("is_legal_delivery", lambda x: x.eq(True).sum()),
        pressure_fours=("is_four", "sum"),
        pressure_sixes=("is_six", "sum")
    )
)

high_pressure_batting["pressure_strike_rate"] = (
    high_pressure_batting["pressure_runs"]
    / high_pressure_batting["pressure_balls"] * 100
).where(
    high_pressure_batting["pressure_balls"] > 0, 0
).round(2)

print("High-pressure batting records:", len(high_pressure_batting))
print("Unique batters:", high_pressure_batting["striker_id"].nunique())
print("Unique matches:", high_pressure_batting["match_id"].nunique())

display(high_pressure_batting.head(10))
```

```python
batter_match_analysis = standard_batting_full.merge(
    high_pressure_batting[
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "pressure_runs",
            "pressure_balls",
            "pressure_fours",
            "pressure_sixes",
            "pressure_strike_rate"
        ]
    ],
    on=[
        "match_id",
        "innings",
        "batting_team_id",
        "striker_id"
    ],
    how="left",
    validate="one_to_one"
)

pressure_columns = [
    "pressure_runs",
    "pressure_balls",
    "pressure_fours",
    "pressure_sixes"
]

batter_match_analysis[pressure_columns] = (
    batter_match_analysis[pressure_columns].fillna(0).astype(int)
)

batter_match_analysis["pressure_strike_rate"] = (
    batter_match_analysis["pressure_strike_rate"].fillna(0)
)

print("Combined batting records:", len(batter_match_analysis))
print("Unique batters:", batter_match_analysis["striker_id"].nunique())
print("Missing pressure values:")
display(batter_match_analysis[pressure_columns].isna().sum())

display(batter_match_analysis.head(10))
```

```python

# ==========================================
# 1. FULL INNINGS BATTING STATISTICS
# ==========================================

standard_batting_full = (
    standard_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        overall_runs=("batter_runs", "sum"),
        overall_balls_faced=(
            "is_batter_ball_faced",
            lambda x: x.eq(True).sum()
        ),
        overall_fours=("is_four", "sum"),
        overall_sixes=("is_six", "sum")
    )
)

standard_batting_full["overall_strike_rate"] = (
    standard_batting_full["overall_runs"]
    / standard_batting_full["overall_balls_faced"] * 100
).where(
    standard_batting_full["overall_balls_faced"] > 0, 0
).round(2)


# ==========================================
# 2. BATTING UNDER PRESSURE STATISTICS
# ==========================================

high_pressure_deliveries = standard_pressure_deliveries[
    standard_pressure_deliveries["is_high_batting_pressure"]
].copy()

batting_under_pressure = (
    high_pressure_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        under_pressure_runs=("batter_runs", "sum"),
        under_pressure_balls_faced=(
            "is_batter_ball_faced",
            lambda x: x.eq(True).sum()
        ),
        under_pressure_fours=("is_four", "sum"),
        under_pressure_sixes=("is_six", "sum")
    )
)

batting_under_pressure["under_pressure_strike_rate"] = (
    batting_under_pressure["under_pressure_runs"]
    / batting_under_pressure["under_pressure_balls_faced"] * 100
).where(
    batting_under_pressure["under_pressure_balls_faced"] > 0, 0
).round(2)


# ==========================================
# 3. MERGE OVERALL + UNDER PRESSURE DATA
# ==========================================

batter_match_analysis = standard_batting_full.merge(
    batting_under_pressure[
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "under_pressure_runs",
            "under_pressure_balls_faced",
            "under_pressure_fours",
            "under_pressure_sixes",
            "under_pressure_strike_rate"
        ]
    ],
    on=[
        "match_id",
        "innings",
        "batting_team_id",
        "striker_id"
    ],
    how="left",
    validate="one_to_one"
)


# Players without any under-pressure deliveries in an innings
under_pressure_count_columns = [
    "under_pressure_runs",
    "under_pressure_balls_faced",
    "under_pressure_fours",
    "under_pressure_sixes"
]

batter_match_analysis[under_pressure_count_columns] = (
    batter_match_analysis[under_pressure_count_columns]
    .fillna(0)
    .astype(int)
)

batter_match_analysis["under_pressure_strike_rate"] = (
    batter_match_analysis["under_pressure_strike_rate"].fillna(0)
)


# ==========================================
# 4. VALIDATION
# ==========================================

print("Overall batting records:", len(standard_batting_full))
print("Under-pressure batting records:", len(batting_under_pressure))
print("Combined batting records:", len(batter_match_analysis))

print(
    "Unique batters:",
    batter_match_analysis["striker_id"].nunique()
)

print("\nOverall runs:", batter_match_analysis["overall_runs"].sum())
print(
    "Runs under pressure:",
    batter_match_analysis["under_pressure_runs"].sum()
)

print(
    "\nOverall balls faced:",
    batter_match_analysis["overall_balls_faced"].sum()
)
print(
    "Balls faced under pressure:",
    batter_match_analysis["under_pressure_balls_faced"].sum()
)

print("\nMissing values:")
display(
    batter_match_analysis[
        [
            "overall_runs",
            "overall_balls_faced",
            "under_pressure_runs",
            "under_pressure_balls_faced",
            "overall_strike_rate",
            "under_pressure_strike_rate"
        ]
    ].isna().sum()
)

print(
    "\nDuplicate batter-match-innings records:",
    batter_match_analysis.duplicated(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id"
        ]
    ).sum()
)
```

```python

# ==========================================
# 5. PLAYER-LEVEL BATTING LEADERBOARD
# ==========================================

batter_rankings = (
    batter_match_analysis
    .groupby(
        ["striker_id", "striker_name"],
        as_index=False
    )
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),

        overall_runs=("overall_runs", "sum"),
        overall_balls_faced=("overall_balls_faced", "sum"),
        overall_fours=("overall_fours", "sum"),
        overall_sixes=("overall_sixes", "sum"),

        under_pressure_runs=("under_pressure_runs", "sum"),
        under_pressure_balls_faced=("under_pressure_balls_faced", "sum"),
        under_pressure_fours=("under_pressure_fours", "sum"),
        under_pressure_sixes=("under_pressure_sixes", "sum")
    )
)

# Overall strike rate
batter_rankings["overall_strike_rate"] = (
    batter_rankings["overall_runs"]
    / batter_rankings["overall_balls_faced"] * 100
).where(
    batter_rankings["overall_balls_faced"] > 0, 0
).round(2)

# Strike rate under pressure
batter_rankings["under_pressure_strike_rate"] = (
    batter_rankings["under_pressure_runs"]
    / batter_rankings["under_pressure_balls_faced"] * 100
).where(
    batter_rankings["under_pressure_balls_faced"] > 0, 0
).round(2)

# Qualification criteria
qualified_batter_rankings = batter_rankings[
    (batter_rankings["overall_runs"] >= 1000)
    & (batter_rankings["innings"] >= 10)
].copy()

# Rank primarily by performance under pressure
qualified_batter_rankings = (
    qualified_batter_rankings
    .sort_values(
        [
            "under_pressure_runs",
            "under_pressure_strike_rate",
            "under_pressure_balls_faced",
            "overall_runs",
            "overall_strike_rate"
        ],
        ascending=[False, False, False, False, False]
    )
    .reset_index(drop=True)
)

qualified_batter_rankings.insert(
    0,
    "rank",
    range(1, len(qualified_batter_rankings) + 1)
)

print("Total batters:", len(batter_rankings))
print("Qualified batters:", len(qualified_batter_rankings))

display(qualified_batter_rankings.head(20))
```

```python

# ==========================================
# 6. ATTACH MATCH CONTEXT
# ==========================================

match_dashboard_context = (
    matches_analysis[
        [
            "match_id",
            "is_knockout",
            "stage",
            "stage_priority",
            "match_date",
            "match_condition",
            "result_category",
            "comp_id"
        ]
    ]
    .merge(
        competitions_copy[
            [
                "comp_id",
                "comp_name",
                "match_format",
                "season"
            ]
        ],
        on="comp_id",
        how="left",
        validate="many_to_one"
    )
)

# Remove old context columns if they already exist
context_columns = [
    "is_knockout",
    "stage",
    "stage_priority",
    "match_date",
    "match_condition",
    "result_category",
    "comp_id",
    "comp_name",
    "match_format",
    "season"
]

batter_match_analysis = (
    batter_match_analysis
    .drop(columns=context_columns, errors="ignore")
    .merge(
        match_dashboard_context,
        on="match_id",
        how="left",
        validate="many_to_one"
    )
)

# Four-level innings priority
conditions = [
    batter_match_analysis["is_knockout"]
    & batter_match_analysis["innings"].eq(2),

    batter_match_analysis["is_knockout"]
    & batter_match_analysis["innings"].eq(1),

    ~batter_match_analysis["is_knockout"]
    & batter_match_analysis["innings"].eq(2),

    ~batter_match_analysis["is_knockout"]
    & batter_match_analysis["innings"].eq(1)
]

batter_match_analysis["analysis_priority"] = np.select(
    conditions,
    [1, 2, 3, 4],
    default=5
)

batter_match_analysis = batter_match_analysis.sort_values(
    ["analysis_priority", "stage_priority", "match_date"],
    ascending=[True, True, False]
).reset_index(drop=True)

# Validation
print("Total records:", len(batter_match_analysis))
print("Unique matches:", batter_match_analysis["match_id"].nunique())

print("\nKnockout distribution:")
display(
    batter_match_analysis["is_knockout"].value_counts(dropna=False)
)

print("\nAnalysis priority distribution:")
display(
    batter_match_analysis["analysis_priority"].value_counts().sort_index()
)

print("\nMissing match context:")
display(
    batter_match_analysis[
        [
            "is_knockout",
            "stage",
            "match_date",
            "comp_name",
            "match_format"
        ]
    ].isna().sum()
)
```

```python

# ==========================================
# 8. BOWLING UNDER PRESSURE - DELIVERY LEVEL
# ==========================================

high_pressure_bowling_deliveries = (
    standard_pressure_deliveries[
        standard_pressure_deliveries["is_high_batting_pressure"]
    ].copy()
)

# Validate required columns
required_columns = [
    "match_id",
    "innings",
    "bowling_team_id",
    "bowler_id",
    "bowler_name",
    "bowler_runs_conceded",
    "is_legal_delivery",
    "is_bowler_credited_wicket",
    "is_high_batting_pressure"
]

print("Missing required columns:")
print(
    set(required_columns)
    - set(high_pressure_bowling_deliveries.columns)
)

print("\nTotal high-pressure deliveries:",
      len(high_pressure_bowling_deliveries))

print("Unique matches:",
      high_pressure_bowling_deliveries["match_id"].nunique())

print("Unique bowlers:",
      high_pressure_bowling_deliveries["bowler_id"].nunique())

print("\nBowling team and bowler missing values:")
display(
    high_pressure_bowling_deliveries[
        ["bowling_team_id", "bowler_id", "bowler_name"]
    ].isna().sum()
)

print("\nUnder-pressure delivery validation:")
print(
    "Non-pressure deliveries:",
    (~high_pressure_bowling_deliveries["is_high_batting_pressure"]).sum()
)
```

```python

# ==========================================
# 9. BOWLING UNDER PRESSURE - MATCH LEVEL
# ==========================================

bowling_under_pressure = (
    high_pressure_bowling_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id",
            "bowler_name"
        ],
        as_index=False
    )
    .agg(
        under_pressure_legal_balls=(
            "is_legal_delivery",
            lambda x: x.eq(True).sum()
        ),
        under_pressure_runs_conceded=(
            "bowler_runs_conceded",
            "sum"
        ),
        under_pressure_wickets=(
            "is_bowler_credited_wicket",
            "sum"
        )
    )
)

# Economy under pressure
bowling_under_pressure["under_pressure_economy"] = (
    bowling_under_pressure["under_pressure_runs_conceded"] * 6
    / bowling_under_pressure["under_pressure_legal_balls"]
).where(
    bowling_under_pressure["under_pressure_legal_balls"] > 0
).round(2)

# Bowling average under pressure
bowling_under_pressure["under_pressure_bowling_average"] = (
    bowling_under_pressure["under_pressure_runs_conceded"]
    / bowling_under_pressure["under_pressure_wickets"]
).where(
    bowling_under_pressure["under_pressure_wickets"] > 0
).round(2)

# Bowling strike rate under pressure
bowling_under_pressure["under_pressure_bowling_strike_rate"] = (
    bowling_under_pressure["under_pressure_legal_balls"]
    / bowling_under_pressure["under_pressure_wickets"]
).where(
    bowling_under_pressure["under_pressure_wickets"] > 0
).round(2)

# Validation
print("Bowling-under-pressure records:", len(bowling_under_pressure))
print("Unique bowlers:", bowling_under_pressure["bowler_id"].nunique())
print("Unique matches:", bowling_under_pressure["match_id"].nunique())

print("\nMissing values:")
display(bowling_under_pressure.isna().sum())

print("\nDuplicate bowler-match-innings records:",
      bowling_under_pressure.duplicated(
          ["match_id", "innings", "bowler_id"]
      ).sum())

print("\nTotal runs conceded under pressure:",
      bowling_under_pressure["under_pressure_runs_conceded"].sum())

print("Total wickets under pressure:",
      bowling_under_pressure["under_pressure_wickets"].sum())

print("\nSample records:")
display(
    bowling_under_pressure.head(10)
)
```

```python
print("Delivery-level vs aggregated validation")

print(
    "Runs conceded match:",
    high_pressure_bowling_deliveries["bowler_runs_conceded"].sum()
    == bowling_under_pressure["under_pressure_runs_conceded"].sum()
)

print(
    "Wickets match:",
    high_pressure_bowling_deliveries["is_bowler_credited_wicket"].sum()
    == bowling_under_pressure["under_pressure_wickets"].sum()
)

print(
    "Legal balls match:",
    high_pressure_bowling_deliveries["is_legal_delivery"].eq(True).sum()
    == bowling_under_pressure["under_pressure_legal_balls"].sum()
)

print(
    "Duplicate records:",
    bowling_under_pressure.duplicated(
        ["match_id", "innings", "bowler_id"]
    ).sum()
)
```

```python

# ==========================================
# 10. ATTACH MATCH CONTEXT TO
#     BOWLING UNDER PRESSURE
# ==========================================

bowling_under_pressure = (
    bowling_under_pressure
    .drop(
        columns=[
            "is_knockout", "stage", "stage_priority",
            "match_date", "match_condition", "result_category",
            "comp_id", "comp_name", "match_format", "season"
        ],
        errors="ignore"
    )
    .merge(
        match_dashboard_context,
        on="match_id",
        how="left",
        validate="many_to_one"
    )
)

conditions = [
    bowling_under_pressure["is_knockout"]
    & bowling_under_pressure["innings"].eq(2),

    bowling_under_pressure["is_knockout"]
    & bowling_under_pressure["innings"].eq(1),

    ~bowling_under_pressure["is_knockout"]
    & bowling_under_pressure["innings"].eq(2),

    ~bowling_under_pressure["is_knockout"]
    & bowling_under_pressure["innings"].eq(1)
]

bowling_under_pressure["analysis_priority"] = np.select(
    conditions,
    [1, 2, 3, 4],
    default=5
)

bowling_under_pressure = bowling_under_pressure.sort_values(
    ["analysis_priority", "stage_priority", "match_date"],
    ascending=[True, True, False]
).reset_index(drop=True)

# Validation
print("Total records:", len(bowling_under_pressure))
print("Unique matches:", bowling_under_pressure["match_id"].nunique())
print("Unique bowlers:", bowling_under_pressure["bowler_id"].nunique())

print("\nPriority distribution:")
display(
    bowling_under_pressure["analysis_priority"].value_counts().sort_index()
)

print("\nMissing match context:")
display(
    bowling_under_pressure[
        ["is_knockout", "stage", "match_date", "comp_name", "match_format"]
    ].isna().sum()
)
```

```python

# ============================================================
# 11. BOWLING UNDER-PRESSURE CLASSIFICATION
# ============================================================

# Create an independent delivery-level dataset
# Do not modify standard_pressure_deliveries.

bowling_situation_deliveries = (
    standard_pressure_deliveries
    .drop(
        columns=[
            "is_knockout",
            "match_date"
        ],
        errors="ignore"
    )
    .merge(
        match_dashboard_context[
            [
                "match_id",
                "is_knockout",
                "match_date"
            ]
        ],
        on="match_id",
        how="left",
        validate="many_to_one"
    )
    .copy(deep=True)
)


# ============================================================
# 1. BOWLING SITUATION FACTORS
# ============================================================

# Factor 1: Defending a low first-innings total (<120)
# Applies only to the team bowling in the second innings.

bowling_situation_deliveries["is_defending_low_total"] = (
    bowling_situation_deliveries["innings"].eq(2)
    & bowling_situation_deliveries["first_innings_total"].lt(120)
)


# Factor 2: Opposition has 7–10 wickets remaining

bowling_situation_deliveries["opposition_has_wickets_in_hand"] = (
    bowling_situation_deliveries["wickets_remaining"].between(7, 10)
)


# Factor 3: Death overs (16–20)

bowling_situation_deliveries["is_bowling_death_over"] = (
    bowling_situation_deliveries["current_over"].between(16, 20)
)


# ============================================================
# 2. COUNT BOWLING SITUATION FACTORS
# ============================================================

bowling_factor_columns = [
    "is_defending_low_total",
    "opposition_has_wickets_in_hand",
    "is_bowling_death_over"
]

bowling_situation_deliveries["bowling_factor_count"] = (
    bowling_situation_deliveries[bowling_factor_columns]
    .sum(axis=1)
)


# ============================================================
# 3. CLASSIFY BOWLING SITUATION
# ============================================================

bowling_situation_deliveries["bowling_situation"] = "Normal"

bowling_situation_deliveries.loc[
    bowling_situation_deliveries["bowling_factor_count"] == 1,
    "bowling_situation"
] = "Moderate"

bowling_situation_deliveries.loc[
    bowling_situation_deliveries["bowling_factor_count"] == 2,
    "bowling_situation"
] = "High"

bowling_situation_deliveries.loc[
    bowling_situation_deliveries["bowling_factor_count"] == 3,
    "bowling_situation"
] = "Very High"


# ============================================================
# 4. MATCH IMPORTANCE
# ============================================================

# Knockout matches receive higher priority than ordinary matches.

bowling_situation_deliveries["match_priority"] = np.where(
    bowling_situation_deliveries["is_knockout"],
    1,  # Knockout
    2   # Ordinary
)

bowling_situation_deliveries["match_importance"] = np.where(
    bowling_situation_deliveries["is_knockout"],
    "Knockout",
    "Ordinary"
)


# ============================================================
# 5. BOWLING SITUATION PRIORITY
# ============================================================

situation_priority = {
    "Very High": 1,
    "High": 2,
    "Moderate": 3,
    "Normal": 4
}

bowling_situation_deliveries["bowling_situation_priority"] = (
    bowling_situation_deliveries["bowling_situation"]
    .map(situation_priority)
)


# ============================================================
# 6. COMBINED ANALYSIS PRIORITY
# ============================================================

# 1–4: Knockout matches
# 5–8: Ordinary matches

bowling_situation_deliveries["analysis_priority"] = (
    (bowling_situation_deliveries["match_priority"] - 1) * 4
    + bowling_situation_deliveries["bowling_situation_priority"]
)


# ============================================================
# 7. SORT DELIVERIES
# ============================================================

bowling_situation_deliveries = (
    bowling_situation_deliveries
    .sort_values(
        [
            "analysis_priority",
            "match_date",
            "match_id"
        ],
        ascending=[
            True,
            False,
            True
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# 8. VALIDATION
# ============================================================

print("=" * 55)
print("BOWLING UNDER-PRESSURE CLASSIFICATION")
print("=" * 55)

print("\nTotal deliveries:", len(bowling_situation_deliveries))

print(
    "Unique matches:",
    bowling_situation_deliveries["match_id"].nunique()
)

print(
    "Unique bowlers:",
    bowling_situation_deliveries["bowler_id"].nunique()
)


print("\n1. MATCH IMPORTANCE")
display(
    bowling_situation_deliveries[
        "match_importance"
    ].value_counts(dropna=False)
)


print("\n2. BOWLING SITUATION")
display(
    bowling_situation_deliveries[
        "bowling_situation"
    ].value_counts()
)


print("\n3. BOWLING FACTOR COUNT")
display(
    bowling_situation_deliveries[
        "bowling_factor_count"
    ].value_counts().sort_index()
)


print("\n4. COMBINED ANALYSIS PRIORITY")
display(
    bowling_situation_deliveries[
        "analysis_priority"
    ].value_counts().sort_index()
)


print("\n5. MATCH IMPORTANCE VS BOWLING SITUATION")
display(
    pd.crosstab(
        bowling_situation_deliveries["match_importance"],
        bowling_situation_deliveries["bowling_situation"]
    )
)


print("\n6. MISSING VALUES")
display(
    bowling_situation_deliveries[
        [
            "is_knockout",
            "match_date",
            "first_innings_total",
            "bowling_factor_count",
            "bowling_situation",
            "analysis_priority"
        ]
    ].isna().sum()
)


print("\n7. DUPLICATE DELIVERY CHECK")

print(
    "Duplicate delivery keys:",
    bowling_situation_deliveries.duplicated(
        [
            "match_id",
            "innings",
            "innings_ball_num"
        ]
    ).sum()
)


print("\n8. ORIGINAL DATA CHECK")

print(
    "Original standard pressure deliveries unchanged:",
    len(standard_pressure_deliveries)
    == len(bowling_situation_deliveries)
)

print("\nClassification completed.")
```

```python

# ============================================================
# 12. BOWLING PERFORMANCE UNDER PRESSURE
# ============================================================

# 1. Select difficult bowling situations

bowling_under_pressure_deliveries = (
    bowling_situation_deliveries[
        bowling_situation_deliveries["bowling_situation"].isin(
            ["Moderate", "High", "Very High"]
        )
    ].copy(deep=True)
)


# ============================================================
# 2. ENSURE MATCH AND SITUATION PRIORITIES EXIST
# ============================================================

# Recreate match importance fields to avoid missing-column errors.

bowling_under_pressure_deliveries["match_priority"] = np.where(
    bowling_under_pressure_deliveries["is_knockout"],
    1,
    2
)

bowling_under_pressure_deliveries["match_importance"] = np.where(
    bowling_under_pressure_deliveries["is_knockout"],
    "Knockout",
    "Ordinary"
)

situation_priority = {
    "Very High": 1,
    "High": 2,
    "Moderate": 3,
    "Normal": 4
}

bowling_under_pressure_deliveries["bowling_situation_priority"] = (
    bowling_under_pressure_deliveries["bowling_situation"]
    .map(situation_priority)
)

bowling_under_pressure_deliveries["analysis_priority"] = (
    (bowling_under_pressure_deliveries["match_priority"] - 1) * 4
    + bowling_under_pressure_deliveries["bowling_situation_priority"]
)


# ============================================================
# 3. AGGREGATE BOWLING PERFORMANCE
# ============================================================

bowling_under_pressure = (
    bowling_under_pressure_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id",
            "bowler_name",
            "match_importance",
            "match_priority",
            "bowling_situation",
            "bowling_situation_priority",
            "analysis_priority"
        ],
        as_index=False
    )
    .agg(
        under_pressure_legal_balls=(
            "is_legal_delivery",
            lambda x: x.eq(True).sum()
        ),
        under_pressure_runs_conceded=(
            "bowler_runs_conceded",
            "sum"
        ),
        under_pressure_wickets=(
            "is_bowler_credited_wicket",
            "sum"
        )
    )
)


# ============================================================
# 4. CALCULATE BOWLING METRICS
# ============================================================

bowling_under_pressure["under_pressure_overs"] = (
    bowling_under_pressure["under_pressure_legal_balls"] // 6
    + (
        bowling_under_pressure["under_pressure_legal_balls"] % 6
    ) / 10
)

bowling_under_pressure["under_pressure_economy"] = (
    bowling_under_pressure["under_pressure_runs_conceded"] * 6
    / bowling_under_pressure["under_pressure_legal_balls"]
).where(
    bowling_under_pressure["under_pressure_legal_balls"] > 0
).round(2)

bowling_under_pressure["under_pressure_bowling_average"] = (
    bowling_under_pressure["under_pressure_runs_conceded"]
    / bowling_under_pressure["under_pressure_wickets"]
).where(
    bowling_under_pressure["under_pressure_wickets"] > 0
).round(2)

bowling_under_pressure["under_pressure_bowling_strike_rate"] = (
    bowling_under_pressure["under_pressure_legal_balls"]
    / bowling_under_pressure["under_pressure_wickets"]
).where(
    bowling_under_pressure["under_pressure_wickets"] > 0
).round(2)


# ============================================================
# 5. SORT BY PRIORITY
# ============================================================

bowling_under_pressure = (
    bowling_under_pressure
    .sort_values(
        [
            "analysis_priority",
            "match_id",
            "innings",
            "bowler_id"
        ],
        ascending=[True, False, True, True]
    )
    .reset_index(drop=True)
)


# ============================================================
# 6. VALIDATION
# ============================================================

print("=" * 60)
print("BOWLING PERFORMANCE UNDER PRESSURE")
print("=" * 60)

print("\nTotal records:", len(bowling_under_pressure))

print(
    "Unique bowlers:",
    bowling_under_pressure["bowler_id"].nunique()
)

print(
    "Unique matches:",
    bowling_under_pressure["match_id"].nunique()
)

print(
    "Total runs conceded:",
    bowling_under_pressure["under_pressure_runs_conceded"].sum()
)

print(
    "Total legal balls:",
    bowling_under_pressure["under_pressure_legal_balls"].sum()
)

print(
    "Total bowler-credited wickets:",
    bowling_under_pressure["under_pressure_wickets"].sum()
)


print("\n1. Bowling situation distribution:")
display(
    bowling_under_pressure["bowling_situation"].value_counts()
)


print("\n2. Match importance distribution:")
display(
    bowling_under_pressure["match_importance"].value_counts()
)


print("\n3. Analysis priority distribution:")
display(
    bowling_under_pressure["analysis_priority"].value_counts().sort_index()
)


print("\n4. Missing values:")
display(
    bowling_under_pressure[
        [
            "bowler_id",
            "under_pressure_legal_balls",
            "under_pressure_runs_conceded",
            "under_pressure_wickets",
            "under_pressure_economy",
            "under_pressure_bowling_average",
            "under_pressure_bowling_strike_rate"
        ]
    ].isna().sum()
)


print("\n5. Duplicate aggregation keys:")

print(
    bowling_under_pressure.duplicated(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id",
            "bowling_situation"
        ]
    ).sum()
)


print("\n6. Delivery-level vs aggregated validation:")

print(
    "Runs conceded match:",
    bowling_under_pressure_deliveries["bowler_runs_conceded"].sum()
    == bowling_under_pressure["under_pressure_runs_conceded"].sum()
)

print(
    "Wickets match:",
    bowling_under_pressure_deliveries["is_bowler_credited_wicket"].sum()
    == bowling_under_pressure["under_pressure_wickets"].sum()
)

print(
    "Legal balls match:",
    bowling_under_pressure_deliveries["is_legal_delivery"].eq(True).sum()
    == bowling_under_pressure["under_pressure_legal_balls"].sum()
)

print("\nBowling aggregation completed.")
```

```python

# ============================================================
# 13. COMBINE OVERALL AND UNDER-PRESSURE BOWLING
# ============================================================

# 1. Combine all difficult situations into one
#    bowler-match-innings record.

bowling_under_pressure_match = (
    bowling_under_pressure
    .groupby(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id"
        ],
        as_index=False
    )
    .agg(
        under_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        under_pressure_runs_conceded=(
            "under_pressure_runs_conceded", "sum"
        ),
        under_pressure_wickets=(
            "under_pressure_wickets", "sum"
        )
    )
)


# 2. Attach difficult-situation statistics
#    to the complete bowling match dataset.

pressure_columns = [
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets"
]

bowler_match_analysis = bowler_match_analysis.drop(
    columns=[
        "under_pressure_legal_balls",
        "under_pressure_runs_conceded",
        "under_pressure_wickets",
        "under_pressure_overs",
        "under_pressure_economy",
        "under_pressure_bowling_average",
        "under_pressure_bowling_strike_rate"
    ],
    errors="ignore"
)

bowler_match_analysis = bowler_match_analysis.merge(
    bowling_under_pressure_match,
    on=[
        "match_id",
        "innings",
        "bowling_team_id",
        "bowler_id"
    ],
    how="left",
    validate="one_to_one"
)


# 3. Bowlers without difficult-situation deliveries
#    receive zero contribution in those situations.

bowler_match_analysis[pressure_columns] = (
    bowler_match_analysis[pressure_columns]
    .fillna(0)
    .astype(int)
)


# 4. Calculate under-pressure bowling metrics.

bowler_match_analysis["under_pressure_overs"] = (
    bowler_match_analysis["under_pressure_legal_balls"] // 6
    + (
        bowler_match_analysis["under_pressure_legal_balls"] % 6
    ) / 10
)

bowler_match_analysis["under_pressure_economy"] = (
    bowler_match_analysis["under_pressure_runs_conceded"] * 6
    / bowler_match_analysis["under_pressure_legal_balls"]
).where(
    bowler_match_analysis["under_pressure_legal_balls"] > 0
).round(2)

bowler_match_analysis["under_pressure_bowling_average"] = (
    bowler_match_analysis["under_pressure_runs_conceded"]
    / bowler_match_analysis["under_pressure_wickets"]
).where(
    bowler_match_analysis["under_pressure_wickets"] > 0
).round(2)

bowler_match_analysis["under_pressure_bowling_strike_rate"] = (
    bowler_match_analysis["under_pressure_legal_balls"]
    / bowler_match_analysis["under_pressure_wickets"]
).where(
    bowler_match_analysis["under_pressure_wickets"] > 0
).round(2)


# ============================================================
# 5. VALIDATION
# ============================================================

print("Overall bowling records:", len(bowler_match_analysis))

print(
    "Unique bowlers:",
    bowler_match_analysis["bowler_id"].nunique()
)

print(
    "Unique matches:",
    bowler_match_analysis["match_id"].nunique()
)

print(
    "Overall bowling runs:",
    bowler_match_analysis["total_runs_conceded"].sum()
)

print(
    "Under-pressure bowling runs:",
    bowler_match_analysis["under_pressure_runs_conceded"].sum()
)

print(
    "Overall wickets:",
    bowler_match_analysis["wickets"].sum()
)

print(
    "Under-pressure wickets:",
    bowler_match_analysis["under_pressure_wickets"].sum()
)


print("\nMissing pressure contribution values:")
display(
    bowler_match_analysis[pressure_columns].isna().sum()
)


print("\nDuplicate bowler-match-innings records:")

print(
    bowler_match_analysis.duplicated(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id"
        ]
    ).sum()
)


print("\nPressure runs reconciliation:")

print(
    bowler_match_analysis["under_pressure_runs_conceded"].sum()
    == bowling_under_pressure["under_pressure_runs_conceded"].sum()
)

print("\nPressure wickets reconciliation:")

print(
    bowler_match_analysis["under_pressure_wickets"].sum()
    == bowling_under_pressure["under_pressure_wickets"].sum()
)

print("\nPressure legal balls reconciliation:")

print(
    bowler_match_analysis["under_pressure_legal_balls"].sum()
    == bowling_under_pressure["under_pressure_legal_balls"].sum()
)
```

```python

# ============================================================
# 14. PLAYER-LEVEL BOWLING AGGREGATION
# ============================================================

bowler_data = bowler_match_analysis.copy(deep=True)

# Separate knockout and ordinary under-pressure contributions.

bowler_data["knockout_under_pressure_runs"] = np.where(
    bowler_data["is_knockout"],
    bowler_data["under_pressure_runs_conceded"],
    0
)

bowler_data["knockout_under_pressure_balls"] = np.where(
    bowler_data["is_knockout"],
    bowler_data["under_pressure_legal_balls"],
    0
)

bowler_data["knockout_under_pressure_wickets"] = np.where(
    bowler_data["is_knockout"],
    bowler_data["under_pressure_wickets"],
    0
)

bowler_data["ordinary_under_pressure_runs"] = np.where(
    ~bowler_data["is_knockout"],
    bowler_data["under_pressure_runs_conceded"],
    0
)

bowler_data["ordinary_under_pressure_balls"] = np.where(
    ~bowler_data["is_knockout"],
    bowler_data["under_pressure_legal_balls"],
    0
)

bowler_data["ordinary_under_pressure_wickets"] = np.where(
    ~bowler_data["is_knockout"],
    bowler_data["under_pressure_wickets"],
    0
)


# Aggregate by player.

bowler_rankings = (
    bowler_data
    .groupby(
        ["bowler_id", "bowler_name"],
        as_index=False
    )
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),

        overall_legal_balls=("legal_balls", "sum"),
        overall_runs_conceded=("total_runs_conceded", "sum"),
        overall_wickets=("wickets", "sum"),

        under_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        under_pressure_runs_conceded=(
            "under_pressure_runs_conceded", "sum"
        ),
        under_pressure_wickets=(
            "under_pressure_wickets", "sum"
        ),

        knockout_under_pressure_balls=(
            "knockout_under_pressure_balls", "sum"
        ),
        knockout_under_pressure_runs=(
            "knockout_under_pressure_runs", "sum"
        ),
        knockout_under_pressure_wickets=(
            "knockout_under_pressure_wickets", "sum"
        ),

        ordinary_under_pressure_balls=(
            "ordinary_under_pressure_balls", "sum"
        ),
        ordinary_under_pressure_runs=(
            "ordinary_under_pressure_runs", "sum"
        ),
        ordinary_under_pressure_wickets=(
            "ordinary_under_pressure_wickets", "sum"
        )
    )
)


# Calculate overall bowling metrics.

bowler_rankings["overall_economy"] = (
    bowler_rankings["overall_runs_conceded"] * 6
    / bowler_rankings["overall_legal_balls"]
).where(
    bowler_rankings["overall_legal_balls"] > 0
).round(2)

bowler_rankings["overall_bowling_average"] = (
    bowler_rankings["overall_runs_conceded"]
    / bowler_rankings["overall_wickets"]
).where(
    bowler_rankings["overall_wickets"] > 0
).round(2)

bowler_rankings["overall_bowling_strike_rate"] = (
    bowler_rankings["overall_legal_balls"]
    / bowler_rankings["overall_wickets"]
).where(
    bowler_rankings["overall_wickets"] > 0
).round(2)


# Calculate combined under-pressure bowling metrics.

bowler_rankings["under_pressure_economy"] = (
    bowler_rankings["under_pressure_runs_conceded"] * 6
    / bowler_rankings["under_pressure_legal_balls"]
).where(
    bowler_rankings["under_pressure_legal_balls"] > 0
).round(2)

bowler_rankings["under_pressure_bowling_average"] = (
    bowler_rankings["under_pressure_runs_conceded"]
    / bowler_rankings["under_pressure_wickets"]
).where(
    bowler_rankings["under_pressure_wickets"] > 0
).round(2)

bowler_rankings["under_pressure_bowling_strike_rate"] = (
    bowler_rankings["under_pressure_legal_balls"]
    / bowler_rankings["under_pressure_wickets"]
).where(
    bowler_rankings["under_pressure_wickets"] > 0
).round(2)


# Calculate knockout under-pressure metrics.

bowler_rankings["knockout_under_pressure_economy"] = (
    bowler_rankings["knockout_under_pressure_runs"] * 6
    / bowler_rankings["knockout_under_pressure_balls"]
).where(
    bowler_rankings["knockout_under_pressure_balls"] > 0
).round(2)


# Calculate ordinary under-pressure metrics.

bowler_rankings["ordinary_under_pressure_economy"] = (
    bowler_rankings["ordinary_under_pressure_runs"] * 6
    / bowler_rankings["ordinary_under_pressure_balls"]
).where(
    bowler_rankings["ordinary_under_pressure_balls"] > 0
).round(2)


# Preliminary view: most overall wickets first.
# This is NOT the final ranking yet.

bowler_rankings = (
    bowler_rankings
    .sort_values(
        ["overall_wickets", "overall_economy"],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)


# ============================================================
# VALIDATION
# ============================================================

print("Total bowlers:", len(bowler_rankings))

print(
    "Total overall runs:",
    bowler_rankings["overall_runs_conceded"].sum()
)

print(
    "Total overall wickets:",
    bowler_rankings["overall_wickets"].sum()
)

print(
    "Total under-pressure runs:",
    bowler_rankings["under_pressure_runs_conceded"].sum()
)

print(
    "Total under-pressure wickets:",
    bowler_rankings["under_pressure_wickets"].sum()
)

print(
    "Knockout under-pressure wickets:",
    bowler_rankings["knockout_under_pressure_wickets"].sum()
)

print(
    "Ordinary under-pressure wickets:",
    bowler_rankings["ordinary_under_pressure_wickets"].sum()
)

print("\nMissing values in key metrics:")
display(
    bowler_rankings[
        [
            "overall_economy",
            "overall_bowling_average",
            "overall_bowling_strike_rate",
            "under_pressure_economy",
            "under_pressure_bowling_average",
            "under_pressure_bowling_strike_rate"
        ]
    ].isna().sum()
)

print("\nTop 20 bowlers by overall wickets:")
display(bowler_rankings.head(20))
```

```python

# ============================================================
# 15. QUALIFIED BOWLER RANKINGS
# ============================================================

qualified_bowler_rankings = bowler_rankings[
    (bowler_rankings["innings"] >= 10)
    & (bowler_rankings["overall_legal_balls"] >= 150)
].copy()


# Sort by match importance and bowling performance.

qualified_bowler_rankings = (
    qualified_bowler_rankings
    .sort_values(
        [
            "knockout_under_pressure_wickets",
            "knockout_under_pressure_economy",
            "knockout_under_pressure_balls",
            "knockout_under_pressure_wickets",
            "ordinary_under_pressure_wickets",
            "ordinary_under_pressure_economy",
            "ordinary_under_pressure_balls",
            "ordinary_under_pressure_wickets",
            "overall_wickets",
            "overall_economy",
            "overall_bowling_strike_rate"
        ],
        ascending=[
            False,
            True,
            False,
            False,
            False,
            True,
            False,
            False,
            False,
            True,
            True
        ],
        na_position="last"
    )
    .reset_index(drop=True)
)

qualified_bowler_rankings.insert(
    0,
    "rank",
    range(1, len(qualified_bowler_rankings) + 1)
)


# ============================================================
# VALIDATION
# ============================================================

print("Total bowlers:", len(bowler_rankings))
print("Qualified bowlers:", len(qualified_bowler_rankings))

print(
    "Minimum innings:",
    qualified_bowler_rankings["innings"].min()
)

print(
    "Minimum overall legal balls:",
    qualified_bowler_rankings["overall_legal_balls"].min()
)

print(
    "Minimum overs:",
    qualified_bowler_rankings["overall_legal_balls"].min() / 6
)

print("\nTop 20 qualified bowlers:")
display(qualified_bowler_rankings.head(20))
```

```python

# ============================================================
# 15. FINAL QUALIFIED BOWLER RANKINGS
# ============================================================

qualified_bowler_rankings = bowler_rankings[
    (bowler_rankings["innings"] >= 10)
    & (bowler_rankings["overall_legal_balls"] >= 150)
].copy()

# Calculate knockout bowling strike rate.

qualified_bowler_rankings["knockout_under_pressure_strike_rate"] = (
    qualified_bowler_rankings["knockout_under_pressure_balls"]
    / qualified_bowler_rankings["knockout_under_pressure_wickets"]
).where(
    qualified_bowler_rankings["knockout_under_pressure_wickets"] > 0
).round(2)

# Calculate ordinary bowling strike rate.

qualified_bowler_rankings["ordinary_under_pressure_strike_rate"] = (
    qualified_bowler_rankings["ordinary_under_pressure_balls"]
    / qualified_bowler_rankings["ordinary_under_pressure_wickets"]
).where(
    qualified_bowler_rankings["ordinary_under_pressure_wickets"] > 0
).round(2)

# Final ranking hierarchy.

ranking_columns = [
    "knockout_under_pressure_wickets",
    "knockout_under_pressure_economy",
    "knockout_under_pressure_strike_rate",
    "ordinary_under_pressure_wickets",
    "ordinary_under_pressure_economy",
    "ordinary_under_pressure_strike_rate",
    "overall_wickets",
    "overall_economy",
    "overall_bowling_strike_rate"
]

ascending_order = [
    False, True, True,
    False, True, True,
    False, True, True
]

qualified_bowler_rankings = (
    qualified_bowler_rankings
    .sort_values(
        ranking_columns,
        ascending=ascending_order,
        na_position="last"
    )
    .reset_index(drop=True)
)

qualified_bowler_rankings.insert(
    0, "rank", range(1, len(qualified_bowler_rankings) + 1)
)

print("Qualified bowlers:", len(qualified_bowler_rankings))

display(qualified_bowler_rankings.head(20))
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 1
# Identify all revised matches
# ============================================================

revised_matches = matches_analysis[
    matches_analysis["has_revised_conditions"] == True
].copy()

revised_matches = (
    revised_matches
    .merge(
        competitions_copy[
            ["comp_id", "comp_name", "match_format", "season"]
        ],
        on="comp_id",
        how="left",
        validate="many_to_one"
    )
    .sort_values(["match_date", "match_id"])
    .reset_index(drop=True)
)

print("Total revised matches:", len(revised_matches))
print("Unique match IDs:", revised_matches["match_id"].nunique())

display(
    revised_matches[
        [
            "match_id",
            "match_date",
            "comp_name",
            "match_format",
            "season",
            "home_team",
            "away_team",
            "first_inn_revised_overs",
            "second_inn_revised_overs",
            "revised_target",
            "result",
            "result_category",
            "has_super_over"
        ]
    ]
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 2
# Validate innings coverage
# ============================================================

revised_regular_innings = innings_analysis[
    innings_analysis["match_id"].isin(revised_matches["match_id"])
    & innings_analysis["innings"].isin([1, 2])
].copy()

revised_innings_summary = (
    revised_regular_innings
    .groupby("match_id")
    .agg(
        innings_count=("innings", "nunique"),
        innings_present=(
            "innings",
            lambda x: ", ".join(
                map(str, sorted(x.unique()))
            )
        ),
        recorded_runs=("total_runs", "sum"),
        recorded_wickets=("wickets", "sum")
    )
    .reset_index()
)

revised_innings_summary = (
    revised_matches[
        [
            "match_id",
            "match_date",
            "comp_name",
            "result_category"
        ]
    ]
    .merge(
        revised_innings_summary,
        on="match_id",
        how="left",
        validate="one_to_one"
    )
)

revised_innings_summary["innings_count"] = (
    revised_innings_summary["innings_count"].fillna(0).astype(int)
)

revised_innings_summary["innings_present"] = (
    revised_innings_summary["innings_present"].fillna("None")
)

print("Revised matches:", len(revised_matches))
print("Matches with innings data:",
      revised_innings_summary["innings_count"].gt(0).sum())
print("Matches with both innings:",
      revised_innings_summary["innings_count"].eq(2).sum())
print("Matches with one innings:",
      revised_innings_summary["innings_count"].eq(1).sum())
print("Matches without innings:",
      revised_innings_summary["innings_count"].eq(0).sum())

print("\nInnings coverage distribution:")
display(
    revised_innings_summary["innings_count"]
    .value_counts()
    .sort_index()
)

print("\nMatch-wise innings coverage:")
display(revised_innings_summary)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 3 (CORRECTED)
# Inspect innings-level performance and ball limits
# ============================================================

# Count delivery records by match and innings
revised_delivery_counts = (
    ball_by_ball_analysis[
        ball_by_ball_analysis["match_id"].isin(revised_matches["match_id"])
        & ball_by_ball_analysis["innings"].isin([1, 2])
    ]
    .groupby(["match_id", "innings"])
    .agg(
        delivery_records=("match_id", "size"),
        legal_balls_known=(
            "is_legal_delivery",
            lambda x: int(x.eq(True).sum())
        ),
        unknown_legality=(
            "is_legal_delivery",
            lambda x: int(x.isna().sum())
        )
    )
    .reset_index()
)

# Add only metadata columns not already present
metadata_columns = [
    "match_date",
    "comp_name",
    "first_inn_revised_overs",
    "second_inn_revised_overs",
    "revised_target",
    "result_category"
]

missing_metadata = [
    col for col in metadata_columns
    if col not in revised_regular_innings.columns
]

revised_innings_detail = revised_regular_innings.merge(
    revised_matches[["match_id"] + missing_metadata],
    on="match_id",
    how="left",
    validate="many_to_one"
)

# Add delivery counts
revised_innings_detail = revised_innings_detail.merge(
    revised_delivery_counts,
    on=["match_id", "innings"],
    how="left",
    validate="one_to_one"
)

# Select applicable revised ball limit
revised_innings_detail["revised_balls_limit"] = (
    revised_innings_detail["first_inn_revised_overs"]
)

second_innings_mask = revised_innings_detail["innings"] == 2

revised_innings_detail.loc[
    second_innings_mask, "revised_balls_limit"
] = revised_innings_detail.loc[
    second_innings_mask, "second_inn_revised_overs"
]

# Sort innings
revised_innings_detail = (
    revised_innings_detail
    .sort_values(["match_date", "match_id", "innings"])
    .reset_index(drop=True)
)

print("Total revised innings records:", len(revised_innings_detail))
print(
    "Innings with a recorded revised ball limit:",
    revised_innings_detail["revised_balls_limit"].notna().sum()
)
print(
    "Innings without a recorded revised ball limit:",
    revised_innings_detail["revised_balls_limit"].isna().sum()
)

display(
    revised_innings_detail[
        [
            "match_id",
            "match_date",
            "comp_name",
            "innings",
            "total_runs",
            "wickets",
            "total_deliveries",
            "delivery_records",
            "legal_balls_known",
            "unknown_legality",
            "revised_balls_limit",
            "revised_target",
            "result_category"
        ]
    ]
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 4
# Validate revised ball limits and delivery legality
# ============================================================

revised_legality_check = (
    ball_by_ball_analysis[
        ball_by_ball_analysis["match_id"].isin(revised_matches["match_id"])
        & ball_by_ball_analysis["innings"].isin([1, 2])
    ]
    .groupby(["match_id", "innings"])
    .agg(
        delivery_records=("match_id", "size"),
        legal_balls_known=(
            "is_legal_delivery",
            lambda s: int(s.eq(True).sum())
        ),
        unknown_legality_count=(
            "is_legal_delivery",
            lambda s: int(s.isna().sum())
        )
    )
    .reset_index()
)

revised_limit_check = revised_innings_detail[
    [
        "match_id",
        "innings",
        "revised_balls_limit",
        "total_runs",
        "wickets"
    ]
].merge(
    revised_legality_check,
    on=["match_id", "innings"],
    how="left",
    validate="one_to_one"
)

revised_limit_check["balls_remaining"] = (
    revised_limit_check["revised_balls_limit"]
    - revised_limit_check["legal_balls_known"]
)

revised_limit_check["limit_status"] = "No recorded limit"

has_limit = revised_limit_check["revised_balls_limit"].notna()

revised_limit_check.loc[
    has_limit & (revised_limit_check["balls_remaining"] > 0),
    "limit_status"
] = "Below revised limit"

revised_limit_check.loc[
    has_limit & (revised_limit_check["balls_remaining"] == 0),
    "limit_status"
] = "Reached revised limit"

revised_limit_check.loc[
    has_limit & (revised_limit_check["balls_remaining"] < 0),
    "limit_status"
] = "Exceeds revised limit"

print("Total innings checked:", len(revised_limit_check))

print("\nLimit status distribution:")
display(revised_limit_check["limit_status"].value_counts())

print("\nInnings with unknown delivery legality:")
display(
    revised_limit_check[
        revised_limit_check["unknown_legality_count"] > 0
    ][
        [
            "match_id",
            "innings",
            "delivery_records",
            "legal_balls_known",
            "unknown_legality_count",
            "revised_balls_limit",
            "limit_status"
        ]
    ]
)

print("\nInnings exceeding revised ball limits:")
display(
    revised_limit_check[
        revised_limit_check["limit_status"] == "Exceeds revised limit"
    ]
)

print("\nComplete revised limit validation:")
display(revised_limit_check)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 5
# Correct limit status for unknown delivery legality
# ============================================================

revised_limit_check["limit_status_final"] = "No recorded limit"

has_limit = revised_limit_check["revised_balls_limit"].notna()
has_unknown = revised_limit_check["unknown_legality_count"] > 0
known_legality = has_limit & ~has_unknown

revised_limit_check.loc[
    known_legality & (revised_limit_check["balls_remaining"] > 0),
    "limit_status_final"
] = "Below revised limit"

revised_limit_check.loc[
    known_legality & (revised_limit_check["balls_remaining"] == 0),
    "limit_status_final"
] = "Reached revised limit"

revised_limit_check.loc[
    known_legality & (revised_limit_check["balls_remaining"] < 0),
    "limit_status_final"
] = "Exceeds revised limit"

revised_limit_check.loc[
    has_limit & has_unknown,
    "limit_status_final"
] = "Uncertain - unknown legality"

print("Final limit status distribution:")
display(
    revised_limit_check["limit_status_final"]
    .value_counts()
)

print("\nInnings requiring uncertainty handling:")
display(
    revised_limit_check[
        revised_limit_check["limit_status_final"]
        == "Uncertain - unknown legality"
    ][
        [
            "match_id",
            "innings",
            "revised_balls_limit",
            "legal_balls_known",
            "unknown_legality_count",
            "balls_remaining"
        ]
    ]
)

print("\nAny innings exceeding revised limits?")
print(
    (
        revised_limit_check["limit_status_final"]
        == "Exceeds revised limit"
    ).any()
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 6
# Create revised-match delivery dataset
# ============================================================

revised_deliveries = ball_by_ball_analysis[
    ball_by_ball_analysis["match_id"].isin(
        revised_matches["match_id"]
    )
    & ball_by_ball_analysis["innings"].isin([1, 2])
].copy()

print("Total revised-match delivery records:", len(revised_deliveries))

print(
    "Unique revised matches:",
    revised_deliveries["match_id"].nunique()
)

print(
    "\nInnings distribution:"
)
display(revised_deliveries["innings"].value_counts().sort_index())

print(
    "\nDelivery legality distribution:"
)
display(revised_deliveries["is_legal_delivery"].value_counts(dropna=False))

print(
    "\nDuplicate delivery keys:",
    revised_deliveries.duplicated(
        ["match_id", "innings", "innings_ball_num"]
    ).sum()
)

print(
    "\nRevised matches without delivery records:",
    len(
        set(revised_matches["match_id"])
        - set(revised_deliveries["match_id"])
    )
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 7
# Calculate revised-match batting performance
# ============================================================

revised_deliveries["is_batter_ball_faced"] = (
    revised_deliveries["is_legal_delivery"].eq(True)
    | revised_deliveries["delivery_legality"].eq("No-ball")
)

revised_batting = (
    revised_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        runs=("batter_runs", "sum"),
        balls_faced=("is_batter_ball_faced", "sum"),
        fours=("is_four", "sum"),
        sixes=("is_six", "sum")
    )
)

revised_batting["strike_rate"] = (
    revised_batting["runs"]
    .div(revised_batting["balls_faced"])
    .mul(100)
).fillna(0).round(2)

print("Total batting records:", len(revised_batting))
print("Unique batters:", revised_batting["striker_id"].nunique())
print("Unique matches:", revised_batting["match_id"].nunique())

print("\nTotal batting runs:", revised_batting["runs"].sum())
print("Total balls faced:", revised_batting["balls_faced"].sum())
print("Total fours:", revised_batting["fours"].sum())
print("Total sixes:", revised_batting["sixes"].sum())

print("\nMissing values:")
display(revised_batting.isna().sum())

print("\nTop 10 run scorers in revised matches:")
display(
    revised_batting
    .groupby("striker_name", as_index=False)
    .agg(
        innings=("match_id", "count"),
        runs=("runs", "sum"),
        balls_faced=("balls_faced", "sum"),
        fours=("fours", "sum"),
        sixes=("sixes", "sum")
    )
    .sort_values("runs", ascending=False)
    .head(10)
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 8
# Validate batting totals against innings totals
# ============================================================

revised_batting_validation = (
    revised_batting
    .groupby(["match_id", "innings"], as_index=False)
    .agg(
        batter_runs=("runs", "sum"),
        batter_fours=("fours", "sum"),
        batter_sixes=("sixes", "sum")
    )
    .merge(
        innings_analysis[
            [
                "match_id",
                "innings",
                "batter_runs",
                "fours",
                "sixes"
            ]
        ].rename(
            columns={
                "batter_runs": "innings_batter_runs",
                "fours": "innings_fours",
                "sixes": "innings_sixes"
            }
        ),
        on=["match_id", "innings"],
        how="left",
        validate="one_to_one"
    )
)

revised_batting_validation["runs_match"] = (
    revised_batting_validation["batter_runs"]
    == revised_batting_validation["innings_batter_runs"]
)

revised_batting_validation["fours_match"] = (
    revised_batting_validation["batter_fours"]
    == revised_batting_validation["innings_fours"]
)

revised_batting_validation["sixes_match"] = (
    revised_batting_validation["batter_sixes"]
    == revised_batting_validation["innings_sixes"]
)

print("Total innings validated:", len(revised_batting_validation))

print("\nValidation summary:")
print("Batter runs match:", revised_batting_validation["runs_match"].all())
print("Fours match:", revised_batting_validation["fours_match"].all())
print("Sixes match:", revised_batting_validation["sixes_match"].all())

print("\nAny mismatches:")
display(
    revised_batting_validation[
        ~(
            revised_batting_validation["runs_match"]
            & revised_batting_validation["fours_match"]
            & revised_batting_validation["sixes_match"]
        )
    ]
)
```

```python
# ============================================================
# REVISED MATCH ANALYSIS - STEP 9
# Calculate bowling performance
# ============================================================

# Bowler-credited dismissals only
bowler_credited_dismissals = [
    "Caught",
    "Bowled",
    "LBW",
    "Stump",
    "Hit Wicket"
]

# Legal balls exclude wides and no-balls.
# Unknown legality is excluded from the legal-ball count.
revised_deliveries["is_legal_bowling_ball"] = (
    revised_deliveries["is_legal_delivery"].eq(True)
)

# Runs conceded by bowler:
# Batter runs + extras1 on wides and no-balls.
# Byes, leg-byes and extras2 are excluded.
revised_deliveries["bowler_runs_conceded"] = (
    revised_deliveries["batter_runs"]
    + revised_deliveries["extras1_runs"].where(
        revised_deliveries["delivery_legality"].isin(
            ["Wide", "No-ball"]
        ),
        0
    )
)

# Bowler-credited wickets
revised_deliveries["bowler_wicket"] = (
    revised_deliveries["is_wicket"].eq(1)
    & revised_deliveries["dismissal_type"].isin(
        bowler_credited_dismissals
    )
)

# Aggregate match-level bowling figures
revised_bowling = (
    revised_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id",
            "bowler_name"
        ],
        as_index=False
    )
    .agg(
        legal_balls=("is_legal_bowling_ball", "sum"),
        runs_conceded=("bowler_runs_conceded", "sum"),
        wickets=("bowler_wicket", "sum")
    )
)

# Calculate overs and economy
revised_bowling["overs"] = (
    revised_bowling["legal_balls"] // 6
    + (revised_bowling["legal_balls"] % 6) / 10
)

revised_bowling["economy"] = (
    revised_bowling["runs_conceded"]
    .div(revised_bowling["legal_balls"])
    .mul(6)
).fillna(0).round(2)

print("Revised bowling records:", len(revised_bowling))
print("Unique bowlers:", revised_bowling["bowler_id"].nunique())
print("Unique matches:", revised_bowling["match_id"].nunique())

print("\nBowling totals:")
print("Legal balls:", revised_bowling["legal_balls"].sum())
print("Runs conceded:", revised_bowling["runs_conceded"].sum())
print("Bowler-credited wickets:", revised_bowling["wickets"].sum())

print("\nMissing values:")
display(revised_bowling.isna().sum())

print("\nTop 10 wicket-takers:")
display(
    revised_bowling
    .groupby("bowler_name", as_index=False)
    .agg(
        innings=("match_id", "count"),
        wickets=("wickets", "sum"),
        runs_conceded=("runs_conceded", "sum"),
        legal_balls=("legal_balls", "sum")
    )
    .sort_values(
        ["wickets", "runs_conceded"],
        ascending=[False, True]
    )
    .head(10)
)
```

```python
# ============================================================
# REVISED MATCH ANALYSIS - STEP 10
# Validate bowling calculations
# ============================================================

# 1. Validate legal balls
expected_legal_balls = (
    revised_deliveries["is_legal_delivery"].eq(True).sum()
)

calculated_legal_balls = revised_bowling["legal_balls"].sum()

# 2. Validate bowler runs conceded
expected_bowler_runs = (
    revised_deliveries["batter_runs"].sum()
    + revised_deliveries.loc[
        revised_deliveries["delivery_legality"].isin(
            ["Wide", "No-ball"]
        ),
        "extras1_runs"
    ].sum()
)

calculated_bowler_runs = revised_bowling["runs_conceded"].sum()

# 3. Validate bowler-credited wickets
expected_bowler_wickets = (
    revised_deliveries["is_wicket"].eq(1)
    & revised_deliveries["dismissal_type"].isin(
        bowler_credited_dismissals
    )
).sum()

calculated_bowler_wickets = revised_bowling["wickets"].sum()

# 4. Compare all wickets with bowler-credited wickets
total_recorded_wickets = revised_deliveries["is_wicket"].eq(1).sum()

uncredited_wickets = (
    total_recorded_wickets - expected_bowler_wickets
)

# 5. Check unknown delivery legality
unknown_legality_deliveries = (
    revised_deliveries["is_legal_delivery"].isna().sum()
)

# 6. Check duplicate bowling records
duplicate_bowling_records = revised_bowling.duplicated(
    subset=[
        "match_id",
        "innings",
        "bowling_team_id",
        "bowler_id"
    ]
).sum()

print("LEGAL BALL VALIDATION")
print("Expected legal balls:", expected_legal_balls)
print("Calculated legal balls:", calculated_legal_balls)
print("Match:", expected_legal_balls == calculated_legal_balls)

print("\nBOWLER RUNS VALIDATION")
print("Expected bowler runs:", expected_bowler_runs)
print("Calculated bowler runs:", calculated_bowler_runs)
print("Match:", expected_bowler_runs == calculated_bowler_runs)

print("\nWICKET VALIDATION")
print("All recorded wickets:", total_recorded_wickets)
print("Expected bowler-credited wickets:", expected_bowler_wickets)
print("Calculated bowler-credited wickets:", calculated_bowler_wickets)
print("Uncredited wickets:", uncredited_wickets)
print(
    "Match:",
    expected_bowler_wickets == calculated_bowler_wickets
)

print("\nDATA QUALITY")
print("Unknown legality deliveries:", unknown_legality_deliveries)
print("Duplicate bowling records:", duplicate_bowling_records)
```

```python
# ============================================================
# SUPER OVER ANALYSIS - STEP 12
# Identify recorded Super Over innings
# ============================================================

super_over_deliveries = ball_by_ball_analysis[
    ball_by_ball_analysis["innings"].isin([3, 4, 5, 6])
].copy()

print("Total Super Over delivery records:",
      len(super_over_deliveries))

print("Unique matches:",
      super_over_deliveries["match_id"].nunique())

print("\nSuper Over innings distribution:")
display(
    super_over_deliveries
    .groupby(["match_id", "innings"], as_index=False)
    .agg(
        delivery_records=("innings_ball_num", "count"),
        total_runs=("team_runs", "sum"),
        wickets=("is_wicket", "sum")
    )
    .sort_values(["match_id", "innings"])
)

print("\nSuper Over match details:")
display(
    super_over_deliveries[
        ["match_id"]
    ].drop_duplicates()
    .merge(
        matches_analysis[
            ["match_id", "comp_id", "match_date",
             "home_team", "away_team", "result"]
        ],
        on="match_id",
        how="left"
    )
    .merge(
        competitions_copy[
            ["comp_id", "comp_name", "match_format"]
        ],
        on="comp_id",
        how="left"
    )
    .sort_values("match_date")
)
```

```python

# ============================================================
# SUPER OVER ANALYSIS - STEP 13
# Calculate batting performance
# ============================================================

# 1. Identify balls faced
# Legal deliveries and no-balls count as balls faced.
# Wides are excluded.

super_over_deliveries["is_batter_ball_faced"] = (
    super_over_deliveries["is_legal_delivery"].eq(True)
    | super_over_deliveries["delivery_legality"].eq("No-ball")
)


# 2. Calculate match-level batting performance

super_over_batting = (
    super_over_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        runs=("batter_runs", "sum"),
        balls_faced=("is_batter_ball_faced", "sum"),
        fours=("is_four", "sum"),
        sixes=("is_six", "sum")
    )
)

# Calculate match-level strike rate
super_over_batting["strike_rate"] = (
    super_over_batting["runs"]
    .div(super_over_batting["balls_faced"])
    .mul(100)
).fillna(0).round(2)


# 3. Display overall batting summary

print("SUPER OVER BATTING SUMMARY")
print("-" * 45)

print("Batting records:", len(super_over_batting))
print("Unique batters:", super_over_batting["striker_id"].nunique())
print("Unique matches:", super_over_batting["match_id"].nunique())

print("\nBatting totals:")
print("Runs:", super_over_batting["runs"].sum())
print("Balls faced:", super_over_batting["balls_faced"].sum())
print("Fours:", super_over_batting["fours"].sum())
print("Sixes:", super_over_batting["sixes"].sum())


# 4. Display match-level batting records

print("\nSUPER OVER MATCH-LEVEL BATTING")
print("-" * 45)

display(
    super_over_batting
    .sort_values(
        ["match_id", "innings", "runs"],
        ascending=[True, True, False]
    )
)


# 5. Aggregate player-level Super Over statistics
# Group by player_id to avoid combining players with
# identical names.

super_over_batting_rankings = (
    super_over_batting
    .groupby("striker_id", as_index=False)
    .agg(
        striker_name=("striker_name", "first"),
        innings=("match_id", "count"),
        runs=("runs", "sum"),
        balls_faced=("balls_faced", "sum"),
        fours=("fours", "sum"),
        sixes=("sixes", "sum")
    )
)

# Calculate aggregate player strike rate
super_over_batting_rankings["strike_rate"] = (
    super_over_batting_rankings["runs"]
    .div(super_over_batting_rankings["balls_faced"])
    .mul(100)
).fillna(0).round(2)


# 6. Display Top 10 Super Over run scorers

print("\nTOP 10 SUPER OVER RUN SCORERS")
print("-" * 45)

display(
    super_over_batting_rankings
    .sort_values(
        ["runs", "strike_rate"],
        ascending=[False, False]
    )
    .head(10)
)


# 7. Data quality checks

print("\nDATA QUALITY CHECKS")
print("-" * 45)

print(
    "Duplicate player-match-innings records:",
    super_over_batting.duplicated(
        subset=["match_id", "innings", "striker_id"]
    ).sum()
)

print(
    "Missing values:",
    super_over_batting.isna().sum().sum()
)

print(
    "Negative batting runs:",
    (super_over_batting["runs"] < 0).sum()
)
```

```python

# ============================================================
# SUPER OVER ANALYSIS - STEP 14
# Calculate bowling performance
# ============================================================

# 1. Legal bowling deliveries
super_over_deliveries["is_legal_bowling_ball"] = (
    super_over_deliveries["is_legal_delivery"].eq(True)
)

# 2. Bowler runs conceded
# Batter runs + extras1 on wides and no-balls.
# Byes, leg-byes and extras2 are excluded.

super_over_deliveries["bowler_runs_conceded"] = (
    super_over_deliveries["batter_runs"]
    + super_over_deliveries["extras1_runs"].where(
        super_over_deliveries["delivery_legality"].isin(
            ["Wide", "No-ball"]
        ),
        0
    )
)

# 3. Bowler-credited wickets

super_over_deliveries["bowler_wicket"] = (
    super_over_deliveries["is_wicket"].eq(1)
    & super_over_deliveries["dismissal_type"].isin(
        bowler_credited_dismissals
    )
)

# 4. Aggregate bowling figures

super_over_bowling = (
    super_over_deliveries
    .groupby(
        [
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id",
            "bowler_name"
        ],
        as_index=False
    )
    .agg(
        legal_balls=("is_legal_bowling_ball", "sum"),
        runs_conceded=("bowler_runs_conceded", "sum"),
        wickets=("bowler_wicket", "sum")
    )
)

# 5. Calculate overs and economy

super_over_bowling["overs"] = (
    super_over_bowling["legal_balls"] // 6
    + (super_over_bowling["legal_balls"] % 6) / 10
)

super_over_bowling["economy"] = (
    super_over_bowling["runs_conceded"]
    .div(super_over_bowling["legal_balls"])
    .mul(6)
).fillna(0).round(2)

# 6. Summary

print("SUPER OVER BOWLING SUMMARY")
print("-" * 45)

print("Bowling records:", len(super_over_bowling))
print("Unique bowlers:", super_over_bowling["bowler_id"].nunique())
print("Unique matches:", super_over_bowling["match_id"].nunique())

print("\nBowling totals:")
print("Legal balls:", super_over_bowling["legal_balls"].sum())
print("Runs conceded:", super_over_bowling["runs_conceded"].sum())
print("Bowler-credited wickets:", super_over_bowling["wickets"].sum())

# 7. Match-level bowling records

print("\nSUPER OVER MATCH-LEVEL BOWLING")
print("-" * 45)

display(
    super_over_bowling
    .sort_values(
        ["match_id", "innings", "wickets"],
        ascending=[True, True, False]
    )
)

# 8. Player-level bowling statistics

super_over_bowling_rankings = (
    super_over_bowling
    .groupby("bowler_id", as_index=False)
    .agg(
        bowler_name=("bowler_name", "first"),
        innings=("match_id", "count"),
        legal_balls=("legal_balls", "sum"),
        runs_conceded=("runs_conceded", "sum"),
        wickets=("wickets", "sum")
    )
)

super_over_bowling_rankings["overs"] = (
    super_over_bowling_rankings["legal_balls"] // 6
    + (super_over_bowling_rankings["legal_balls"] % 6) / 10
)

super_over_bowling_rankings["economy"] = (
    super_over_bowling_rankings["runs_conceded"]
    .div(super_over_bowling_rankings["legal_balls"])
    .mul(6)
).fillna(0).round(2)

# 9. Top 10 wicket-takers

print("\nTOP 10 SUPER OVER WICKET-TAKERS")
print("-" * 45)

display(
    super_over_bowling_rankings
    .sort_values(
        ["wickets", "economy"],
        ascending=[False, True]
    )
    .head(10)
)

# 10. Data quality checks

print("\nDATA QUALITY CHECKS")
print("-" * 45)

print(
    "Duplicate bowling records:",
    super_over_bowling.duplicated(
        subset=[
            "match_id",
            "innings",
            "bowling_team_id",
            "bowler_id"
        ]
    ).sum()
)

print("Missing values:")
display(super_over_bowling.isna().sum())
```

```python

# ============================================================
# SUPER OVER ANALYSIS - STEP 15
# Validate bowling calculations
# ============================================================

# 1. Legal balls
expected_legal_balls = (
    super_over_deliveries["is_legal_delivery"].eq(True).sum()
)

calculated_legal_balls = super_over_bowling["legal_balls"].sum()

# 2. Bowler runs conceded
expected_bowler_runs = (
    super_over_deliveries["batter_runs"].sum()
    + super_over_deliveries.loc[
        super_over_deliveries["delivery_legality"].isin(
            ["Wide", "No-ball"]
        ),
        "extras1_runs"
    ].sum()
)

calculated_bowler_runs = super_over_bowling["runs_conceded"].sum()

# 3. Bowler-credited wickets
expected_bowler_wickets = (
    super_over_deliveries["is_wicket"].eq(1)
    & super_over_deliveries["dismissal_type"].isin(
        bowler_credited_dismissals
    )
).sum()

calculated_bowler_wickets = super_over_bowling["wickets"].sum()

# 4. All recorded wickets
total_recorded_wickets = (
    super_over_deliveries["is_wicket"].eq(1).sum()
)

uncredited_wickets = (
    total_recorded_wickets - expected_bowler_wickets
)

# 5. Unknown legality
unknown_legality_deliveries = (
    super_over_deliveries["is_legal_delivery"].isna().sum()
)

# 6. Duplicate records
duplicate_bowling_records = super_over_bowling.duplicated(
    subset=[
        "match_id",
        "innings",
        "bowling_team_id",
        "bowler_id"
    ]
).sum()

print("LEGAL BALL VALIDATION")
print("Expected:", expected_legal_balls)
print("Calculated:", calculated_legal_balls)
print("Match:", expected_legal_balls == calculated_legal_balls)

print("\nBOWLER RUNS VALIDATION")
print("Expected:", expected_bowler_runs)
print("Calculated:", calculated_bowler_runs)
print("Match:", expected_bowler_runs == calculated_bowler_runs)

print("\nWICKET VALIDATION")
print("All recorded wickets:", total_recorded_wickets)
print("Expected credited wickets:", expected_bowler_wickets)
print("Calculated credited wickets:", calculated_bowler_wickets)
print("Uncredited wickets:", uncredited_wickets)
print("Match:", expected_bowler_wickets == calculated_bowler_wickets)

print("\nDATA QUALITY")
print("Unknown legality:", unknown_legality_deliveries)
print("Duplicate bowling records:", duplicate_bowling_records)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 16
# Prepare delivery-level context for batting pressure analysis
# ============================================================

# 1. Select revised match metadata
revised_match_context = matches_analysis.loc[
    matches_analysis["match_condition"] == "REVISED",
    [
        "match_id",
        "first_inn_revised_overs",
        "second_inn_revised_overs",
        "revised_target",
        "match_date",
        "comp_id",
        "win_team_id",
        "result"
    ]
].copy()

# 2. Prepare innings-level scoring context
revised_innings_context = innings_analysis.loc[
    innings_analysis["match_id"].isin(
        revised_match_context["match_id"]
    ),
    [
        "match_id",
        "innings",
        "batting_team_id",
        "bowling_team_id",
        "total_runs",
        "wickets"
    ]
].copy()

# 3. Merge match and innings context into deliveries
revised_pressure_deliveries = revised_deliveries.merge(
    revised_match_context,
    on="match_id",
    how="left",
    validate="many_to_one"
)

revised_pressure_deliveries = revised_pressure_deliveries.merge(
    revised_innings_context,
    on=[
        "match_id",
        "innings",
        "batting_team_id",
        "bowling_team_id"
    ],
    how="left",
    suffixes=("", "_innings"),
    validate="many_to_one"
)

# 4. Display summary
print("REVISED PRESSURE DELIVERY SUMMARY")
print("---------------------------------")
print("Delivery records:", len(revised_pressure_deliveries))
print("Unique matches:", revised_pressure_deliveries["match_id"].nunique())

print("\nInnings distribution:")
print(
    revised_pressure_deliveries["innings"]
    .value_counts()
    .sort_index()
)

print("\nMissing context:")
context_columns = [
    "first_inn_revised_overs",
    "second_inn_revised_overs",
    "revised_target",
    "total_runs",
    "wickets"
]

print(
    revised_pressure_deliveries[context_columns]
    .isna()
    .sum()
)

print("\nDuplicate delivery keys:")
print(
    revised_pressure_deliveries.duplicated(
        subset=["match_id", "innings", "innings_ball_num"]
    ).sum()
)

print("\nSample records:")
display(
    revised_pressure_deliveries[
        [
            "match_id",
            "innings",
            "overid",
            "innings_ball_num",
            "batting_team_id",
            "bowling_team_id",
            "runs",
            "is_wicket",
            "first_inn_revised_overs",
            "second_inn_revised_overs",
            "revised_target",
            "total_runs",
            "wickets"
        ]
    ].head(10)
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 17
# Inspect revised overs and innings quotas
# ============================================================

# 1. Add match format
revised_match_diagnostic = revised_match_context.merge(
    competitions_copy[["comp_id", "match_format"]],
    on="comp_id",
    how="left",
    validate="many_to_one"
)

# 2. Display revised match metadata
print("REVISED MATCH QUOTA METADATA")
print("---------------------------------")

display(
    revised_match_diagnostic[
        [
            "match_id",
            "match_format",
            "first_inn_revised_overs",
            "second_inn_revised_overs",
            "revised_target"
        ]
    ].sort_values("match_id")
)

# 3. Check revised quota availability by innings
revised_pressure_deliveries["applicable_revised_quota"] = (
    revised_pressure_deliveries["first_inn_revised_overs"].where(
        revised_pressure_deliveries["innings"].eq(1),
        revised_pressure_deliveries["second_inn_revised_overs"]
    )
)

print("\nAPPLICABLE REVISED QUOTA BY INNINGS")
print("---------------------------------")

for innings_number in [1, 2]:
    innings_data = revised_pressure_deliveries[
        revised_pressure_deliveries["innings"] == innings_number
    ]

    print(f"\nInnings {innings_number}")
    print("Delivery records:", len(innings_data))
    print(
        "Matches with revised quota:",
        innings_data.loc[
            innings_data["applicable_revised_quota"].notna(),
            "match_id"
        ].nunique()
    )
    print(
        "Matches without revised quota:",
        innings_data.loc[
            innings_data["applicable_revised_quota"].isna(),
            "match_id"
        ].nunique()
    )

# 4. Check whether recorded revised quotas are in balls
quota_values = (
    revised_match_diagnostic[
        ["first_inn_revised_overs", "second_inn_revised_overs"]
    ]
    .stack()
)

print("\nREVISED QUOTA UNIT CHECK")
print("---------------------------------")
print("Recorded quota values:", sorted(quota_values.unique()))
print(
    "Values not divisible by 6:",
    (quota_values % 6 != 0).sum()
)

# 5. Show matches where the applicable quota is missing
missing_quota_matches = (
    revised_pressure_deliveries.loc[
        revised_pressure_deliveries["applicable_revised_quota"].isna(),
        ["match_id", "innings"]
    ]
    .drop_duplicates()
    .sort_values(["match_id", "innings"])
)

print("\nMATCHES WITHOUT AN APPLICABLE REVISED QUOTA")
print("---------------------------------")
display(missing_quota_matches)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 18
# Investigate innings without explicit revised quotas
# ============================================================

# 1. Get innings-level delivery statistics
missing_quota_innings_detail = (
    revised_pressure_deliveries[
        revised_pressure_deliveries["applicable_revised_quota"].isna()
    ]
    .groupby(
        ["match_id", "innings"],
        as_index=False
    )
    .agg(
        delivery_records=("innings_ball_num", "count"),
        legal_deliveries=("is_legal_delivery", lambda x: x.eq(True).sum()),
        max_over_id=("overid", "max"),
        max_innings_ball_num=("innings_ball_num", "max"),
        total_runs=("total_runs", "first"),
        wickets=("wickets", "first"),
        first_inn_revised_overs=("first_inn_revised_overs", "first"),
        second_inn_revised_overs=("second_inn_revised_overs", "first"),
        revised_target=("revised_target", "first"),
        result=("result", "first")
    )
)

# 2. Attach match format
missing_quota_innings_detail = missing_quota_innings_detail.merge(
    revised_match_diagnostic[
        ["match_id", "match_format"]
    ],
    on="match_id",
    how="left",
    validate="many_to_one"
)

# 3. Sort and display
missing_quota_innings_detail = missing_quota_innings_detail.sort_values(
    ["match_id", "innings"]
).reset_index(drop=True)

print("INVESTIGATION: INNINGS WITHOUT EXPLICIT REVISED QUOTA")
print("------------------------------------------------------")
display(missing_quota_innings_detail)

# 4. Show the complete match-level innings summary for these matches
investigation_match_ids = (
    missing_quota_innings_detail["match_id"].unique()
)

print("\nALL AVAILABLE INNINGS FOR THESE MATCHES")
print("---------------------------------------")

display(
    innings_analysis.loc[
        innings_analysis["match_id"].isin(investigation_match_ids),
        [
            "match_id",
            "innings",
            "batting_team_id",
            "bowling_team_id",
            "total_runs",
            "wickets"
        ]
    ].sort_values(["match_id", "innings"])
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 19
# Inspect final over of innings without explicit quota
# ============================================================

last_over_details = []

for _, row in missing_quota_innings_detail.iterrows():

    match_id = row["match_id"]
    innings_number = row["innings"]
    last_over = row["max_over_id"]

    over_data = revised_pressure_deliveries[
        (revised_pressure_deliveries["match_id"] == match_id)
        & (revised_pressure_deliveries["innings"] == innings_number)
        & (revised_pressure_deliveries["overid"] == last_over)
    ].copy()

    last_over_details.append({
        "match_id": match_id,
        "innings": innings_number,
        "last_over": last_over,
        "deliveries_in_last_over": len(over_data),
        "legal_balls_in_last_over": (
            over_data["is_legal_delivery"].eq(True).sum()
        ),
        "runs_in_last_over": over_data["team_runs"].sum(),
        "wickets_in_last_over": over_data["is_wicket"].eq(True).sum(),
        "last_innings_ball_num": over_data["innings_ball_num"].max(),
        "total_innings_wickets": row["wickets"]
    })

last_over_diagnostic = pd.DataFrame(last_over_details)

print("FINAL OVER DIAGNOSTICS")
print("---------------------------------")
display(last_over_diagnostic.sort_values(["match_id", "innings"]))
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 20
# Assign innings quotas and identify death overs
# ============================================================

import math

# 1. Start with recorded revised quota
revised_pressure_deliveries["innings_quota_balls"] = (
    revised_pressure_deliveries["applicable_revised_quota"]
)

revised_pressure_deliveries["quota_source"] = "RECORDED_REVISED"

# 2. Missing first-innings quota:
# Use original 20-over T20/T20I quota = 120 balls
missing_first = (
    revised_pressure_deliveries["innings"].eq(1)
    & revised_pressure_deliveries["innings_quota_balls"].isna()
)

revised_pressure_deliveries.loc[
    missing_first, "innings_quota_balls"
] = 120

revised_pressure_deliveries.loc[
    missing_first, "quota_source"
] = "ORIGINAL_20_OVERS"

# 3. Missing second-innings quota:
# Use first-innings revised quota when available.
missing_second = (
    revised_pressure_deliveries["innings"].eq(2)
    & revised_pressure_deliveries["innings_quota_balls"].isna()
)

revised_pressure_deliveries.loc[
    missing_second, "innings_quota_balls"
] = revised_pressure_deliveries.loc[
    missing_second, "first_inn_revised_overs"
]

revised_pressure_deliveries.loc[
    missing_second, "quota_source"
] = "INFERRED_FROM_FIRST_INNINGS"

# Fallback only if the first-innings quota is also missing
remaining_missing = (
    revised_pressure_deliveries["innings_quota_balls"].isna()
)

revised_pressure_deliveries.loc[
    remaining_missing, "innings_quota_balls"
] = 120

revised_pressure_deliveries.loc[
    remaining_missing, "quota_source"
] = "ORIGINAL_20_OVERS_FALLBACK"

# 4. Convert quota from balls to overs
revised_pressure_deliveries["innings_quota_overs"] = (
    revised_pressure_deliveries["innings_quota_balls"] / 6
).astype(int)

# 5. Calculate final 25% death overs
revised_pressure_deliveries["death_overs_count"] = (
    revised_pressure_deliveries["innings_quota_overs"]
    .apply(lambda overs: math.ceil(overs * 0.25))
)

revised_pressure_deliveries["death_over_start"] = (
    revised_pressure_deliveries["innings_quota_overs"]
    - revised_pressure_deliveries["death_overs_count"]
    + 1
)

# 6. Identify death-over deliveries
revised_pressure_deliveries["is_death_over"] = (
    revised_pressure_deliveries["overid"].between(
        revised_pressure_deliveries["death_over_start"],
        revised_pressure_deliveries["innings_quota_overs"]
    )
)

# 7. Build innings-level quota summary
revised_quota_summary = (
    revised_pressure_deliveries.groupby(
        ["match_id", "innings"],
        as_index=False
    )
    .agg(
        quota_balls=("innings_quota_balls", "first"),
        quota_overs=("innings_quota_overs", "first"),
        quota_source=("quota_source", "first"),
        death_over_start=("death_over_start", "first"),
        death_over_end=("innings_quota_overs", "first"),
        delivery_records=("innings_ball_num", "count"),
        death_over_deliveries=("is_death_over", "sum")
    )
)

print("REVISED INNINGS QUOTA SUMMARY")
print("---------------------------------")
print("Total innings:", len(revised_quota_summary))
print("\nQuota sources:")
print(revised_quota_summary["quota_source"].value_counts())

print("\nMissing quotas:")
print(revised_pressure_deliveries["innings_quota_balls"].isna().sum())

print("\nQuota values not divisible by 6:")
print(
    (revised_pressure_deliveries["innings_quota_balls"] % 6 != 0).sum()
)

print("\nDeath-over distribution:")
display(
    revised_quota_summary[
        [
            "match_id",
            "innings",
            "quota_balls",
            "quota_overs",
            "quota_source",
            "death_over_start",
            "death_over_end",
            "death_over_deliveries"
        ]
    ].sort_values(["match_id", "innings"])
)

print("\nDeath-over examples:")
display(
    revised_pressure_deliveries[
        [
            "match_id",
            "innings",
            "overid",
            "innings_quota_overs",
            "death_over_start",
            "is_death_over"
        ]
    ].drop_duplicates().head(15)
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 22
# Calculate batting pressure delivery by delivery
# ============================================================

import numpy as np
import pandas as pd

# 1. Create a separate working copy
revised_batting_pressure = revised_pressure_deliveries.copy()

# Sort deliveries chronologically within each innings
revised_batting_pressure = revised_batting_pressure.sort_values(
    ["match_id", "innings", "innings_ball_num"]
).copy()

group_cols = ["match_id", "innings"]

# 2. Calculate runs before the current delivery
revised_batting_pressure["runs_before_delivery"] = (
    revised_batting_pressure.groupby(group_cols)["team_runs"].cumsum()
    - revised_batting_pressure["team_runs"]
)

# 3. Calculate wickets before the current delivery
revised_batting_pressure["wicket_flag"] = (
    revised_batting_pressure["is_wicket"]
    .fillna(0)
    .astype(int)
)

revised_batting_pressure["wickets_before_delivery"] = (
    revised_batting_pressure.groupby(group_cols)["wicket_flag"].cumsum()
    - revised_batting_pressure["wicket_flag"]
)

# 4. Calculate legal balls before the current delivery
# Unknown legality is NOT counted as a confirmed legal ball.
revised_batting_pressure["legal_ball_flag"] = (
    revised_batting_pressure["is_legal_delivery"]
    .eq(True)
    .fillna(False)
    .astype(int)
)

revised_batting_pressure["legal_balls_before_delivery"] = (
    revised_batting_pressure.groupby(group_cols)["legal_ball_flag"].cumsum()
    - revised_batting_pressure["legal_ball_flag"]
)

# 5. Calculate remaining wickets and balls
revised_batting_pressure["wickets_remaining"] = (
    10 - revised_batting_pressure["wickets_before_delivery"]
)

revised_batting_pressure["legal_balls_remaining"] = (
    revised_batting_pressure["innings_quota_balls"]
    - revised_batting_pressure["legal_balls_before_delivery"]
)

# 6. Identify first-innings pressure
# Pressure applies when only 1–3 wickets remain.
revised_batting_pressure["low_wickets_remaining"] = (
    revised_batting_pressure["wickets_remaining"].between(1, 3)
)

revised_batting_pressure["first_innings_pressure"] = (
    revised_batting_pressure["innings"].eq(1)
    & revised_batting_pressure["low_wickets_remaining"]
)

# 7. Determine second-innings target
first_innings_targets = (
    revised_innings_context[
        revised_innings_context["innings"].eq(1)
    ][["match_id", "total_runs"]]
    .rename(columns={"total_runs": "first_innings_runs"})
)

revised_batting_pressure = revised_batting_pressure.merge(
    first_innings_targets,
    on="match_id",
    how="left",
    validate="many_to_one"
)

# Use revised target when available.
# Otherwise use first-innings score + 1.
revised_batting_pressure["target_for_chase"] = (
    revised_batting_pressure["revised_target"]
    .fillna(revised_batting_pressure["first_innings_runs"] + 1)
)

# 8. Calculate runs required before current delivery
revised_batting_pressure["runs_required"] = (
    revised_batting_pressure["target_for_chase"]
    - revised_batting_pressure["runs_before_delivery"]
)

# 9. Calculate required run rate
revised_batting_pressure["required_run_rate"] = np.where(
    revised_batting_pressure["legal_balls_remaining"] > 0,
    (
        revised_batting_pressure["runs_required"] * 6
        / revised_batting_pressure["legal_balls_remaining"]
    ),
    np.nan
)

revised_batting_pressure["high_required_run_rate"] = (
    revised_batting_pressure["required_run_rate"] >= 8
)

# 10. Second innings pressure
# At least 2 of 3 conditions must be satisfied:
# - Required run rate >= 8
# - Only 1–3 wickets remain
# - Death over

pressure_factors = [
    "high_required_run_rate",
    "low_wickets_remaining",
    "is_death_over"
]

revised_batting_pressure["second_innings_pressure_factors"] = (
    revised_batting_pressure[pressure_factors]
    .fillna(False)
    .astype(int)
    .sum(axis=1)
)

revised_batting_pressure["second_innings_pressure"] = (
    revised_batting_pressure["innings"].eq(2)
    & revised_batting_pressure[
        "second_innings_pressure_factors"
    ].ge(2)
)

# 11. Final batting pressure classification
revised_batting_pressure["is_under_pressure"] = (
    revised_batting_pressure["first_innings_pressure"]
    | revised_batting_pressure["second_innings_pressure"]
)

# 12. Summary
print("REVISED BATTING PRESSURE SUMMARY")
print("---------------------------------")

print("Total deliveries:", len(revised_batting_pressure))

print(
    "First innings pressure deliveries:",
    revised_batting_pressure["first_innings_pressure"].sum()
)

print(
    "Second innings pressure deliveries:",
    revised_batting_pressure["second_innings_pressure"].sum()
)

print(
    "Total under-pressure deliveries:",
    revised_batting_pressure["is_under_pressure"].sum()
)

print("\nPressure classification distribution:")

display(
    revised_batting_pressure.groupby(
        ["innings", "is_under_pressure"],
        dropna=False
    ).size().reset_index(name="deliveries")
)

print("\nSecond innings pressure factor combinations:")

display(
    revised_batting_pressure[
        revised_batting_pressure["innings"].eq(2)
    ]
    .groupby("second_innings_pressure_factors")
    .size()
    .reset_index(name="deliveries")
)

# 13. Data validation
print("\nDATA VALIDATION")
print("---------------------------------")

print("Missing pressure classifications:",
      revised_batting_pressure["is_under_pressure"].isna().sum())

print("Duplicate delivery keys:",
      revised_batting_pressure.duplicated(
          ["match_id", "innings", "innings_ball_num"]
      ).sum())

print("Unknown delivery legality:",
      revised_batting_pressure["is_legal_delivery"].isna().sum())

print("Missing chase targets in second innings:",
      revised_batting_pressure.loc[
          revised_batting_pressure["innings"].eq(2),
          "target_for_chase"
      ].isna().sum())

print("Negative remaining legal balls:",
      revised_batting_pressure["legal_balls_remaining"].lt(0).sum())
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 23
# Aggregate batting under-pressure statistics
# ============================================================

# 1. Select only under-pressure deliveries
revised_high_pressure_deliveries = (
    revised_batting_pressure[
        revised_batting_pressure["is_under_pressure"]
    ].copy()
)

# 2. Identify batter balls faced
# Legal deliveries and confirmed no-balls count as balls faced.
# Wides are excluded.
revised_high_pressure_deliveries["is_batter_ball_faced"] = (
    revised_high_pressure_deliveries["is_legal_delivery"].eq(True)
    | revised_high_pressure_deliveries["delivery_legality"].eq("No-ball")
)

# 3. Aggregate by batter and innings
revised_batting_under_pressure = (
    revised_high_pressure_deliveries.groupby(
        [
            "match_id",
            "innings",
            "batting_team_id",
            "striker_id",
            "striker_name"
        ],
        as_index=False
    )
    .agg(
        under_pressure_runs=("batter_runs", "sum"),
        under_pressure_balls_faced=("is_batter_ball_faced", "sum"),
        under_pressure_fours=("is_four", "sum"),
        under_pressure_sixes=("is_six", "sum"),
        under_pressure_deliveries=("innings_ball_num", "count")
    )
)

# 4. Calculate strike rate
revised_batting_under_pressure["under_pressure_strike_rate"] = (
    revised_batting_under_pressure["under_pressure_runs"]
    * 100
    / revised_batting_under_pressure["under_pressure_balls_faced"].replace(0, np.nan)
)

revised_batting_under_pressure[
    "under_pressure_strike_rate"
] = revised_batting_under_pressure[
    "under_pressure_strike_rate"
].round(2)

# 5. Summary
print("REVISED BATTING UNDER-PRESSURE SUMMARY")
print("---------------------------------")

print("Records:", len(revised_batting_under_pressure))
print("Unique batters:", revised_batting_under_pressure["striker_id"].nunique())
print("Matches:", revised_batting_under_pressure["match_id"].nunique())

print("\nTotal under-pressure statistics:")
print(
    revised_batting_under_pressure[
        [
            "under_pressure_runs",
            "under_pressure_balls_faced",
            "under_pressure_fours",
            "under_pressure_sixes",
            "under_pressure_deliveries"
        ]
    ].sum()
)

print("\nTop 10 revised under-pressure run scorers:")
display(
    revised_batting_under_pressure.groupby(
        ["striker_id", "striker_name"],
        as_index=False
    )
    .agg(
        under_pressure_runs=("under_pressure_runs", "sum"),
        under_pressure_balls=("under_pressure_balls_faced", "sum"),
        under_pressure_fours=("under_pressure_fours", "sum"),
        under_pressure_sixes=("under_pressure_sixes", "sum")
    )
    .sort_values(
        ["under_pressure_runs", "under_pressure_balls"],
        ascending=[False, False]
    )
    .head(10)
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 24
# Validate revised batting under-pressure statistics
# ============================================================

# 1. Calculate expected totals directly from pressure deliveries
expected_totals = pd.Series({
    "runs": revised_high_pressure_deliveries["batter_runs"].sum(),
    "balls_faced": revised_high_pressure_deliveries["is_batter_ball_faced"].sum(),
    "fours": revised_high_pressure_deliveries["is_four"].sum(),
    "sixes": revised_high_pressure_deliveries["is_six"].sum(),
    "deliveries": len(revised_high_pressure_deliveries)
})

# 2. Calculate actual totals from aggregated batting statistics
actual_totals = pd.Series({
    "runs": revised_batting_under_pressure["under_pressure_runs"].sum(),
    "balls_faced": revised_batting_under_pressure["under_pressure_balls_faced"].sum(),
    "fours": revised_batting_under_pressure["under_pressure_fours"].sum(),
    "sixes": revised_batting_under_pressure["under_pressure_sixes"].sum(),
    "deliveries": revised_batting_under_pressure["under_pressure_deliveries"].sum()
})

# 3. Compare expected and actual totals
validation_summary = pd.DataFrame({
    "expected": expected_totals,
    "actual": actual_totals
})

validation_summary["match"] = (
    validation_summary["expected"] == validation_summary["actual"]
)

print("REVISED BATTING PRESSURE VALIDATION")
print("---------------------------------")
display(validation_summary)

# 4. Validate individual batter-innings records
group_cols = [
    "match_id",
    "innings",
    "batting_team_id",
    "striker_id",
    "striker_name"
]

expected_records = (
    revised_high_pressure_deliveries.groupby(
        group_cols,
        as_index=False
    )
    .agg(
        expected_runs=("batter_runs", "sum"),
        expected_balls=("is_batter_ball_faced", "sum"),
        expected_fours=("is_four", "sum"),
        expected_sixes=("is_six", "sum"),
        expected_deliveries=("innings_ball_num", "count")
    )
)

actual_records = revised_batting_under_pressure[
    group_cols + [
        "under_pressure_runs",
        "under_pressure_balls_faced",
        "under_pressure_fours",
        "under_pressure_sixes",
        "under_pressure_deliveries"
    ]
].copy()

record_comparison = expected_records.merge(
    actual_records,
    on=group_cols,
    how="outer",
    indicator=True,
    validate="one_to_one"
)

record_comparison["record_matches"] = (
    record_comparison["_merge"].eq("both")
    & record_comparison["expected_runs"].eq(
        record_comparison["under_pressure_runs"]
    )
    & record_comparison["expected_balls"].eq(
        record_comparison["under_pressure_balls_faced"]
    )
    & record_comparison["expected_fours"].eq(
        record_comparison["under_pressure_fours"]
    )
    & record_comparison["expected_sixes"].eq(
        record_comparison["under_pressure_sixes"]
    )
    & record_comparison["expected_deliveries"].eq(
        record_comparison["under_pressure_deliveries"]
    )
)

print("\nINDIVIDUAL BATTER RECORD VALIDATION")
print("---------------------------------")
print("Expected records:", len(expected_records))
print("Actual records:", len(actual_records))
print("Mismatched records:", (~record_comparison["record_matches"]).sum())

if (~record_comparison["record_matches"]).any():
    display(
        record_comparison[
            ~record_comparison["record_matches"]
        ]
    )

# 5. Duplicate and missing-value checks
print("\nDATA QUALITY CHECKS")
print("---------------------------------")

print(
    "Duplicate batter-innings records:",
    revised_batting_under_pressure.duplicated(group_cols).sum()
)

print(
    "Missing batter IDs:",
    revised_batting_under_pressure["striker_id"].isna().sum()
)

print(
    "Missing runs:",
    revised_batting_under_pressure["under_pressure_runs"].isna().sum()
)

print(
    "Missing balls faced:",
    revised_batting_under_pressure["under_pressure_balls_faced"].isna().sum()
)

# 6. Final validation status
validation_passed = (
    validation_summary["match"].all()
    and record_comparison["record_matches"].all()
    and revised_batting_under_pressure.duplicated(group_cols).sum() == 0
)

print("\nFINAL VALIDATION STATUS")
print("---------------------------------")
print("PASSED" if validation_passed else "FAILED")
```

```python
# ============================================================
# REVISED MATCH ANALYSIS - STEP 25
# Create revised bowling-pressure classification
# ============================================================

# 1. Start with the existing revised pressure context.
# Preserve the original batting-pressure classification separately.

revised_bowling_pressure = revised_batting_pressure.copy()

revised_bowling_pressure["is_batting_under_pressure"] = (
    revised_bowling_pressure["is_under_pressure"].copy()
)

# 2. Factor 1:
# Defending a first-innings total below 120.
# Applicable only when bowling in the second innings.

revised_bowling_pressure["defending_low_total"] = (
    revised_bowling_pressure["innings"].eq(2)
    & revised_bowling_pressure["first_innings_runs"].lt(120)
).fillna(False)

# 3. Factor 2:
# Opposition has 7–10 wickets remaining.

revised_bowling_pressure[
    "opposition_7_10_wickets_remaining"
] = (
    revised_bowling_pressure["wickets_remaining"].between(7, 10)
).fillna(False)

# 4. Factor 3:
# Bowling during death overs.

revised_bowling_pressure["death_over_pressure"] = (
    revised_bowling_pressure["is_death_over"].eq(True)
).fillna(False)

# 5. Calculate total bowling-pressure factors.

bowling_factor_columns = [
    "defending_low_total",
    "opposition_7_10_wickets_remaining",
    "death_over_pressure"
]

revised_bowling_pressure["bowling_pressure_factors"] = (
    revised_bowling_pressure[bowling_factor_columns]
    .sum(axis=1)
    .astype(int)
)

# 6. Classify bowling-pressure situation.

revised_bowling_pressure["bowling_pressure_situation"] = (
    revised_bowling_pressure["bowling_pressure_factors"]
    .map({
        0: "Normal",
        1: "Moderate",
        2: "High",
        3: "Very High"
    })
)

# 7. Final bowling-pressure flag.

revised_bowling_pressure["is_under_pressure"] = (
    revised_bowling_pressure["bowling_pressure_factors"].ge(1)
)

# 8. Summary.

print("REVISED BOWLING PRESSURE SUMMARY")
print("---------------------------------")

print("Total deliveries:", len(revised_bowling_pressure))

print("\nBowling pressure situations:")
display(
    revised_bowling_pressure[
        "bowling_pressure_situation"
    ].value_counts().rename_axis("situation").to_frame("deliveries")
)

print(
    "\nUnder-pressure deliveries:",
    revised_bowling_pressure["is_under_pressure"].sum()
)

print(
    "Normal deliveries:",
    (~revised_bowling_pressure["is_under_pressure"]).sum()
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 26
# Aggregate revised bowling under-pressure statistics
# ============================================================

# 1. Filter under-pressure deliveries
revised_high_pressure_bowling = (
    revised_bowling_pressure[
        revised_bowling_pressure["is_under_pressure"]
    ].copy()
)

# 2. Calculate legal balls bowled
# Wide and no-ball deliveries are not legal bowling balls.
revised_high_pressure_bowling["is_legal_bowling_ball"] = (
    revised_high_pressure_bowling["is_legal_delivery"]
    .eq(True)
)

# 3. Calculate bowler runs conceded
# Batter runs are included.
# Wide and no-ball penalties (extras1_runs) are included.
# Byes and leg-byes are excluded.
revised_high_pressure_bowling["bowler_runs_conceded"] = (
    revised_high_pressure_bowling["batter_runs"].fillna(0)
    + np.where(
        revised_high_pressure_bowling["delivery_legality"].isin(
            ["Wide", "No-ball"]
        ),
        revised_high_pressure_bowling["extras1_runs"].fillna(0),
        0
    )
)

# 4. Calculate bowler-credited wickets
revised_high_pressure_bowling["bowler_wicket"] = (
    revised_high_pressure_bowling["is_wicket"].eq(1)
    & revised_high_pressure_bowling["dismissal_type"].isin(
        bowler_credited_dismissals
    )
).astype(int)

# 5. Group by match, innings and bowler
bowling_group_cols = [
    "match_id",
    "innings",
    "bowling_team_id",
    "bowler_id",
    "bowler_name"
]

revised_bowling_under_pressure = (
    revised_high_pressure_bowling.groupby(
        bowling_group_cols,
        as_index=False
    )
    .agg(
        under_pressure_legal_balls=(
            "is_legal_bowling_ball", "sum"
        ),
        under_pressure_runs=(
            "bowler_runs_conceded", "sum"
        ),
        under_pressure_wickets=(
            "bowler_wicket", "sum"
        ),
        under_pressure_deliveries=(
            "innings_ball_num", "count"
        )
    )
)

# 6. Calculate bowling metrics
revised_bowling_under_pressure["under_pressure_overs"] = (
    revised_bowling_under_pressure[
        "under_pressure_legal_balls"
    ] / 6
)

revised_bowling_under_pressure["under_pressure_economy"] = (
    revised_bowling_under_pressure["under_pressure_runs"]
    * 6
    / revised_bowling_under_pressure[
        "under_pressure_legal_balls"
    ].replace(0, np.nan)
).round(2)

revised_bowling_under_pressure["under_pressure_average"] = (
    revised_bowling_under_pressure["under_pressure_runs"]
    / revised_bowling_under_pressure[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

revised_bowling_under_pressure["under_pressure_strike_rate"] = (
    revised_bowling_under_pressure["under_pressure_legal_balls"]
    / revised_bowling_under_pressure[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

# 7. Summary
print("REVISED BOWLING UNDER-PRESSURE SUMMARY")
print("---------------------------------")

print(
    "Bowling records:",
    len(revised_bowling_under_pressure)
)

print(
    "Unique bowlers:",
    revised_bowling_under_pressure["bowler_id"].nunique()
)

print(
    "Matches:",
    revised_bowling_under_pressure["match_id"].nunique()
)

print(
    "Under-pressure deliveries:",
    revised_bowling_under_pressure[
        "under_pressure_deliveries"
    ].sum()
)

print(
    "Legal balls:",
    revised_bowling_under_pressure[
        "under_pressure_legal_balls"
    ].sum()
)

print(
    "Runs conceded:",
    revised_bowling_under_pressure[
        "under_pressure_runs"
    ].sum()
)

print(
    "Bowler-credited wickets:",
    revised_bowling_under_pressure[
        "under_pressure_wickets"
    ].sum()
)

# 8. Top 10 bowlers by pressure wickets
print("\nTOP 10 REVISED BOWLERS UNDER PRESSURE")

display(
    revised_bowling_under_pressure
    .groupby(
        ["bowler_id", "bowler_name"],
        as_index=False
    )
    .agg(
        under_pressure_wickets=(
            "under_pressure_wickets", "sum"
        ),
        under_pressure_runs=(
            "under_pressure_runs", "sum"
        ),
        under_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        )
    )
    .sort_values(
        ["under_pressure_wickets", "under_pressure_runs"],
        ascending=[False, True]
    )
    .head(10)
    .reset_index(drop=True)
)

# 9. Data quality checks
print("\nDATA QUALITY CHECKS")
print("---------------------------------")

print(
    "Duplicate bowler-innings records:",
    revised_bowling_under_pressure.duplicated(
        bowling_group_cols
    ).sum()
)

print(
    "Missing bowler IDs:",
    revised_bowling_under_pressure["bowler_id"].isna().sum()
)

print(
    "Missing pressure runs:",
    revised_bowling_under_pressure[
        "under_pressure_runs"
    ].isna().sum()
)

print(
    "Missing legal balls:",
    revised_bowling_under_pressure[
        "under_pressure_legal_balls"
    ].isna().sum()
)
```

```python

# ============================================================
# REVISED MATCH ANALYSIS - STEP 27
# Validate revised bowling under-pressure aggregation
# ============================================================

# 1. Expected totals directly from delivery-level data
expected_totals = pd.Series({
    "deliveries": len(revised_high_pressure_bowling),
    "legal_balls": revised_high_pressure_bowling[
        "is_legal_bowling_ball"
    ].sum(),
    "runs": revised_high_pressure_bowling[
        "bowler_runs_conceded"
    ].sum(),
    "wickets": revised_high_pressure_bowling[
        "bowler_wicket"
    ].sum()
})

# 2. Actual totals from aggregated bowler records
actual_totals = pd.Series({
    "deliveries": revised_bowling_under_pressure[
        "under_pressure_deliveries"
    ].sum(),
    "legal_balls": revised_bowling_under_pressure[
        "under_pressure_legal_balls"
    ].sum(),
    "runs": revised_bowling_under_pressure[
        "under_pressure_runs"
    ].sum(),
    "wickets": revised_bowling_under_pressure[
        "under_pressure_wickets"
    ].sum()
})

# 3. Compare expected and actual totals
validation_summary = pd.DataFrame({
    "expected": expected_totals,
    "actual": actual_totals
})

validation_summary["match"] = (
    validation_summary["expected"]
    == validation_summary["actual"]
)

print("REVISED BOWLING PRESSURE VALIDATION")
print("---------------------------------")

display(validation_summary)

# 4. Validate individual bowler-innings records
bowling_group_cols = [
    "match_id",
    "innings",
    "bowling_team_id",
    "bowler_id",
    "bowler_name"
]

expected_records = (
    revised_high_pressure_bowling.groupby(
        bowling_group_cols,
        as_index=False
    )
    .agg(
        expected_legal_balls=(
            "is_legal_bowling_ball", "sum"
        ),
        expected_runs=(
            "bowler_runs_conceded", "sum"
        ),
        expected_wickets=(
            "bowler_wicket", "sum"
        ),
        expected_deliveries=(
            "innings_ball_num", "count"
        )
    )
)

actual_records = revised_bowling_under_pressure[
    bowling_group_cols + [
        "under_pressure_legal_balls",
        "under_pressure_runs",
        "under_pressure_wickets",
        "under_pressure_deliveries"
    ]
].copy()

record_comparison = expected_records.merge(
    actual_records,
    on=bowling_group_cols,
    how="outer",
    indicator=True,
    validate="one_to_one"
)

record_comparison["record_matches"] = (
    record_comparison["_merge"].eq("both")
    & record_comparison["expected_legal_balls"].eq(
        record_comparison["under_pressure_legal_balls"]
    )
    & record_comparison["expected_runs"].eq(
        record_comparison["under_pressure_runs"]
    )
    & record_comparison["expected_wickets"].eq(
        record_comparison["under_pressure_wickets"]
    )
    & record_comparison["expected_deliveries"].eq(
        record_comparison["under_pressure_deliveries"]
    )
)

print("\nINDIVIDUAL BOWLER RECORD VALIDATION")
print("---------------------------------")

print("Expected records:", len(expected_records))
print("Actual records:", len(actual_records))

print(
    "Mismatched records:",
    (~record_comparison["record_matches"]).sum()
)

if (~record_comparison["record_matches"]).any():
    display(
        record_comparison[
            ~record_comparison["record_matches"]
        ]
    )

# 5. Final validation
validation_passed = (
    validation_summary["match"].all()
    and record_comparison["record_matches"].all()
    and revised_bowling_under_pressure.duplicated(
        bowling_group_cols
    ).sum() == 0
)

print("\nFINAL VALIDATION STATUS")
print("---------------------------------")
print("PASSED" if validation_passed else "FAILED")

```

```python

# ============================================================
# COMBINED ANALYSIS - STEP 29
# Create combined batting performance dataset
# ============================================================

import pandas as pd
import numpy as np

# Common match metadata
metadata_cols = [
    "match_id",
    "is_knockout",
    "stage",
    "stage_priority",
    "match_date",
    "result_category",
    "comp_id",
    "comp_name",
    "match_format",
    "season"
]

# ============================================================
# 1. STANDARD BATTING
# ============================================================

standard_batting_combined = batter_match_analysis.copy()

standard_batting_combined = standard_batting_combined.rename(
    columns={
        "striker_id": "player_id",
        "striker_name": "player_name",
        "batting_team_id": "team_id"
    }
)

standard_batting_combined["match_type"] = "STANDARD"

# ============================================================
# 2. REVISED BATTING
# ============================================================

revised_batting_combined = revised_batting.merge(
    revised_batting_under_pressure,
    on=[
        "match_id",
        "innings",
        "batting_team_id",
        "striker_id",
        "striker_name"
    ],
    how="left",
    validate="one_to_one"
)

# Rename overall batting statistics
revised_batting_combined = revised_batting_combined.rename(
    columns={
        "striker_id": "player_id",
        "striker_name": "player_name",
        "batting_team_id": "team_id",
        "runs": "overall_runs",
        "balls_faced": "overall_balls_faced",
        "fours": "overall_fours",
        "sixes": "overall_sixes",
        "strike_rate": "overall_strike_rate"
    }
)

# Fill missing pressure counting statistics with zero
revised_pressure_count_cols = [
    "under_pressure_runs",
    "under_pressure_balls_faced",
    "under_pressure_fours",
    "under_pressure_sixes"
]

revised_batting_combined[
    revised_pressure_count_cols
] = revised_batting_combined[
    revised_pressure_count_cols
].fillna(0)

# Recalculate pressure strike rate
revised_batting_combined["under_pressure_strike_rate"] = (
    revised_batting_combined["under_pressure_runs"]
    * 100
    / revised_batting_combined[
        "under_pressure_balls_faced"
    ].replace(0, np.nan)
).round(2)

revised_batting_combined["match_type"] = "REVISED"

# Attach match metadata
revised_batting_combined = revised_batting_combined.merge(
    match_dashboard_context[metadata_cols],
    on="match_id",
    how="left",
    validate="many_to_one"
)

# ============================================================
# 3. SUPER OVER BATTING
# ============================================================

super_over_batting_combined = super_over_batting.copy()

super_over_batting_combined = (
    super_over_batting_combined.rename(
        columns={
            "striker_id": "player_id",
            "striker_name": "player_name",
            "batting_team_id": "team_id",
            "runs": "overall_runs",
            "balls_faced": "overall_balls_faced",
            "fours": "overall_fours",
            "sixes": "overall_sixes",
            "strike_rate": "overall_strike_rate"
        }
    )
)

# Recalculate overall strike rate.
# Zero balls faced means undefined strike rate.
super_over_batting_combined["overall_strike_rate"] = (
    super_over_batting_combined["overall_runs"]
    * 100
    / super_over_batting_combined[
        "overall_balls_faced"
    ].replace(0, np.nan)
).round(2)

# Every recorded Super Over performance counts as under pressure.
super_over_batting_combined["under_pressure_runs"] = (
    super_over_batting_combined["overall_runs"]
)

super_over_batting_combined["under_pressure_balls_faced"] = (
    super_over_batting_combined["overall_balls_faced"]
)

super_over_batting_combined["under_pressure_fours"] = (
    super_over_batting_combined["overall_fours"]
)

super_over_batting_combined["under_pressure_sixes"] = (
    super_over_batting_combined["overall_sixes"]
)

super_over_batting_combined["under_pressure_strike_rate"] = (
    super_over_batting_combined["overall_strike_rate"]
)

super_over_batting_combined["match_type"] = "SUPER_OVER"

super_over_batting_combined = (
    super_over_batting_combined.merge(
        match_dashboard_context[metadata_cols],
        on="match_id",
        how="left",
        validate="many_to_one"
    )
)

# ============================================================
# 4. SELECT COMMON COLUMNS
# ============================================================

combined_batting_columns = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "player_name",
    "match_type",
    "overall_runs",
    "overall_balls_faced",
    "overall_fours",
    "overall_sixes",
    "overall_strike_rate",
    "under_pressure_runs",
    "under_pressure_balls_faced",
    "under_pressure_fours",
    "under_pressure_sixes",
    "under_pressure_strike_rate",
    "is_knockout",
    "stage",
    "stage_priority",
    "match_date",
    "result_category",
    "comp_id",
    "comp_name",
    "match_format",
    "season"
]

# Align all three datasets
standard_batting_final = standard_batting_combined[
    combined_batting_columns
].copy()

revised_batting_final = revised_batting_combined[
    combined_batting_columns
].copy()

super_over_batting_final = super_over_batting_combined[
    combined_batting_columns
].copy()

# ============================================================
# 5. COMBINE ALL BATTING PERFORMANCES
# ============================================================

combined_batting = pd.concat(
    [
        standard_batting_final,
        revised_batting_final,
        super_over_batting_final
    ],
    ignore_index=True
)

# ============================================================
# 6. SUMMARY
# ============================================================

print("COMBINED BATTING DATASET")
print("---------------------------------")

print("Total records:", len(combined_batting))

print(
    "Unique players:",
    combined_batting["player_id"].nunique()
)

print(
    "Unique matches:",
    combined_batting["match_id"].nunique()
)

print("\nRecords by match type:")

display(
    combined_batting["match_type"]
    .value_counts()
    .rename_axis("match_type")
    .reset_index(name="records")
)

print("\nOverall batting totals:")

display(
    combined_batting.groupby("match_type")[
        [
            "overall_runs",
            "overall_balls_faced",
            "overall_fours",
            "overall_sixes"
        ]
    ].sum()
)

print("\nUnder-pressure batting totals:")

display(
    combined_batting.groupby("match_type")[
        [
            "under_pressure_runs",
            "under_pressure_balls_faced",
            "under_pressure_fours",
            "under_pressure_sixes"
        ]
    ].sum()
)

print("\nCombined dataset sample:")

display(combined_batting.head())

print("\nDataset shape:", combined_batting.shape)
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 30
# Validate combined batting dataset
# ============================================================

import pandas as pd
import numpy as np

print("COMBINED BATTING VALIDATION")
print("=" * 50)

# ------------------------------------------------------------
# 1. CHECK DUPLICATE PLAYER-INNINGS RECORDS
# ------------------------------------------------------------

duplicate_key = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "match_type"
]

duplicate_records = combined_batting[
    combined_batting.duplicated(
        subset=duplicate_key,
        keep=False
    )
]

print("\n1. Duplicate player-innings records")
print("Duplicate rows:", len(duplicate_records))
print("Duplicate keys:", duplicate_records[duplicate_key].drop_duplicates().shape[0])

# ------------------------------------------------------------
# 2. CHECK MISSING IDENTIFIERS
# ------------------------------------------------------------

identifier_cols = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "player_name",
    "match_type"
]

print("\n2. Missing identifiers")

display(
    combined_batting[identifier_cols]
    .isna()
    .sum()
    .rename("missing_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 3. CHECK MISSING PERFORMANCE METRICS
# ------------------------------------------------------------

count_metric_cols = [
    "overall_runs",
    "overall_balls_faced",
    "overall_fours",
    "overall_sixes",
    "under_pressure_runs",
    "under_pressure_balls_faced",
    "under_pressure_fours",
    "under_pressure_sixes"
]

print("\n3. Missing counting metrics")

display(
    combined_batting[count_metric_cols]
    .isna()
    .sum()
    .rename("missing_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 4. CHECK NEGATIVE VALUES
# ------------------------------------------------------------

negative_counts = (
    combined_batting[count_metric_cols] < 0
).sum()

print("\n4. Negative metric values")

display(
    negative_counts
    .rename("negative_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 5. CHECK PRESSURE METRICS AGAINST OVERALL METRICS
# ------------------------------------------------------------

pressure_checks = {
    "Runs": (
        combined_batting["under_pressure_runs"]
        <= combined_batting["overall_runs"]
    ),
    "Balls faced": (
        combined_batting["under_pressure_balls_faced"]
        <= combined_batting["overall_balls_faced"]
    ),
    "Fours": (
        combined_batting["under_pressure_fours"]
        <= combined_batting["overall_fours"]
    ),
    "Sixes": (
        combined_batting["under_pressure_sixes"]
        <= combined_batting["overall_sixes"]
    )
}

print("\n5. Under-pressure metrics consistency")

for metric, condition in pressure_checks.items():
    print(
        f"{metric}:",
        "PASS" if condition.all() else "FAIL",
        "| Failed rows:",
        int((~condition).sum())
    )

# ------------------------------------------------------------
# 6. VERIFY SOURCE TOTALS
# ------------------------------------------------------------

expected_totals = {
    "STANDARD": {
        "overall_runs": 260459,
        "overall_balls_faced": 186829,
        "overall_fours": 22518,
        "overall_sixes": 12341,
        "under_pressure_runs": 27914,
        "under_pressure_balls_faced": 19861,
        "under_pressure_fours": 2071,
        "under_pressure_sixes": 1587
    },
    "REVISED": {
        "overall_runs": 6385,
        "overall_balls_faced": 4248,
        "overall_fours": 558,
        "overall_sixes": 332,
        "under_pressure_runs": 655,
        "under_pressure_balls_faced": 426,
        "under_pressure_fours": 54,
        "under_pressure_sixes": 38
    },
    "SUPER_OVER": {
        "overall_runs": 44,
        "overall_balls_faced": 33,
        "overall_fours": 6,
        "overall_sixes": 1,
        "under_pressure_runs": 44,
        "under_pressure_balls_faced": 33,
        "under_pressure_fours": 6,
        "under_pressure_sixes": 1
    }
}

actual_totals = (
    combined_batting
    .groupby("match_type")[count_metric_cols]
    .sum()
)

validation_rows = []

for match_type, expected in expected_totals.items():

    for metric, expected_value in expected.items():

        actual_value = actual_totals.loc[
            match_type, metric
        ]

        validation_rows.append({
            "match_type": match_type,
            "metric": metric,
            "expected": expected_value,
            "actual": actual_value,
            "passed": np.isclose(
                expected_value,
                actual_value
            )
        })

totals_validation = pd.DataFrame(validation_rows)

print("\n6. Source totals reconciliation")

display(totals_validation)

print(
    "Total mismatches:",
    int((~totals_validation["passed"]).sum())
)

# ------------------------------------------------------------
# 7. VERIFY SUPER OVER PRESSURE RULE
# ------------------------------------------------------------

super_over_data = combined_batting[
    combined_batting["match_type"] == "SUPER_OVER"
]

super_over_checks = {
    "Runs": (
        super_over_data["overall_runs"]
        == super_over_data["under_pressure_runs"]
    ),
    "Balls faced": (
        super_over_data["overall_balls_faced"]
        == super_over_data["under_pressure_balls_faced"]
    ),
    "Fours": (
        super_over_data["overall_fours"]
        == super_over_data["under_pressure_fours"]
    ),
    "Sixes": (
        super_over_data["overall_sixes"]
        == super_over_data["under_pressure_sixes"]
    )
}

print("\n7. Super Over pressure validation")

for metric, condition in super_over_checks.items():
    print(
        f"{metric}:",
        "PASS" if condition.all() else "FAIL"
    )

# ------------------------------------------------------------
# 8. FINAL SUMMARY
# ------------------------------------------------------------

print("\n8. FINAL VALIDATION SUMMARY")
print("-" * 50)

print("Total records:", len(combined_batting))
print("Unique players:", combined_batting["player_id"].nunique())
print("Unique matches:", combined_batting["match_id"].nunique())

print(
    "Duplicate player-innings rows:",
    len(duplicate_records)
)

print(
    "Missing essential identifiers:",
    int(combined_batting[identifier_cols].isna().sum().sum())
)

print(
    "Missing counting metrics:",
    int(combined_batting[count_metric_cols].isna().sum().sum())
)

print(
    "Source total mismatches:",
    int((~totals_validation["passed"]).sum())
)

print("\nValidation cell completed.")
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 31
# Create combined bowling performance dataset
# ============================================================

import pandas as pd
import numpy as np

metadata_cols = [
    "match_id",
    "is_knockout",
    "stage",
    "stage_priority",
    "match_date",
    "result_category",
    "comp_id",
    "comp_name",
    "match_format",
    "season"
]

# ============================================================
# 1. STANDARD BOWLING
# ============================================================

standard_bowling_combined = bowler_match_analysis.copy()

standard_bowling_combined = standard_bowling_combined.rename(
    columns={
        "bowling_team_id": "team_id",
        "bowler_id": "player_id",
        "bowler_name": "player_name",
        "legal_balls": "overall_legal_balls",
        "total_runs_conceded": "overall_runs_conceded",
        "wickets": "overall_wickets",
        "overs": "overall_overs",
        "economy": "overall_economy",
        "under_pressure_legal_balls": "under_pressure_legal_balls",
        "under_pressure_runs_conceded": "under_pressure_runs_conceded",
        "under_pressure_wickets": "under_pressure_wickets",
        "under_pressure_overs": "under_pressure_overs",
        "under_pressure_economy": "under_pressure_economy",
        "under_pressure_bowling_average": "under_pressure_bowling_average",
        "under_pressure_bowling_strike_rate": "under_pressure_bowling_strike_rate"
    }
)

standard_bowling_combined["match_type"] = "STANDARD"

# ============================================================
# 2. REVISED BOWLING
# ============================================================

revised_bowling_combined = revised_bowling.merge(
    revised_bowling_under_pressure,
    on=[
        "match_id",
        "innings",
        "bowling_team_id",
        "bowler_id",
        "bowler_name"
    ],
    how="left",
    validate="one_to_one"
)

revised_bowling_combined = revised_bowling_combined.rename(
    columns={
        "bowling_team_id": "team_id",
        "bowler_id": "player_id",
        "bowler_name": "player_name",
        "legal_balls": "overall_legal_balls",
        "runs_conceded": "overall_runs_conceded",
        "wickets": "overall_wickets",
        "overs": "overall_overs",
        "economy": "overall_economy",
        "under_pressure_runs": "under_pressure_runs_conceded",
        "under_pressure_average": "under_pressure_bowling_average",
        "under_pressure_strike_rate": "under_pressure_bowling_strike_rate"
    }
)

# Fill missing pressure counting metrics with zero
revised_pressure_count_cols = [
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets"
]

revised_bowling_combined[
    revised_pressure_count_cols
] = revised_bowling_combined[
    revised_pressure_count_cols
].fillna(0)

# Recalculate pressure overs
revised_bowling_combined["under_pressure_overs"] = (
    revised_bowling_combined["under_pressure_legal_balls"] / 6
).round(2)

# Recalculate pressure economy
revised_bowling_combined["under_pressure_economy"] = (
    revised_bowling_combined["under_pressure_runs_conceded"] * 6
    / revised_bowling_combined[
        "under_pressure_legal_balls"
    ].replace(0, np.nan)
).round(2)

# Recalculate pressure bowling average
revised_bowling_combined["under_pressure_bowling_average"] = (
    revised_bowling_combined["under_pressure_runs_conceded"]
    / revised_bowling_combined[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

# Recalculate pressure bowling strike rate
revised_bowling_combined["under_pressure_bowling_strike_rate"] = (
    revised_bowling_combined["under_pressure_legal_balls"]
    / revised_bowling_combined[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

revised_bowling_combined["match_type"] = "REVISED"

# Attach match metadata
revised_bowling_combined = revised_bowling_combined.merge(
    match_dashboard_context[metadata_cols],
    on="match_id",
    how="left",
    validate="many_to_one"
)

# ============================================================
# 3. SUPER OVER BOWLING
# ============================================================

super_over_bowling_combined = super_over_bowling.copy()

super_over_bowling_combined = super_over_bowling_combined.rename(
    columns={
        "bowling_team_id": "team_id",
        "bowler_id": "player_id",
        "bowler_name": "player_name",
        "legal_balls": "overall_legal_balls",
        "runs_conceded": "overall_runs_conceded",
        "wickets": "overall_wickets",
        "overs": "overall_overs",
        "economy": "overall_economy"
    }
)

# Every recorded Super Over performance counts as under pressure.
super_over_bowling_combined["under_pressure_legal_balls"] = (
    super_over_bowling_combined["overall_legal_balls"]
)

super_over_bowling_combined["under_pressure_runs_conceded"] = (
    super_over_bowling_combined["overall_runs_conceded"]
)

super_over_bowling_combined["under_pressure_wickets"] = (
    super_over_bowling_combined["overall_wickets"]
)

super_over_bowling_combined["under_pressure_overs"] = (
    super_over_bowling_combined["overall_overs"]
)

super_over_bowling_combined["under_pressure_economy"] = (
    super_over_bowling_combined["overall_economy"]
)

super_over_bowling_combined["under_pressure_bowling_average"] = (
    super_over_bowling_combined["overall_runs_conceded"]
    / super_over_bowling_combined[
        "overall_wickets"
    ].replace(0, np.nan)
).round(2)

super_over_bowling_combined["under_pressure_bowling_strike_rate"] = (
    super_over_bowling_combined["overall_legal_balls"]
    / super_over_bowling_combined[
        "overall_wickets"
    ].replace(0, np.nan)
).round(2)

super_over_bowling_combined["match_type"] = "SUPER_OVER"

super_over_bowling_combined = super_over_bowling_combined.merge(
    match_dashboard_context[metadata_cols],
    on="match_id",
    how="left",
    validate="many_to_one"
)

# ============================================================
# 4. SELECT COMMON COLUMNS
# ============================================================

combined_bowling_columns = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "player_name",
    "match_type",
    "overall_legal_balls",
    "overall_runs_conceded",
    "overall_wickets",
    "overall_overs",
    "overall_economy",
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets",
    "under_pressure_overs",
    "under_pressure_economy",
    "under_pressure_bowling_average",
    "under_pressure_bowling_strike_rate",
    "is_knockout",
    "stage",
    "stage_priority",
    "match_date",
    "result_category",
    "comp_id",
    "comp_name",
    "match_format",
    "season"
]

standard_bowling_final = standard_bowling_combined[
    combined_bowling_columns
].copy()

revised_bowling_final = revised_bowling_combined[
    combined_bowling_columns
].copy()

super_over_bowling_final = super_over_bowling_combined[
    combined_bowling_columns
].copy()

# ============================================================
# 5. COMBINE ALL BOWLING PERFORMANCES
# ============================================================

combined_bowling = pd.concat(
    [
        standard_bowling_final,
        revised_bowling_final,
        super_over_bowling_final
    ],
    ignore_index=True
)

# ============================================================
# 6. SUMMARY
# ============================================================

print("COMBINED BOWLING DATASET")
print("---------------------------------")

print("Total records:", len(combined_bowling))

print(
    "Unique players:",
    combined_bowling["player_id"].nunique()
)

print(
    "Unique matches:",
    combined_bowling["match_id"].nunique()
)

print("\nRecords by match type:")

display(
    combined_bowling["match_type"]
    .value_counts()
    .rename_axis("match_type")
    .reset_index(name="records")
)

print("\nOverall bowling totals:")

display(
    combined_bowling.groupby("match_type")[
        [
            "overall_legal_balls",
            "overall_runs_conceded",
            "overall_wickets"
        ]
    ].sum()
)

print("\nUnder-pressure bowling totals:")

display(
    combined_bowling.groupby("match_type")[
        [
            "under_pressure_legal_balls",
            "under_pressure_runs_conceded",
            "under_pressure_wickets"
        ]
    ].sum()
)

print("\nCombined dataset sample:")

display(combined_bowling.head())

print("\nDataset shape:", combined_bowling.shape)
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 32
# Validate combined bowling dataset
# ============================================================

import pandas as pd
import numpy as np

print("COMBINED BOWLING VALIDATION")
print("=" * 50)

# ------------------------------------------------------------
# 1. CHECK DUPLICATE BOWLER-INNINGS RECORDS
# ------------------------------------------------------------

duplicate_key = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "match_type"
]

duplicate_records = combined_bowling[
    combined_bowling.duplicated(
        subset=duplicate_key,
        keep=False
    )
]

print("\n1. Duplicate bowler-innings records")
print("Duplicate rows:", len(duplicate_records))
print(
    "Duplicate keys:",
    duplicate_records[duplicate_key].drop_duplicates().shape[0]
)

# ------------------------------------------------------------
# 2. CHECK MISSING IDENTIFIERS
# ------------------------------------------------------------

identifier_cols = [
    "match_id",
    "innings",
    "team_id",
    "player_id",
    "player_name",
    "match_type"
]

print("\n2. Missing identifiers")

display(
    combined_bowling[identifier_cols]
    .isna()
    .sum()
    .rename("missing_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 3. CHECK MISSING COUNTING METRICS
# ------------------------------------------------------------

count_metric_cols = [
    "overall_legal_balls",
    "overall_runs_conceded",
    "overall_wickets",
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets"
]

print("\n3. Missing counting metrics")

display(
    combined_bowling[count_metric_cols]
    .isna()
    .sum()
    .rename("missing_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 4. CHECK NEGATIVE VALUES
# ------------------------------------------------------------

negative_counts = (
    combined_bowling[count_metric_cols] < 0
).sum()

print("\n4. Negative metric values")

display(
    negative_counts
    .rename("negative_count")
    .reset_index()
    .rename(columns={"index": "column"})
)

# ------------------------------------------------------------
# 5. CHECK UNDER-PRESSURE METRICS AGAINST OVERALL METRICS
# ------------------------------------------------------------

pressure_checks = {
    "Legal balls": (
        combined_bowling["under_pressure_legal_balls"]
        <= combined_bowling["overall_legal_balls"]
    ),
    "Runs conceded": (
        combined_bowling["under_pressure_runs_conceded"]
        <= combined_bowling["overall_runs_conceded"]
    ),
    "Wickets": (
        combined_bowling["under_pressure_wickets"]
        <= combined_bowling["overall_wickets"]
    )
}

print("\n5. Under-pressure metrics consistency")

for metric, condition in pressure_checks.items():
    print(
        f"{metric}:",
        "PASS" if condition.all() else "FAIL",
        "| Failed rows:",
        int((~condition).sum())
    )

# ------------------------------------------------------------
# 6. VERIFY SOURCE TOTALS
# ------------------------------------------------------------

expected_totals = {
    "STANDARD": {
        "overall_legal_balls": 186258,
        "overall_runs_conceded": 269893,
        "overall_wickets": 9867,
        "under_pressure_legal_balls": 158781,
        "under_pressure_runs_conceded": 236351,
        "under_pressure_wickets": 8550
    },
    "REVISED": {
        "overall_legal_balls": 4233,
        "overall_runs_conceded": 6605,
        "overall_wickets": 296,
        "under_pressure_legal_balls": 3833,
        "under_pressure_runs_conceded": 6085,
        "under_pressure_wickets": 274
    },
    "SUPER_OVER": {
        "overall_legal_balls": 33,
        "overall_runs_conceded": 52,
        "overall_wickets": 11,
        "under_pressure_legal_balls": 33,
        "under_pressure_runs_conceded": 52,
        "under_pressure_wickets": 11
    }
}

actual_totals = (
    combined_bowling
    .groupby("match_type")[count_metric_cols]
    .sum()
)

validation_rows = []

for match_type, expected in expected_totals.items():

    for metric, expected_value in expected.items():

        actual_value = actual_totals.loc[
            match_type, metric
        ]

        validation_rows.append({
            "match_type": match_type,
            "metric": metric,
            "expected": expected_value,
            "actual": actual_value,
            "passed": np.isclose(
                expected_value,
                actual_value
            )
        })

totals_validation = pd.DataFrame(validation_rows)

print("\n6. Source totals reconciliation")

display(totals_validation)

print(
    "Total mismatches:",
    int((~totals_validation["passed"]).sum())
)

# ------------------------------------------------------------
# 7. VERIFY SUPER OVER PRESSURE RULE
# ------------------------------------------------------------

super_over_data = combined_bowling[
    combined_bowling["match_type"] == "SUPER_OVER"
]

super_over_checks = {
    "Legal balls": (
        super_over_data["overall_legal_balls"]
        == super_over_data["under_pressure_legal_balls"]
    ),
    "Runs conceded": (
        super_over_data["overall_runs_conceded"]
        == super_over_data["under_pressure_runs_conceded"]
    ),
    "Wickets": (
        super_over_data["overall_wickets"]
        == super_over_data["under_pressure_wickets"]
    )
}

print("\n7. Super Over pressure validation")

for metric, condition in super_over_checks.items():
    print(
        f"{metric}:",
        "PASS" if condition.all() else "FAIL"
    )

# ------------------------------------------------------------
# 8. VERIFY BOWLING RATE CALCULATIONS
# ------------------------------------------------------------

# Economy = runs conceded * 6 / legal balls
valid_overall_balls = combined_bowling[
    "overall_legal_balls"
] > 0

expected_overall_economy = (
    combined_bowling.loc[valid_overall_balls, "overall_runs_conceded"]
    * 6
    / combined_bowling.loc[valid_overall_balls, "overall_legal_balls"]
).round(2)

overall_economy_check = np.isclose(
    expected_overall_economy,
    combined_bowling.loc[
        valid_overall_balls, "overall_economy"
    ],
    equal_nan=True
)

valid_pressure_balls = combined_bowling[
    "under_pressure_legal_balls"
] > 0

expected_pressure_economy = (
    combined_bowling.loc[
        valid_pressure_balls, "under_pressure_runs_conceded"
    ] * 6
    / combined_bowling.loc[
        valid_pressure_balls, "under_pressure_legal_balls"
    ]
).round(2)

pressure_economy_check = np.isclose(
    expected_pressure_economy,
    combined_bowling.loc[
        valid_pressure_balls, "under_pressure_economy"
    ],
    equal_nan=True
)

print("\n8. Economy rate validation")
print(
    "Overall economy:",
    "PASS" if overall_economy_check.all() else "FAIL",
    "| Mismatches:",
    int((~overall_economy_check).sum())
)

print(
    "Under-pressure economy:",
    "PASS" if pressure_economy_check.all() else "FAIL",
    "| Mismatches:",
    int((~pressure_economy_check).sum())
)

# ------------------------------------------------------------
# 9. FINAL SUMMARY
# ------------------------------------------------------------

print("\n9. FINAL VALIDATION SUMMARY")
print("-" * 50)

print("Total records:", len(combined_bowling))
print("Unique players:", combined_bowling["player_id"].nunique())
print("Unique matches:", combined_bowling["match_id"].nunique())

print("Duplicate bowler-innings rows:", len(duplicate_records))

print(
    "Missing essential identifiers:",
    int(combined_bowling[identifier_cols].isna().sum().sum())
)

print(
    "Missing counting metrics:",
    int(combined_bowling[count_metric_cols].isna().sum().sum())
)

print(
    "Source total mismatches:",
    int((~totals_validation["passed"]).sum())
)

print("\nValidation cell completed.")
```

```python

# ============================================================
# COMBINED ANALYSIS - STEP 33
# Player-level bowling statistics
# Regular career = STANDARD + REVISED
# SUPER_OVER statistics remain separate
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. KEEP ALL PLAYERS IN THE MASTER
# ------------------------------------------------------------

all_bowlers = combined_bowling[
    ["player_id", "player_name"]
].drop_duplicates()

# ------------------------------------------------------------
# 2. AGGREGATE REGULAR CAREER BOWLING
# STANDARD + REVISED ONLY
# ------------------------------------------------------------

regular_bowling = combined_bowling[
    combined_bowling["match_type"].isin(["STANDARD", "REVISED"])
].copy()

player_bowling_overall = (
    regular_bowling
    .groupby(["player_id", "player_name"], as_index=False)
    .agg(
        matches_played=("match_id", "nunique"),
        bowling_innings=("match_id", "count"),
        overall_legal_balls=("overall_legal_balls", "sum"),
        overall_runs_conceded=("overall_runs_conceded", "sum"),
        overall_wickets=("overall_wickets", "sum"),
        under_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        under_pressure_runs_conceded=(
            "under_pressure_runs_conceded", "sum"
        ),
        under_pressure_wickets=(
            "under_pressure_wickets", "sum"
        )
    )
)

# ------------------------------------------------------------
# 3. AGGREGATE SUPER OVER STATISTICS SEPARATELY
# ------------------------------------------------------------

super_over_bowling = combined_bowling[
    combined_bowling["match_type"] == "SUPER_OVER"
].copy()

player_super_over_bowling = (
    super_over_bowling
    .groupby(["player_id", "player_name"], as_index=False)
    .agg(
        super_over_matches=("match_id", "nunique"),
        super_over_innings=("match_id", "count"),
        super_over_legal_balls=("overall_legal_balls", "sum"),
        super_over_runs_conceded=("overall_runs_conceded", "sum"),
        super_over_wickets=("overall_wickets", "sum"),
        super_over_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        super_over_pressure_runs_conceded=(
            "under_pressure_runs_conceded", "sum"
        ),
        super_over_pressure_wickets=(
            "under_pressure_wickets", "sum"
        )
    )
)

# ------------------------------------------------------------
# 4. MERGE REGULAR + SUPER OVER STATISTICS
# ------------------------------------------------------------

player_bowling_master = (
    all_bowlers
    .merge(
        player_bowling_overall,
        on=["player_id", "player_name"],
        how="left",
        validate="one_to_one"
    )
    .merge(
        player_super_over_bowling,
        on=["player_id", "player_name"],
        how="left",
        validate="one_to_one"
    )
)

# Fill missing count metrics with zero
count_columns = [
    col for col in player_bowling_master.columns
    if col not in ["player_id", "player_name"]
]

player_bowling_master[count_columns] = (
    player_bowling_master[count_columns].fillna(0)
)

# ------------------------------------------------------------
# 5. CALCULATE REGULAR CAREER BOWLING RATES
# ------------------------------------------------------------

player_bowling_master["overall_overs"] = (
    player_bowling_master["overall_legal_balls"] / 6
).round(2)

player_bowling_master["overall_economy"] = (
    player_bowling_master["overall_runs_conceded"] * 6
    / player_bowling_master["overall_legal_balls"].replace(0, np.nan)
).round(2)

player_bowling_master["overall_bowling_average"] = (
    player_bowling_master["overall_runs_conceded"]
    / player_bowling_master["overall_wickets"].replace(0, np.nan)
).round(2)

player_bowling_master["overall_bowling_strike_rate"] = (
    player_bowling_master["overall_legal_balls"]
    / player_bowling_master["overall_wickets"].replace(0, np.nan)
).round(2)

# ------------------------------------------------------------
# 6. CALCULATE REGULAR UNDER-PRESSURE BOWLING RATES
# ------------------------------------------------------------

player_bowling_master["under_pressure_overs"] = (
    player_bowling_master["under_pressure_legal_balls"] / 6
).round(2)

player_bowling_master["under_pressure_economy"] = (
    player_bowling_master["under_pressure_runs_conceded"] * 6
    / player_bowling_master[
        "under_pressure_legal_balls"
    ].replace(0, np.nan)
).round(2)

player_bowling_master["under_pressure_bowling_average"] = (
    player_bowling_master["under_pressure_runs_conceded"]
    / player_bowling_master[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

player_bowling_master["under_pressure_bowling_strike_rate"] = (
    player_bowling_master["under_pressure_legal_balls"]
    / player_bowling_master[
        "under_pressure_wickets"
    ].replace(0, np.nan)
).round(2)

# ------------------------------------------------------------
# 7. CALCULATE SUPER OVER RATES SEPARATELY
# ------------------------------------------------------------

player_bowling_master["super_over_overs"] = (
    player_bowling_master["super_over_legal_balls"] / 6
).round(2)

player_bowling_master["super_over_economy"] = (
    player_bowling_master["super_over_runs_conceded"] * 6
    / player_bowling_master[
        "super_over_legal_balls"
    ].replace(0, np.nan)
).round(2)

player_bowling_master["super_over_bowling_average"] = (
    player_bowling_master["super_over_runs_conceded"]
    / player_bowling_master[
        "super_over_wickets"
    ].replace(0, np.nan)
).round(2)

player_bowling_master["super_over_bowling_strike_rate"] = (
    player_bowling_master["super_over_legal_balls"]
    / player_bowling_master[
        "super_over_wickets"
    ].replace(0, np.nan)
).round(2)

# ------------------------------------------------------------
# 8. MATCH-TYPE BREAKDOWN
# Keep STANDARD, REVISED, SUPER_OVER separately
# ------------------------------------------------------------

bowling_match_type_stats = (
    combined_bowling
    .groupby(
        ["player_id", "player_name", "match_type"],
        as_index=False
    )
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),
        legal_balls=("overall_legal_balls", "sum"),
        runs_conceded=("overall_runs_conceded", "sum"),
        wickets=("overall_wickets", "sum"),
        pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        pressure_runs=(
            "under_pressure_runs_conceded", "sum"
        ),
        pressure_wickets=("under_pressure_wickets", "sum")
    )
)

bowling_match_type_metrics = [
    "matches",
    "innings",
    "legal_balls",
    "runs_conceded",
    "wickets",
    "pressure_legal_balls",
    "pressure_runs",
    "pressure_wickets"
]

bowling_match_type_wide = bowling_match_type_stats.pivot(
    index=["player_id", "player_name"],
    columns="match_type",
    values=bowling_match_type_metrics
)

bowling_match_type_wide.columns = [
    f"{metric}_{match_type.lower()}"
    for metric, match_type in bowling_match_type_wide.columns
]

bowling_match_type_wide = (
    bowling_match_type_wide
    .reset_index()
    .fillna(0)
)

player_bowling_master = player_bowling_master.merge(
    bowling_match_type_wide,
    on=["player_id", "player_name"],
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 9. SUMMARY
# ------------------------------------------------------------

print("CORRECTED PLAYER-LEVEL BOWLING MASTER")
print("=" * 60)

print("Unique players:", len(player_bowling_master))

print(
    "Regular career wickets (Standard + Revised):",
    player_bowling_master["overall_wickets"].sum()
)

print(
    "Regular under-pressure wickets:",
    player_bowling_master["under_pressure_wickets"].sum()
)

print(
    "Separate Super Over wickets:",
    player_bowling_master["super_over_wickets"].sum()
)

print(
    "Separate Super Over pressure wickets:",
    player_bowling_master["super_over_pressure_wickets"].sum()
)

print("\nTop 15 players by regular career wickets:")

display(
    player_bowling_master
    .sort_values(
        ["overall_wickets", "overall_economy"],
        ascending=[False, True]
    )
    [[
        "player_id",
        "player_name",
        "matches_played",
        "bowling_innings",
        "overall_overs",
        "overall_runs_conceded",
        "overall_wickets",
        "overall_economy",
        "under_pressure_wickets",
        "under_pressure_economy",
        "super_over_wickets",
        "super_over_pressure_wickets"
    ]]
    .head(15)
)

print("\nMaster dataset shape:", player_bowling_master.shape)
```

```python

# ============================================================
# COMBINED ANALYSIS - STEP 34
# Player-level batting statistics
# Regular career = STANDARD + REVISED
# SUPER_OVER statistics remain separate
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. KEEP ALL PLAYERS IN THE MASTER
# ------------------------------------------------------------

all_batters = combined_batting[
    ["player_id", "player_name"]
].drop_duplicates()

# ------------------------------------------------------------
# 2. AGGREGATE REGULAR CAREER BATTING
# STANDARD + REVISED ONLY
# ------------------------------------------------------------

regular_batting = combined_batting[
    combined_batting["match_type"].isin(["STANDARD", "REVISED"])
].copy()

player_batting_overall = (
    regular_batting
    .groupby(["player_id", "player_name"], as_index=False)
    .agg(
        matches_played=("match_id", "nunique"),
        batting_innings=("match_id", "count"),

        overall_runs=("overall_runs", "sum"),
        overall_balls_faced=("overall_balls_faced", "sum"),
        overall_fours=("overall_fours", "sum"),
        overall_sixes=("overall_sixes", "sum"),

        under_pressure_runs=("under_pressure_runs", "sum"),
        under_pressure_balls_faced=(
            "under_pressure_balls_faced", "sum"
        ),
        under_pressure_fours=("under_pressure_fours", "sum"),
        under_pressure_sixes=("under_pressure_sixes", "sum")
    )
)

# ------------------------------------------------------------
# 3. AGGREGATE SUPER OVER BATTING SEPARATELY
# ------------------------------------------------------------

super_over_batting = combined_batting[
    combined_batting["match_type"] == "SUPER_OVER"
].copy()

player_super_over_batting = (
    super_over_batting
    .groupby(["player_id", "player_name"], as_index=False)
    .agg(
        super_over_matches=("match_id", "nunique"),
        super_over_innings=("match_id", "count"),

        super_over_runs=("overall_runs", "sum"),
        super_over_balls_faced=("overall_balls_faced", "sum"),
        super_over_fours=("overall_fours", "sum"),
        super_over_sixes=("overall_sixes", "sum"),

        super_over_pressure_runs=("under_pressure_runs", "sum"),
        super_over_pressure_balls_faced=(
            "under_pressure_balls_faced", "sum"
        ),
        super_over_pressure_fours=("under_pressure_fours", "sum"),
        super_over_pressure_sixes=("under_pressure_sixes", "sum")
    )
)

# ------------------------------------------------------------
# 4. MERGE REGULAR + SUPER OVER STATISTICS
# ------------------------------------------------------------

player_batting_master = (
    all_batters
    .merge(
        player_batting_overall,
        on=["player_id", "player_name"],
        how="left",
        validate="one_to_one"
    )
    .merge(
        player_super_over_batting,
        on=["player_id", "player_name"],
        how="left",
        validate="one_to_one"
    )
)

# Fill missing count metrics with zero
count_columns = [
    col for col in player_batting_master.columns
    if col not in ["player_id", "player_name"]
]

player_batting_master[count_columns] = (
    player_batting_master[count_columns].fillna(0)
)

# ------------------------------------------------------------
# 5. CALCULATE REGULAR CAREER BATTING RATES
# ------------------------------------------------------------

player_batting_master["overall_strike_rate"] = (
    player_batting_master["overall_runs"] * 100
    / player_batting_master[
        "overall_balls_faced"
    ].replace(0, np.nan)
).round(2)

player_batting_master["under_pressure_strike_rate"] = (
    player_batting_master["under_pressure_runs"] * 100
    / player_batting_master[
        "under_pressure_balls_faced"
    ].replace(0, np.nan)
).round(2)

# ------------------------------------------------------------
# 6. CALCULATE SUPER OVER BATTING RATES SEPARATELY
# ------------------------------------------------------------

player_batting_master["super_over_strike_rate"] = (
    player_batting_master["super_over_runs"] * 100
    / player_batting_master[
        "super_over_balls_faced"
    ].replace(0, np.nan)
).round(2)

player_batting_master["super_over_pressure_strike_rate"] = (
    player_batting_master["super_over_pressure_runs"] * 100
    / player_batting_master[
        "super_over_pressure_balls_faced"
    ].replace(0, np.nan)
).round(2)

# ------------------------------------------------------------
# 7. MATCH-TYPE BREAKDOWN
# Keep STANDARD, REVISED, SUPER_OVER separately
# ------------------------------------------------------------

batting_match_type_stats = (
    combined_batting
    .groupby(
        ["player_id", "player_name", "match_type"],
        as_index=False
    )
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),

        runs=("overall_runs", "sum"),
        balls_faced=("overall_balls_faced", "sum"),
        fours=("overall_fours", "sum"),
        sixes=("overall_sixes", "sum"),

        pressure_runs=("under_pressure_runs", "sum"),
        pressure_balls_faced=(
            "under_pressure_balls_faced", "sum"
        ),
        pressure_fours=("under_pressure_fours", "sum"),
        pressure_sixes=("under_pressure_sixes", "sum")
    )
)

batting_match_type_metrics = [
    "matches",
    "innings",
    "runs",
    "balls_faced",
    "fours",
    "sixes",
    "pressure_runs",
    "pressure_balls_faced",
    "pressure_fours",
    "pressure_sixes"
]

batting_match_type_wide = batting_match_type_stats.pivot(
    index=["player_id", "player_name"],
    columns="match_type",
    values=batting_match_type_metrics
)

batting_match_type_wide.columns = [
    f"{metric}_{match_type.lower()}"
    for metric, match_type in batting_match_type_wide.columns
]

batting_match_type_wide = (
    batting_match_type_wide
    .reset_index()
    .fillna(0)
)

player_batting_master = player_batting_master.merge(
    batting_match_type_wide,
    on=["player_id", "player_name"],
    how="left",
    validate="one_to_one"
)

# ------------------------------------------------------------
# 8. SUMMARY
# ------------------------------------------------------------

print("CORRECTED PLAYER-LEVEL BATTING MASTER")
print("=" * 60)

print("Unique players:", len(player_batting_master))

print(
    "Regular career runs (Standard + Revised):",
    player_batting_master["overall_runs"].sum()
)

print(
    "Regular under-pressure runs:",
    player_batting_master["under_pressure_runs"].sum()
)

print(
    "Separate Super Over runs:",
    player_batting_master["super_over_runs"].sum()
)

print(
    "Separate Super Over pressure runs:",
    player_batting_master["super_over_pressure_runs"].sum()
)

print("\nTop 15 players by regular career runs:")

display(
    player_batting_master
    .sort_values(
        ["overall_runs", "overall_strike_rate"],
        ascending=[False, False]
    )
    [[
        "player_id",
        "player_name",
        "matches_played",
        "batting_innings",
        "overall_runs",
        "overall_balls_faced",
        "overall_strike_rate",
        "under_pressure_runs",
        "under_pressure_balls_faced",
        "under_pressure_strike_rate",
        "super_over_runs",
        "super_over_pressure_runs"
    ]]
    .head(15)
)

print("\nMatch-type breakdown sample:")

display(
    player_batting_master[
        [
            "player_id",
            "player_name",
            "runs_standard",
            "runs_revised",
            "runs_super_over",
            "pressure_runs_standard",
            "pressure_runs_revised",
            "pressure_runs_super_over"
        ]
    ].head(10)
)

print("\nMaster dataset shape:", player_batting_master.shape)
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 35
# VALIDATE CORRECTED PLAYER-LEVEL MASTERS
# ============================================================

import pandas as pd
import numpy as np

print("PLAYER MASTER VALIDATION")
print("=" * 65)

# ------------------------------------------------------------
# 1. EXPECTED SOURCE TOTALS
# ------------------------------------------------------------

expected = {
    "batting": {
        "regular_runs": 266844,
        "regular_balls": 191077,
        "regular_pressure_runs": 28569,
        "regular_pressure_balls": 20287,
        "super_over_runs": 44,
        "super_over_balls": 33,
        "super_over_pressure_runs": 44,
        "super_over_pressure_balls": 33,
        "regular_fours": 23076,
        "regular_sixes": 12673,
        "super_over_fours": 6,
        "super_over_sixes": 1
    },
    "bowling": {
        "regular_wickets": 10163,
        "regular_pressure_wickets": 8824,
        "super_over_wickets": 11,
        "super_over_pressure_wickets": 11,
        "regular_runs": 276498,
        "regular_legal_balls": 190491,
        "regular_pressure_runs": 242436,
        "regular_pressure_legal_balls": 162614,
        "super_over_runs": 52,
        "super_over_legal_balls": 33
    }
}

# ------------------------------------------------------------
# 2. BATTER MASTER VALIDATION
# ------------------------------------------------------------

batting_checks = {
    "Regular runs": (
        player_batting_master["overall_runs"].sum(),
        expected["batting"]["regular_runs"]
    ),
    "Regular balls faced": (
        player_batting_master["overall_balls_faced"].sum(),
        expected["batting"]["regular_balls"]
    ),
    "Regular pressure runs": (
        player_batting_master["under_pressure_runs"].sum(),
        expected["batting"]["regular_pressure_runs"]
    ),
    "Regular pressure balls": (
        player_batting_master["under_pressure_balls_faced"].sum(),
        expected["batting"]["regular_pressure_balls"]
    ),
    "Super Over runs": (
        player_batting_master["super_over_runs"].sum(),
        expected["batting"]["super_over_runs"]
    ),
    "Super Over balls": (
        player_batting_master["super_over_balls_faced"].sum(),
        expected["batting"]["super_over_balls"]
    ),
    "Super Over pressure runs": (
        player_batting_master["super_over_pressure_runs"].sum(),
        expected["batting"]["super_over_pressure_runs"]
    ),
    "Super Over pressure balls": (
        player_batting_master["super_over_pressure_balls_faced"].sum(),
        expected["batting"]["super_over_pressure_balls"]
    ),
    "Regular fours": (
        player_batting_master["overall_fours"].sum(),
        expected["batting"]["regular_fours"]
    ),
    "Regular sixes": (
        player_batting_master["overall_sixes"].sum(),
        expected["batting"]["regular_sixes"]
    ),
    "Super Over fours": (
        player_batting_master["super_over_fours"].sum(),
        expected["batting"]["super_over_fours"]
    ),
    "Super Over sixes": (
        player_batting_master["super_over_sixes"].sum(),
        expected["batting"]["super_over_sixes"]
    )
}

print("\nBATTING TOTAL RECONCILIATION")
print("-" * 65)

for metric, (actual, target) in batting_checks.items():
    passed = np.isclose(actual, target)
    print(
        f"{metric}: Actual={actual}, Expected={target}, "
        f"{'PASS' if passed else 'FAIL'}"
    )

# ------------------------------------------------------------
# 3. BOWLER MASTER VALIDATION
# ------------------------------------------------------------

bowling_checks = {
    "Regular wickets": (
        player_bowling_master["overall_wickets"].sum(),
        expected["bowling"]["regular_wickets"]
    ),
    "Regular pressure wickets": (
        player_bowling_master["under_pressure_wickets"].sum(),
        expected["bowling"]["regular_pressure_wickets"]
    ),
    "Super Over wickets": (
        player_bowling_master["super_over_wickets"].sum(),
        expected["bowling"]["super_over_wickets"]
    ),
    "Super Over pressure wickets": (
        player_bowling_master["super_over_pressure_wickets"].sum(),
        expected["bowling"]["super_over_pressure_wickets"]
    ),
    "Regular runs conceded": (
        player_bowling_master["overall_runs_conceded"].sum(),
        expected["bowling"]["regular_runs"]
    ),
    "Regular legal balls": (
        player_bowling_master["overall_legal_balls"].sum(),
        expected["bowling"]["regular_legal_balls"]
    ),
    "Regular pressure runs conceded": (
        player_bowling_master["under_pressure_runs_conceded"].sum(),
        expected["bowling"]["regular_pressure_runs"]
    ),
    "Regular pressure legal balls": (
        player_bowling_master["under_pressure_legal_balls"].sum(),
        expected["bowling"]["regular_pressure_legal_balls"]
    ),
    "Super Over runs conceded": (
        player_bowling_master["super_over_runs_conceded"].sum(),
        expected["bowling"]["super_over_runs"]
    ),
    "Super Over legal balls": (
        player_bowling_master["super_over_legal_balls"].sum(),
        expected["bowling"]["super_over_legal_balls"]
    )
}

print("\nBOWLING TOTAL RECONCILIATION")
print("-" * 65)

for metric, (actual, target) in bowling_checks.items():
    passed = np.isclose(actual, target)
    print(
        f"{metric}: Actual={actual}, Expected={target}, "
        f"{'PASS' if passed else 'FAIL'}"
    )

# ------------------------------------------------------------
# 4. PLAYER ID AND DUPLICATE VALIDATION
# ------------------------------------------------------------

print("\nPLAYER MASTER STRUCTURE")
print("-" * 65)

for name, df in [
    ("Batting", player_batting_master),
    ("Bowling", player_bowling_master)
]:
    print(f"\n{name}:")
    print("Rows:", len(df))
    print("Unique player IDs:", df["player_id"].nunique())
    print("Duplicate player IDs:", df["player_id"].duplicated().sum())
    print("Missing player IDs:", df["player_id"].isna().sum())
    print("Missing player names:", df["player_name"].isna().sum())

# ------------------------------------------------------------
# 5. PRESSURE STATISTICS MUST NOT EXCEED OVERALL
# ------------------------------------------------------------

print("\nPRESSURE VS OVERALL VALIDATION")
print("-" * 65)

batting_pressure_checks = {
    "Runs": (
        player_batting_master["under_pressure_runs"]
        <= player_batting_master["overall_runs"]
    ),
    "Balls": (
        player_batting_master["under_pressure_balls_faced"]
        <= player_batting_master["overall_balls_faced"]
    ),
    "Fours": (
        player_batting_master["under_pressure_fours"]
        <= player_batting_master["overall_fours"]
    ),
    "Sixes": (
        player_batting_master["under_pressure_sixes"]
        <= player_batting_master["overall_sixes"]
    )
}

for metric, condition in batting_pressure_checks.items():
    print(
        f"Batting {metric}:",
        "PASS" if condition.all() else "FAIL",
        "| Failures:", (~condition).sum()
    )

bowling_pressure_checks = {
    "Wickets": (
        player_bowling_master["under_pressure_wickets"]
        <= player_bowling_master["overall_wickets"]
    ),
    "Runs": (
        player_bowling_master["under_pressure_runs_conceded"]
        <= player_bowling_master["overall_runs_conceded"]
    ),
    "Legal balls": (
        player_bowling_master["under_pressure_legal_balls"]
        <= player_bowling_master["overall_legal_balls"]
    )
}

for metric, condition in bowling_pressure_checks.items():
    print(
        f"Bowling {metric}:",
        "PASS" if condition.all() else "FAIL",
        "| Failures:", (~condition).sum()
    )

# ------------------------------------------------------------
# 6. SUPER OVER PRESSURE SEPARATION
# ------------------------------------------------------------

print("\nSUPER OVER PRESSURE VALIDATION")
print("-" * 65)

so_batting_checks = {
    "Runs": (
        player_batting_master["super_over_runs"]
        == player_batting_master["super_over_pressure_runs"]
    ),
    "Balls": (
        player_batting_master["super_over_balls_faced"]
        == player_batting_master["super_over_pressure_balls_faced"]
    ),
    "Fours": (
        player_batting_master["super_over_fours"]
        == player_batting_master["super_over_pressure_fours"]
    ),
    "Sixes": (
        player_batting_master["super_over_sixes"]
        == player_batting_master["super_over_pressure_sixes"]
    )
}

for metric, condition in so_batting_checks.items():
    print(
        f"Batting {metric}:",
        "PASS" if condition.all() else "FAIL"
    )

so_bowling_checks = {
    "Wickets": (
        player_bowling_master["super_over_wickets"]
        == player_bowling_master["super_over_pressure_wickets"]
    ),
    "Runs": (
        player_bowling_master["super_over_runs_conceded"]
        == player_bowling_master["super_over_pressure_runs_conceded"]
    ),
    "Legal balls": (
        player_bowling_master["super_over_legal_balls"]
        == player_bowling_master["super_over_pressure_legal_balls"]
    )
}

for metric, condition in so_bowling_checks.items():
    print(
        f"Bowling {metric}:",
        "PASS" if condition.all() else "FAIL"
    )

# ------------------------------------------------------------
# 7. FINAL SUMMARY
# ------------------------------------------------------------

all_total_checks = (
    list(batting_checks.values())
    + list(bowling_checks.values())
)

all_totals_passed = all(
    np.isclose(actual, target)
    for actual, target in all_total_checks
)

all_structure_passed = all(
    df["player_id"].is_unique
    and df["player_id"].notna().all()
    and df["player_name"].notna().all()
    for df in [
        player_batting_master,
        player_bowling_master
    ]
)

all_pressure_passed = (
    all(condition.all() for condition in batting_pressure_checks.values())
    and all(condition.all() for condition in bowling_pressure_checks.values())
)

all_so_passed = (
    all(condition.all() for condition in so_batting_checks.values())
    and all(condition.all() for condition in so_bowling_checks.values())
)

print("\n" + "=" * 65)
print("FINAL VALIDATION RESULT")
print("=" * 65)

print("All source totals reconciled:", all_totals_passed)
print("Both player masters structurally valid:", all_structure_passed)
print("Pressure metrics valid:", all_pressure_passed)
print("Super Over separation valid:", all_so_passed)

print(
    "OVERALL RESULT:",
    "PASS" if all([
        all_totals_passed,
        all_structure_passed,
        all_pressure_passed,
        all_so_passed
    ]) else "CHECK REQUIRED"
)
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 36
# MATCH-TYPE BREAKDOWN AND BATTING RATE VALIDATION
# ============================================================

import numpy as np
import pandas as pd

print("MATCH-TYPE AND BATTING RATE VALIDATION")
print("=" * 65)


# ------------------------------------------------------------
# 1. REBUILD EXPECTED BATTING MATCH-TYPE TOTALS
# ------------------------------------------------------------

batting_source = (
    combined_batting
    .groupby(["player_id", "match_type"], as_index=False)
    .agg(
        matches=("match_id", "nunique"),
        innings=("match_id", "count"),
        runs=("overall_runs", "sum"),
        balls_faced=("overall_balls_faced", "sum"),
        fours=("overall_fours", "sum"),
        sixes=("overall_sixes", "sum"),
        pressure_runs=("under_pressure_runs", "sum"),
        pressure_balls_faced=("under_pressure_balls_faced", "sum"),
        pressure_fours=("under_pressure_fours", "sum"),
        pressure_sixes=("under_pressure_sixes", "sum")
    )
)

batting_metrics = [
    "matches",
    "innings",
    "runs",
    "balls_faced",
    "fours",
    "sixes",
    "pressure_runs",
    "pressure_balls_faced",
    "pressure_fours",
    "pressure_sixes"
]

batting_expected = batting_source.pivot(
    index="player_id",
    columns="match_type",
    values=batting_metrics
)

batting_expected.columns = [
    f"{metric}_{match_type.lower()}"
    for metric, match_type in batting_expected.columns
]

batting_expected = batting_expected.fillna(0)


# ------------------------------------------------------------
# 2. COMPARE BATTING MATCH-TYPE BREAKDOWN
# ------------------------------------------------------------

batting_comparison = (
    player_batting_master
    .set_index("player_id")
    .join(batting_expected, how="left", rsuffix="_expected")
)

batting_breakdown_results = []

for col in batting_expected.columns:
    actual_col = col
    expected_col = col + "_expected"

    actual = batting_comparison[actual_col].fillna(0)
    expected_values = batting_comparison[expected_col].fillna(0)

    mismatch_count = (~np.isclose(actual, expected_values)).sum()

    batting_breakdown_results.append({
        "metric": col,
        "mismatches": mismatch_count,
        "status": "PASS" if mismatch_count == 0 else "FAIL"
    })

batting_breakdown_report = pd.DataFrame(batting_breakdown_results)

print("\nBATTING MATCH-TYPE BREAKDOWN")
print("-" * 65)
display(batting_breakdown_report)


# ------------------------------------------------------------
# 3. VALIDATE BATTING STRIKE RATES
# ------------------------------------------------------------

def calculate_strike_rate(runs, balls):
    return (
        runs * 100
        / balls.replace(0, np.nan)
    ).round(2)


batting_rate_checks = {
    "Overall strike rate": (
        player_batting_master["overall_strike_rate"],
        calculate_strike_rate(
            player_batting_master["overall_runs"],
            player_batting_master["overall_balls_faced"]
        )
    ),
    "Under-pressure strike rate": (
        player_batting_master["under_pressure_strike_rate"],
        calculate_strike_rate(
            player_batting_master["under_pressure_runs"],
            player_batting_master["under_pressure_balls_faced"]
        )
    ),
    "Super Over strike rate": (
        player_batting_master["super_over_strike_rate"],
        calculate_strike_rate(
            player_batting_master["super_over_runs"],
            player_batting_master["super_over_balls_faced"]
        )
    ),
    "Super Over pressure strike rate": (
        player_batting_master["super_over_pressure_strike_rate"],
        calculate_strike_rate(
            player_batting_master["super_over_pressure_runs"],
            player_batting_master["super_over_pressure_balls_faced"]
        )
    )
}

print("\nBATTING STRIKE-RATE VALIDATION")
print("-" * 65)

batting_rate_results = []

for metric, (actual, expected_values) in batting_rate_checks.items():
    matches = np.isclose(
        actual,
        expected_values,
        equal_nan=True
    )

    failures = (~matches).sum()

    batting_rate_results.append({
        "metric": metric,
        "mismatches": failures,
        "status": "PASS" if failures == 0 else "FAIL"
    })

display(pd.DataFrame(batting_rate_results))


# ------------------------------------------------------------
# 4. INSPECT BOWLING RATE COLUMNS FOR NEXT VALIDATION
# ------------------------------------------------------------

bowling_rate_columns = [
    col for col in player_bowling_master.columns
    if any(
        keyword in col.lower()
        for keyword in ["economy", "average", "strike_rate"]
    )
]

print("\nBOWLING RATE COLUMNS")
print("-" * 65)

for col in bowling_rate_columns:
    print(col)


# ------------------------------------------------------------
# 5. FINAL SUMMARY
# ------------------------------------------------------------

batting_breakdown_passed = (
    batting_breakdown_report["mismatches"].sum() == 0
)

batting_rates_passed = (
    pd.DataFrame(batting_rate_results)["mismatches"].sum() == 0
)

print("\n" + "=" * 65)
print("STEP 36 SUMMARY")
print("=" * 65)

print("Batting match-type breakdown:", batting_breakdown_passed)
print("Batting strike rates:", batting_rates_passed)

print(
    "STEP 36 RESULT:",
    "PASS" if batting_breakdown_passed and batting_rates_passed
    else "CHECK REQUIRED"
)
```

```python
# ============================================================
# COMBINED ANALYSIS - STEP 37
# BOWLING RATE VALIDATION
# ============================================================

import numpy as np
import pandas as pd

print("BOWLING RATE VALIDATION")
print("=" * 65)


# ------------------------------------------------------------
# 1. RATE CALCULATION HELPER
# ------------------------------------------------------------

def safe_divide(numerator, denominator):
    return numerator / denominator.replace(0, np.nan)


def calculate_bowling_rates(runs, legal_balls, wickets):

    return {
        "economy": (
            safe_divide(runs * 6, legal_balls)
        ).round(2),

        "average": (
            safe_divide(runs, wickets)
        ).round(2),

        "strike_rate": (
            safe_divide(legal_balls, wickets)
        ).round(2)
    }


# ------------------------------------------------------------
# 2. DEFINE ALL BOWLING RATE GROUPS
# ------------------------------------------------------------

rate_groups = {
    "Overall": {
        "runs": "overall_runs_conceded",
        "balls": "overall_legal_balls",
        "wickets": "overall_wickets",
        "economy": "overall_economy",
        "average": "overall_bowling_average",
        "strike_rate": "overall_bowling_strike_rate"
    },

    "Under Pressure": {
        "runs": "under_pressure_runs_conceded",
        "balls": "under_pressure_legal_balls",
        "wickets": "under_pressure_wickets",
        "economy": "under_pressure_economy",
        "average": "under_pressure_bowling_average",
        "strike_rate": "under_pressure_bowling_strike_rate"
    },

    "Super Over": {
        "runs": "super_over_runs_conceded",
        "balls": "super_over_legal_balls",
        "wickets": "super_over_wickets",
        "economy": "super_over_economy",
        "average": "super_over_bowling_average",
        "strike_rate": "super_over_bowling_strike_rate"
    }
}


# ------------------------------------------------------------
# 3. VALIDATE EACH RATE
# ------------------------------------------------------------

bowling_rate_results = []

for group_name, cols in rate_groups.items():

    expected_rates = calculate_bowling_rates(
        player_bowling_master[cols["runs"]],
        player_bowling_master[cols["balls"]],
        player_bowling_master[cols["wickets"]]
    )

    for rate_name, expected_values in expected_rates.items():

        actual_values = player_bowling_master[
            cols[rate_name]
        ]

        matches = np.isclose(
            actual_values,
            expected_values,
            equal_nan=True
        )

        failures = (~matches).sum()

        bowling_rate_results.append({
            "group": group_name,
            "rate": rate_name,
            "mismatches": failures,
            "status": "PASS" if failures == 0 else "FAIL"
        })


bowling_rate_report = pd.DataFrame(
    bowling_rate_results
)

print("\nRATE VALIDATION RESULTS")
print("-" * 65)

display(bowling_rate_report)


# ------------------------------------------------------------
# 4. CHECK ZERO-DENOMINATOR HANDLING
# ------------------------------------------------------------

print("\nZERO-DENOMINATOR VALIDATION")
print("-" * 65)

zero_denominator_checks = []

for group_name, cols in rate_groups.items():

    for rate_name, denominator in [
        ("economy", cols["balls"]),
        ("average", cols["wickets"]),
        ("strike_rate", cols["wickets"])
    ]:

        zero_mask = (
            player_bowling_master[denominator] == 0
        )

        rate_column = cols[rate_name]

        invalid_count = (
            player_bowling_master.loc[
                zero_mask, rate_column
            ].notna().sum()
        )

        zero_denominator_checks.append({
            "group": group_name,
            "rate": rate_name,
            "zero_denominator_players": zero_mask.sum(),
            "invalid_non_null_rates": invalid_count,
            "status": (
                "PASS" if invalid_count == 0 else "FAIL"
            )
        })


zero_denominator_report = pd.DataFrame(
    zero_denominator_checks
)

display(zero_denominator_report)


# ------------------------------------------------------------
# 5. FINAL SUMMARY
# ------------------------------------------------------------

rates_passed = (
    bowling_rate_report["mismatches"].sum() == 0
)

zero_denominators_passed = (
    zero_denominator_report[
        "invalid_non_null_rates"
    ].sum() == 0
)

print("\n" + "=" * 65)
print("STEP 37 SUMMARY")
print("=" * 65)

print("Bowling rate calculations:", rates_passed)
print("Zero-denominator handling:", zero_denominators_passed)

print(
    "STEP 37 RESULT:",
    "PASS" if rates_passed and zero_denominators_passed
    else "CHECK REQUIRED"
)
```

```python
# ============================================================
# STEP 43 - INSPECT COMBINED BOWLING MATCH CONTEXT
# ============================================================

required_columns = [
    "player_id",
    "player_name",
    "match_id",
    "match_type",
    "is_knockout",
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets"
]

print("REQUIRED COLUMN AVAILABILITY")
print("=" * 65)

for column in required_columns:
    print(column, "->", column in combined_bowling.columns)


print("\nMATCH TYPE DISTRIBUTION")
print("=" * 65)

display(
    combined_bowling["match_type"].value_counts().to_frame(
        "records"
    )
)


print("\nKNOCKOUT CLASSIFICATION")
print("=" * 65)

regular_bowling = combined_bowling[
    combined_bowling["match_type"].isin(
        ["STANDARD", "REVISED"]
    )
].copy()

print(
    "Regular bowling records:",
    len(regular_bowling)
)

print(
    "Missing knockout classification:",
    regular_bowling["is_knockout"].isna().sum()
)

display(
    regular_bowling.groupby(
        ["match_type", "is_knockout"],
        dropna=False
    ).size().reset_index(name="records")
)


print("\nPRESSURE METRIC VALIDATION")
print("=" * 65)

pressure_columns = [
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets"
]

display(
    regular_bowling[pressure_columns].describe()
)
```

```python
# ============================================================
# STEP 44 - CALCULATE KNOCKOUT AND ORDINARY PRESSURE METRICS
# ============================================================

# Aggregate knockout pressure performance
knockout_bowling = (
    regular_bowling[
        regular_bowling["is_knockout"] == True
    ]
    .groupby("player_id", as_index=False)
    .agg(
        knockout_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        knockout_pressure_runs=(
            "under_pressure_runs_conceded", "sum"
        ),
        knockout_pressure_wickets=(
            "under_pressure_wickets", "sum"
        )
    )
)


# Aggregate ordinary-match pressure performance
ordinary_bowling = (
    regular_bowling[
        regular_bowling["is_knockout"] == False
    ]
    .groupby("player_id", as_index=False)
    .agg(
        ordinary_pressure_legal_balls=(
            "under_pressure_legal_balls", "sum"
        ),
        ordinary_pressure_runs=(
            "under_pressure_runs_conceded", "sum"
        ),
        ordinary_pressure_wickets=(
            "under_pressure_wickets", "sum"
        )
    )
)


# Start with qualified bowlers only
bowling_ranking_data = qualified_bowlers.copy()


# Merge knockout and ordinary metrics
bowling_ranking_data = bowling_ranking_data.merge(
    knockout_bowling,
    on="player_id",
    how="left",
    validate="one_to_one"
)

bowling_ranking_data = bowling_ranking_data.merge(
    ordinary_bowling,
    on="player_id",
    how="left",
    validate="one_to_one"
)


# Fill missing counting metrics with zero
pressure_count_columns = [
    "knockout_pressure_legal_balls",
    "knockout_pressure_runs",
    "knockout_pressure_wickets",
    "ordinary_pressure_legal_balls",
    "ordinary_pressure_runs",
    "ordinary_pressure_wickets"
]

bowling_ranking_data[pressure_count_columns] = (
    bowling_ranking_data[pressure_count_columns]
    .fillna(0)
)


# Calculate knockout pressure rates
bowling_ranking_data["knockout_pressure_economy"] = (
    bowling_ranking_data["knockout_pressure_runs"] * 6 /
    bowling_ranking_data["knockout_pressure_legal_balls"]
)

bowling_ranking_data["knockout_pressure_average"] = (
    bowling_ranking_data["knockout_pressure_runs"] /
    bowling_ranking_data["knockout_pressure_wickets"]
)

bowling_ranking_data["knockout_pressure_strike_rate"] = (
    bowling_ranking_data["knockout_pressure_legal_balls"] /
    bowling_ranking_data["knockout_pressure_wickets"]
)


# Calculate ordinary pressure rates
bowling_ranking_data["ordinary_pressure_economy"] = (
    bowling_ranking_data["ordinary_pressure_runs"] * 6 /
    bowling_ranking_data["ordinary_pressure_legal_balls"]
)

bowling_ranking_data["ordinary_pressure_average"] = (
    bowling_ranking_data["ordinary_pressure_runs"] /
    bowling_ranking_data["ordinary_pressure_wickets"]
)

bowling_ranking_data["ordinary_pressure_strike_rate"] = (
    bowling_ranking_data["ordinary_pressure_legal_balls"] /
    bowling_ranking_data["ordinary_pressure_wickets"]
)


# Display top-level summary
print("BOWLING RANKING DATA SUMMARY")
print("=" * 65)

print("Qualified bowlers:", len(bowling_ranking_data))

print(
    "Knockout pressure wickets:",
    bowling_ranking_data["knockout_pressure_wickets"].sum()
)

print(
    "Ordinary pressure wickets:",
    bowling_ranking_data["ordinary_pressure_wickets"].sum()
)

print(
    "Combined pressure wickets:",
    bowling_ranking_data["under_pressure_wickets"].sum()
)


print("\nPRESSURE METRIC RECONCILIATION")
print("=" * 65)

print(
    "Knockout + ordinary wickets reconcile:",
    (
        bowling_ranking_data["knockout_pressure_wickets"] +
        bowling_ranking_data["ordinary_pressure_wickets"]
    ).equals(
        bowling_ranking_data["under_pressure_wickets"]
    )
)

print(
    "Knockout + ordinary legal balls reconcile:",
    (
        bowling_ranking_data["knockout_pressure_legal_balls"] +
        bowling_ranking_data["ordinary_pressure_legal_balls"]
    ).equals(
        bowling_ranking_data["under_pressure_legal_balls"]
    )
)

print(
    "Knockout + ordinary runs reconcile:",
    (
        bowling_ranking_data["knockout_pressure_runs"] +
        bowling_ranking_data["ordinary_pressure_runs"]
    ).equals(
        bowling_ranking_data["under_pressure_runs_conceded"]
    )
)


print("\nTOP KNOCKOUT PRESSURE BOWLERS")
print("=" * 65)

display(
    bowling_ranking_data.sort_values(
        "knockout_pressure_wickets",
        ascending=False
    )[
        [
            "player_name",
            "knockout_pressure_wickets",
            "knockout_pressure_legal_balls",
            "knockout_pressure_runs",
            "knockout_pressure_economy"
        ]
    ].head(10)
)
```

```python
# ============================================================
# STEP 45A - STANDARDIZE BOWLING RANKING DATA TYPES
# ============================================================

# Integer metrics
bowling_integer_columns = [
    "bowling_innings",
    "matches_played",
    "overall_legal_balls",
    "overall_runs_conceded",
    "overall_wickets",
    "under_pressure_legal_balls",
    "under_pressure_runs_conceded",
    "under_pressure_wickets",
    "knockout_pressure_legal_balls",
    "knockout_pressure_runs",
    "knockout_pressure_wickets",
    "ordinary_pressure_legal_balls",
    "ordinary_pressure_runs",
    "ordinary_pressure_wickets",
    "super_over_legal_balls",
    "super_over_runs_conceded",
    "super_over_wickets"
]

# Only apply to columns that exist
bowling_integer_columns = [
    col for col in bowling_integer_columns
    if col in bowling_ranking_data.columns
]

for col in bowling_integer_columns:
    bowling_ranking_data[col] = (
        bowling_ranking_data[col]
        .fillna(0)
        .round()
        .astype("int64")
    )


# Round rate metrics to 2 decimal places
bowling_rate_columns = [
    col for col in bowling_ranking_data.columns
    if any(
        metric in col.lower()
        for metric in [
            "economy",
            "average",
            "strike_rate"
        ]
    )
]

for col in bowling_rate_columns:
    bowling_ranking_data[col] = (
        bowling_ranking_data[col].round(2)
    )


print("BOWLING DATA FORMATTING VALIDATION")
print("=" * 65)

print("\nInteger column data types:")
display(
    bowling_ranking_data[bowling_integer_columns].dtypes.to_frame(
        "data_type"
    )
)

print("\nRate column data types:")
display(
    bowling_ranking_data[bowling_rate_columns].dtypes.to_frame(
        "data_type"
    )
)

print(
    "\nAll wicket values are integers:",
    bowling_ranking_data["overall_wickets"].apply(
        lambda x: isinstance(x, (int, __import__("numpy").integer))
    ).all()
)

print(
    "All knockout pressure wickets are integers:",
    bowling_ranking_data["knockout_pressure_wickets"].dtype == "int64"
)
```

```python
# ============================================================
# STEP 45 - CREATE FINAL BOWLING PRESSURE RANKINGS
# ============================================================

qualified_bowling_rankings = (
    bowling_ranking_data
    .sort_values(
        by=[
            "knockout_pressure_wickets",
            "knockout_pressure_economy",
            "knockout_pressure_strike_rate",
            "ordinary_pressure_wickets",
            "ordinary_pressure_economy",
            "ordinary_pressure_strike_rate",
            "overall_wickets",
            "overall_economy",
            "overall_bowling_strike_rate"
        ],
        ascending=[
            False, True, True,
            False, True, True,
            False, True, True
        ],
        na_position="last"
    )
    .reset_index(drop=True)
)

qualified_bowling_rankings.insert(
    0, "rank", range(1, len(qualified_bowling_rankings) + 1)
)

top_10_bowlers = qualified_bowling_rankings.head(10).copy()

display(
    top_10_bowlers[
        [
            "rank",
            "player_id",
            "player_name",
            "bowling_innings",
            "overall_wickets",
            "overall_economy",
            "knockout_pressure_wickets",
            "knockout_pressure_legal_balls",
            "knockout_pressure_economy",
            "ordinary_pressure_wickets",
            "ordinary_pressure_economy",
            "under_pressure_wickets",
            "super_over_wickets"
        ]
    ]
)


print("\nBOWLING RANKING VALIDATION")
print("=" * 65)

print("Qualified bowlers:", len(qualified_bowling_rankings))
print("Top 10 records:", len(top_10_bowlers))
print("Duplicate player IDs:", qualified_bowling_rankings["player_id"].duplicated().sum())
print(
    "Pakistan players:",
    qualified_bowling_rankings["player_id"].isin(pakistan_player_ids).sum()
)
print(
    "Knockout wickets sorted correctly:",
    qualified_bowling_rankings["knockout_pressure_wickets"].is_monotonic_decreasing
)
print(
    "All displayed economy values rounded:",
    top_10_bowlers[
        ["overall_economy", "knockout_pressure_economy", "ordinary_pressure_economy"]
    ].apply(
        lambda col: col.dropna().map(
            lambda x: round(x, 2) == x
        ).all()
    ).all()
)
```

```python
# ============================================================
# STEP 46 - STANDARDIZE BATTING RANKING FORMATTING
# ============================================================

batting_integer_columns = [
    "matches_played",
    "batting_innings",
    "overall_runs",
    "overall_balls_faced",
    "overall_fours",
    "overall_sixes",
    "under_pressure_runs",
    "under_pressure_balls_faced",
    "under_pressure_fours",
    "under_pressure_sixes",
    "super_over_runs",
    "super_over_balls_faced",
    "super_over_fours",
    "super_over_sixes"
]

batting_integer_columns = [
    col for col in batting_integer_columns
    if col in qualified_batting_rankings.columns
]

for col in batting_integer_columns:
    qualified_batting_rankings[col] = (
        qualified_batting_rankings[col]
        .fillna(0)
        .round()
        .astype("int64")
    )


batting_rate_columns = [
    col for col in qualified_batting_rankings.columns
    if "strike_rate" in col.lower()
]

for col in batting_rate_columns:
    qualified_batting_rankings[col] = (
        qualified_batting_rankings[col].round(2)
    )


top_10_batters = qualified_batting_rankings.head(10).copy()


print("BATTING FORMATTING VALIDATION")
print("=" * 65)

print("Integer columns:")
display(
    qualified_batting_rankings[
        batting_integer_columns
    ].dtypes.to_frame("data_type")
)

print("\nStrike-rate columns:")
display(
    qualified_batting_rankings[
        batting_rate_columns
    ].dtypes.to_frame("data_type")
)

print(
    "\nAll batting run values are integers:",
    qualified_batting_rankings["overall_runs"].dtype == "int64"
)

print(
    "All pressure run values are integers:",
    qualified_batting_rankings["under_pressure_runs"].dtype == "int64"
)
```

```python
# ============================================================
# STEP 47 - FINAL BATTING AND BOWLING LEADERBOARD REVIEW
# ============================================================

print("TOP 10 BATTERS")
print("=" * 100)

display(
    top_10_batters[
        [
            "rank",
            "player_name",
            "matches_played",
            "overall_runs",
            "overall_strike_rate",
            "under_pressure_runs",
            "under_pressure_balls_faced",
            "under_pressure_strike_rate",
            "super_over_runs"
        ]
    ]
)

print("\nTOP 10 BOWLERS")
print("=" * 100)

display(
    top_10_bowlers[
        [
            "rank",
            "player_name",
            "matches_played",
            "overall_wickets",
            "overall_economy",
            "knockout_pressure_wickets",
            "knockout_pressure_economy",
            "ordinary_pressure_wickets",
            "under_pressure_wickets",
            "super_over_wickets"
        ]
    ]
)

print("\nFINAL DATASET SUMMARY")
print("=" * 100)

print("Qualified batters:", len(qualified_batting_rankings))
print("Qualified bowlers:", len(qualified_bowling_rankings))

print("Top 10 batters:", len(top_10_batters))
print("Top 10 bowlers:", len(top_10_bowlers))

print(
    "Pakistan excluded from batting:",
    qualified_batting_rankings["player_id"].isin(pakistan_player_ids).sum() == 0
)

print(
    "Pakistan excluded from bowling:",
    qualified_bowling_rankings["player_id"].isin(pakistan_player_ids).sum() == 0
)
```

```python
# 1. Competition mapping: one row per player, competition and franchise
ipl_listing = (
    ipl_mapping[
        ["player_id", "player_name", "comp_id", "comp_name",
         "team_id", "team_name"]
    ]
    .drop_duplicates()
)

# 2. Actual appearances: only records marked is_played=True
ipl_appearances = (
    ipl_participation[
        ipl_participation["is_played"] == True
    ]
    .groupby(["player_id", "comp_id", "comp_name"], dropna=False)
    .agg(
        ipl_matches_played=("match_id", "nunique"),
        ipl_impact_in_appearances=("is_impact", lambda x: (x == 1).sum()),
        ipl_impact_out_appearances=("is_impact_off", lambda x: (x == 1).sum())
    )
    .reset_index()
)

# 3. Combine listing and actual appearances
ipl_player_season_audit = ipl_listing.merge(
    ipl_appearances,
    on=["player_id", "comp_id", "comp_name"],
    how="left"
)

# 4. Missing appearance counts mean no recorded appearances
count_columns = [
    "ipl_matches_played",
    "ipl_impact_in_appearances",
    "ipl_impact_out_appearances"
]

ipl_player_season_audit[count_columns] = (
    ipl_player_season_audit[count_columns]
    .fillna(0)
    .astype(int)
)

ipl_player_season_audit["actually_participated"] = (
    ipl_player_season_audit["ipl_matches_played"] > 0
)

print("IPL player-competition records:", len(ipl_player_season_audit))
print("Unique players:", ipl_player_season_audit["player_id"].nunique())
print(
    "Players with appearances:",
    ipl_player_season_audit.loc[
        ipl_player_season_audit["actually_participated"], "player_id"
    ].nunique()
)
print(
    "Mapping-only players:",
    ipl_player_season_audit.loc[
        ~ipl_player_season_audit["actually_participated"], "player_id"
    ].nunique()
)

display(
    ipl_player_season_audit.sort_values(
        ["comp_name", "player_name"]
    ).head(30)
)
```

```python
print("=" * 65)
print("IPL PLAYER-COMPETITION DUPLICATE CHECK")
print("=" * 65)

# Check duplicate player + competition combinations
duplicate_comp = (
    ipl_listing.groupby(["player_id", "comp_id"])
    .agg(
        mapping_rows=("team_id", "size"),
        unique_teams=("team_id", "nunique"),
        team_names=("team_name", lambda x: ", ".join(sorted(x.dropna().unique())))
    )
    .reset_index()
)

duplicates = duplicate_comp[duplicate_comp["mapping_rows"] > 1]

print("Total mapping rows:", len(ipl_listing))
print("Unique player + competition combinations:",
      ipl_listing[["player_id", "comp_id"]].drop_duplicates().shape[0])
print("Duplicate player + competition combinations:", len(duplicates))
print("Players associated with multiple teams in same competition:",
      (duplicates["unique_teams"] > 1).sum())

display(duplicates.head(30))

print("\n" + "=" * 65)
print("UNIQUE MAPPING-ONLY PLAYER VALIDATION")
print("=" * 65)

mapping_only_season = ipl_player_season_audit[
    ~ipl_player_season_audit["actually_participated"]
]

print("Mapping-only player-season-team rows:",
      len(mapping_only_season))
print("Unique mapping-only players:",
      mapping_only_season["player_id"].nunique())
```

```python
# Aggregate IPL experience to one row per player
ipl_player_experience = (
    ipl_player_season_audit
    .groupby("player_id")
    .agg(
        player_name=("player_name", "first"),
        ipl_competitions_listed=("comp_id", "nunique"),
        ipl_seasons_listed=("comp_name", "nunique"),
        ipl_matches_played=("ipl_matches_played", "sum"),
        ipl_seasons_played=("actually_participated", "sum"),
        ipl_mapping_only_competitions=(
            "actually_participated",
            lambda x: (~x).sum()
        ),
        ipl_impact_in_appearances=(
            "ipl_impact_in_appearances", "sum"
        ),
        ipl_impact_out_appearances=(
            "ipl_impact_out_appearances", "sum"
        )
    )
    .reset_index()
)

ipl_player_experience["ipl_ever_listed"] = True

ipl_player_experience["ipl_ever_played"] = (
    ipl_player_experience["ipl_matches_played"] > 0
)

print("=" * 65)
print("IPL PLAYER EXPERIENCE VALIDATION")
print("=" * 65)

print("Unique listed players:", len(ipl_player_experience))
print(
    "Players with IPL appearances:",
    ipl_player_experience["ipl_ever_played"].sum()
)
print(
    "Players with zero IPL appearances:",
    (~ipl_player_experience["ipl_ever_played"]).sum()
)
print(
    "Total player-match appearances:",
    ipl_player_experience["ipl_matches_played"].sum()
)
print(
    "Total incoming impact appearances:",
    ipl_player_experience["ipl_impact_in_appearances"].sum()
)
print(
    "Total outgoing impact appearances:",
    ipl_player_experience["ipl_impact_out_appearances"].sum()
)

print("\nPlayers with zero IPL appearances:")
display(
    ipl_player_experience[
        ~ipl_player_experience["ipl_ever_played"]
    ].sort_values("player_name").head(30)
)

print("\nFive previously verified players:")
display(
    ipl_player_experience[
        ipl_player_experience["player_id"].isin(
            [35, 1600, 2299, 1829, 7095]
        )
    ]
)
```

```python
import pandas as pd

# 1. Build the complete player universe from competition mapping
player_universe = (
    competition_team_player_mapping[
        ["player_id", "player_name"]
    ]
    .drop_duplicates("player_id")
    .copy()
)

# 2. Prefer performance-table names; use mapping name as fallback
batting_names = (
    player_batting_master
    .drop_duplicates("player_id")
    .set_index("player_id")["player_name"]
)

bowling_names = (
    player_bowling_master
    .drop_duplicates("player_id")
    .set_index("player_id")["player_name"]
)

player_universe["player_name"] = (
    player_universe["player_id"]
    .map(batting_names)
    .fillna(player_universe["player_id"].map(bowling_names))
    .fillna(player_universe["player_name"])
)

# 3. Aggregate competition and team associations
mapping_profile = (
    competition_team_player_mapping
    .groupby("player_id")
    .agg(
        competitions_count=("comp_id", "nunique"),
        competitions_list=(
            "comp_name",
            lambda x: " | ".join(sorted(x.dropna().unique()))
        ),
        teams_list=(
            "team_name",
            lambda x: " | ".join(sorted(x.dropna().unique()))
        )
    )
    .reset_index()
)

# 4. Prepare batting and bowling masters with clear prefixes
batting_master_prefixed = (
    player_batting_master
    .drop(columns=["player_name"])
    .rename(columns={
        col: f"batting_{col}"
        for col in player_batting_master.columns
        if col not in ["player_id", "player_name"]
    })
)

bowling_master_prefixed = (
    player_bowling_master
    .drop(columns=["player_name"])
    .rename(columns={
        col: f"bowling_{col}"
        for col in player_bowling_master.columns
        if col not in ["player_id", "player_name"]
    })
)

# 5. Prepare IPL experience
ipl_experience_prefixed = ipl_player_experience.drop(
    columns=["player_name"]
)

# 6. Merge all sources at player level
tableau_all_players = (
    player_universe
    .merge(
        mapping_profile,
        on="player_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        batting_master_prefixed,
        on="player_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        bowling_master_prefixed,
        on="player_id",
        how="left",
        validate="one_to_one"
    )
    .merge(
        ipl_experience_prefixed,
        on="player_id",
        how="left",
        validate="one_to_one"
    )
)

# 7. Players not listed in IPL should have explicit False / zero values
ipl_bool_cols = ["ipl_ever_listed", "ipl_ever_played"]

for col in ipl_bool_cols:
    tableau_all_players[col] = (
        tableau_all_players[col].fillna(False).astype(bool)
    )

ipl_numeric_cols = [
    "ipl_competitions_listed",
    "ipl_seasons_listed",
    "ipl_matches_played",
    "ipl_seasons_played",
    "ipl_mapping_only_competitions",
    "ipl_impact_in_appearances",
    "ipl_impact_out_appearances"
]

tableau_all_players[ipl_numeric_cols] = (
    tableau_all_players[ipl_numeric_cols].fillna(0).astype(int)
)

# 8. Validate
print("=" * 65)
print("INITIAL TABLEAU PLAYER MASTER")
print("=" * 65)

print("Total rows:", len(tableau_all_players))
print("Unique players:", tableau_all_players["player_id"].nunique())
print("Duplicate player IDs:", tableau_all_players["player_id"].duplicated().sum())
print("Total columns:", len(tableau_all_players.columns))

print("\nPlayers with batting data:",
      tableau_all_players["batting_matches_played"].notna().sum())

print("Players with bowling data:",
      tableau_all_players["bowling_matches_played"].notna().sum())

print("Players listed in IPL:",
      tableau_all_players["ipl_ever_listed"].sum())

print("Players with IPL appearances:",
      tableau_all_players["ipl_ever_played"].sum())

print("\nMissing player names:",
      tableau_all_players["player_name"].isna().sum())

print("\nSample:")
display(tableau_all_players.head(10))
```

```python
# Aggregate match participation at player level
player_participation_summary = (
    player_match_analysis
    .groupby("player_id")
    .agg(
        participation_matches_listed=("match_id", "nunique"),

        participation_matches_played=(
            "match_id",
            lambda x: x[
                player_match_analysis.loc[x.index, "is_played"] == True
            ].nunique()
        ),

        participation_captain_matches=(
            "match_id",
            lambda x: x[
                player_match_analysis.loc[x.index, "is_captain"] == True
            ].nunique()
        ),

        participation_wicketkeeper_matches=(
            "match_id",
            lambda x: x[
                player_match_analysis.loc[x.index, "is_wk"] == True
            ].nunique()
        ),

        participation_impact_in_matches=(
            "match_id",
            lambda x: x[
                player_match_analysis.loc[x.index, "is_impact"] == 1
            ].nunique()
        ),

        participation_impact_out_matches=(
            "match_id",
            lambda x: x[
                player_match_analysis.loc[x.index, "is_impact_off"] == 1
            ].nunique()
        )
    )
    .reset_index()
)

# Merge into the player master
tableau_all_players = tableau_all_players.merge(
    player_participation_summary,
    on="player_id",
    how="left",
    validate="one_to_one"
)

participation_cols = [
    col for col in player_participation_summary.columns
    if col != "player_id"
]

tableau_all_players[participation_cols] = (
    tableau_all_players[participation_cols]
    .fillna(0)
    .astype(int)
)

print("=" * 65)
print("PARTICIPATION SUMMARY VALIDATION")
print("=" * 65)

print("Master rows:", len(tableau_all_players))
print("Unique players:", tableau_all_players["player_id"].nunique())
print("Duplicate IDs:", tableau_all_players["player_id"].duplicated().sum())

for col in participation_cols:
    print(f"{col}: {tableau_all_players[col].sum()}")

print("\nSample:")
display(
    tableau_all_players[
        ["player_id", "player_name"] + participation_cols
    ].head(10)
)
```

```python
# Find duplicate player + match combinations
player_match_duplicates = (
    player_match_analysis
    .groupby(["player_id", "match_id"])
    .agg(
        records=("player_id", "size"),
        is_played_values=("is_played", "nunique"),
        captain_values=("is_captain", "nunique"),
        wk_values=("is_wk", "nunique"),
        impact_values=("is_impact", "nunique"),
        impact_off_values=("is_impact_off", "nunique")
    )
    .reset_index()
)

duplicates = player_match_duplicates[
    player_match_duplicates["records"] > 1
]

print("=" * 65)
print("PLAYER-MATCH DUPLICATE AUDIT")
print("=" * 65)

print("Raw player-match rows:", len(player_match_analysis))
print("Unique player-match combinations:",
      player_match_analysis[["player_id", "match_id"]].drop_duplicates().shape[0])
print("Duplicate player-match combinations:", len(duplicates))
print("Extra duplicate rows:", duplicates["records"].sub(1).sum())

print("\nDuplicate groups with conflicting flags:")
print(
    duplicates[
        (duplicates["is_played_values"] > 1) |
        (duplicates["captain_values"] > 1) |
        (duplicates["wk_values"] > 1) |
        (duplicates["impact_values"] > 1) |
        (duplicates["impact_off_values"] > 1)
    ].shape[0]
)

print("\nDuplicate examples:")
display(duplicates.head(20))

print("\nRaw is_played row counts:")
print(player_match_analysis["is_played"].value_counts(dropna=False))
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# --------------------------------------------------
# 1. Define output and grouping keys
# --------------------------------------------------

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/BestPlayersUnderPressure/08_Tableau_Ready"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "tableau_player_competition_season.csv"

KEYS = ["player_id", "comp_id", "comp_name", "season", "match_format"]

# Normalize match-condition labels without changing original DataFrames
bat = batter_match_analysis.copy()
bowl = bowler_data.copy()

bat["_condition"] = bat["match_condition"].astype(str).str.upper()
bowl["_condition"] = bowl["match_condition"].astype(str).str.upper()

regular_bat = bat[bat["_condition"].isin(["STANDARD", "REVISED"])].copy()
super_bat = bat[bat["_condition"] == "SUPER_OVER"].copy()

regular_bowl = bowl[bowl["_condition"].isin(["STANDARD", "REVISED"])].copy()
super_bowl = bowl[bowl["_condition"] == "SUPER_OVER"].copy()


# --------------------------------------------------
# 2. Aggregation helpers
# --------------------------------------------------

def aggregate_batting(df, prefix):
    if df.empty:
        return pd.DataFrame(columns=KEYS)

    return (
        df.groupby(KEYS, dropna=False)
        .agg(
            **{
                f"{prefix}_matches": ("match_id", "nunique"),
                f"{prefix}_innings": ("match_id", "size"),
                f"{prefix}_runs": ("overall_runs", "sum"),
                f"{prefix}_balls": ("overall_balls_faced", "sum"),
                f"{prefix}_fours": ("overall_fours", "sum"),
                f"{prefix}_sixes": ("overall_sixes", "sum"),
                f"{prefix}_pressure_runs": ("under_pressure_runs", "sum"),
                f"{prefix}_pressure_balls": ("under_pressure_balls_faced", "sum"),
                f"{prefix}_pressure_fours": ("under_pressure_fours", "sum"),
                f"{prefix}_pressure_sixes": ("under_pressure_sixes", "sum"),
            }
        )
        .reset_index()
    )


def aggregate_bowling(df, prefix):
    if df.empty:
        return pd.DataFrame(columns=KEYS)

    return (
        df.groupby(KEYS, dropna=False)
        .agg(
            **{
                f"{prefix}_matches": ("match_id", "nunique"),
                f"{prefix}_innings": ("match_id", "size"),
                f"{prefix}_legal_balls": ("legal_balls", "sum"),
                f"{prefix}_runs_conceded": ("total_runs_conceded", "sum"),
                f"{prefix}_wickets": ("wickets", "sum"),
                f"{prefix}_pressure_legal_balls": ("under_pressure_legal_balls", "sum"),
                f"{prefix}_pressure_runs_conceded": ("under_pressure_runs_conceded", "sum"),
                f"{prefix}_pressure_wickets": ("under_pressure_wickets", "sum"),
                f"{prefix}_knockout_pressure_runs": ("knockout_under_pressure_runs", "sum"),
                f"{prefix}_knockout_pressure_balls": ("knockout_under_pressure_balls", "sum"),
                f"{prefix}_knockout_pressure_wickets": ("knockout_under_pressure_wickets", "sum"),
                f"{prefix}_ordinary_pressure_runs": ("ordinary_under_pressure_runs", "sum"),
                f"{prefix}_ordinary_pressure_balls": ("ordinary_under_pressure_balls", "sum"),
                f"{prefix}_ordinary_pressure_wickets": ("ordinary_under_pressure_wickets", "sum"),
            }
        )
        .reset_index()
    )


# --------------------------------------------------
# 3. Build batting aggregates
# --------------------------------------------------

batting_parts = [
    aggregate_batting(regular_bat, "batting"),
    aggregate_batting(super_bat, "super_over_batting"),
    aggregate_batting(
        regular_bat[regular_bat["_condition"] == "STANDARD"],
        "standard_batting"
    ),
    aggregate_batting(
        regular_bat[regular_bat["_condition"] == "REVISED"],
        "revised_batting"
    ),
]

# Knockout and ordinary batting splits, within regular matches
for label, subset in [
    ("knockout_batting", regular_bat[regular_bat["is_knockout"].eq(True)]),
    ("ordinary_batting", regular_bat[regular_bat["is_knockout"].eq(False)])
]:
    if not subset.empty:
        batting_parts.append(
            subset.groupby(KEYS, dropna=False)
            .agg(
                **{
                    f"{label}_matches": ("match_id", "nunique"),
                    f"{label}_runs": ("overall_runs", "sum"),
                    f"{label}_balls": ("overall_balls_faced", "sum"),
                    f"{label}_pressure_runs": ("under_pressure_runs", "sum"),
                    f"{label}_pressure_balls": ("under_pressure_balls_faced", "sum"),
                }
            )
            .reset_index()
        )

batting_stats = batting_parts[0]
for part in batting_parts[1:]:
    batting_stats = batting_stats.merge(part, on=KEYS, how="outer")


# --------------------------------------------------
# 4. Build bowling aggregates
# --------------------------------------------------

bowling_parts = [
    aggregate_bowling(regular_bowl, "bowling"),
    aggregate_bowling(super_bowl, "super_over_bowling"),
    aggregate_bowling(
        regular_bowl[regular_bowl["_condition"] == "STANDARD"],
        "standard_bowling"
    ),
    aggregate_bowling(
        regular_bowl[regular_bowl["_condition"] == "REVISED"],
        "revised_bowling"
    ),
]

bowling_stats = bowling_parts[0]
for part in bowling_parts[1:]:
    bowling_stats = bowling_stats.merge(part, on=KEYS, how="outer")


# --------------------------------------------------
# 5. Create player + competition + season universe
#    This retains listed players with no appearances.
# --------------------------------------------------

competition_meta = (
    competitions[["comp_id", "season", "match_format"]]
    .drop_duplicates("comp_id")
)

mapping = competition_team_player_mapping.merge(
    competition_meta,
    on="comp_id",
    how="left",
    suffixes=("", "_metadata")
)

# Prefer season and format from competition metadata
if "season_metadata" in mapping.columns:
    mapping["season"] = mapping["season_metadata"].combine_first(mapping["season"])
    mapping.drop(columns=["season_metadata"], inplace=True)

if "match_format_metadata" in mapping.columns:
    mapping["match_format"] = mapping["match_format_metadata"].combine_first(
        mapping["match_format"]
    )
    mapping.drop(columns=["match_format_metadata"], inplace=True)

player_competition = (
    mapping.groupby(KEYS, dropna=False)
    .agg(
        player_name=("player_name", "first"),
        teams_list=(
            "team_name",
            lambda x: ", ".join(sorted(x.dropna().astype(str).unique()))
        ),
    )
    .reset_index()
)


# --------------------------------------------------
# 6. Merge batting and bowling into the player universe
# --------------------------------------------------

tableau_player_competition_season = (
    player_competition
    .merge(batting_stats, on=KEYS, how="left")
    .merge(bowling_stats, on=KEYS, how="left")
)


# --------------------------------------------------
# 7. Fill missing aggregate counts and totals with zero
#    Missing rate metrics will be calculated as null
# --------------------------------------------------

metric_columns = [
    col for col in tableau_player_competition_season.columns
    if col not in KEYS + ["player_name", "teams_list"]
]

tableau_player_competition_season[metric_columns] = (
    tableau_player_competition_season[metric_columns].fillna(0)
)


# --------------------------------------------------
# 8. Calculate derived performance rates
# --------------------------------------------------

def safe_divide(numerator, denominator, multiplier=1):
    return np.where(
        denominator > 0,
        numerator / denominator * multiplier,
        np.nan
    )

# Batting strike rates
for prefix in [
    "batting", "super_over_batting",
    "standard_batting", "revised_batting",
    "knockout_batting", "ordinary_batting"
]:
    runs_col = f"{prefix}_runs"
    balls_col = f"{prefix}_balls"

    if runs_col in tableau_player_competition_season.columns:
        tableau_player_competition_season[f"{prefix}_strike_rate"] = safe_divide(
            tableau_player_competition_season[runs_col],
            tableau_player_competition_season[balls_col],
            100
        )

# Bowling economy, average and strike rate
for prefix in [
    "bowling", "super_over_bowling",
    "standard_bowling", "revised_bowling"
]:
    balls = tableau_player_competition_season[f"{prefix}_legal_balls"]
    runs = tableau_player_competition_season[f"{prefix}_runs_conceded"]
    wickets = tableau_player_competition_season[f"{prefix}_wickets"]

    tableau_player_competition_season[f"{prefix}_economy"] = safe_divide(
        runs, balls, 6
    )
    tableau_player_competition_season[f"{prefix}_average"] = safe_divide(
        runs, wickets
    )
    tableau_player_competition_season[f"{prefix}_strike_rate"] = safe_divide(
        balls, wickets
    )

# Pressure bowling rates
for prefix in ["bowling", "super_over_bowling"]:
    balls = tableau_player_competition_season[f"{prefix}_pressure_legal_balls"]
    runs = tableau_player_competition_season[f"{prefix}_pressure_runs_conceded"]
    wickets = tableau_player_competition_season[f"{prefix}_pressure_wickets"]

    tableau_player_competition_season[f"{prefix}_pressure_economy"] = safe_divide(
        runs, balls, 6
    )
    tableau_player_competition_season[f"{prefix}_pressure_average"] = safe_divide(
        runs, wickets
    )
    tableau_player_competition_season[f"{prefix}_pressure_strike_rate"] = safe_divide(
        balls, wickets
    )


# --------------------------------------------------
# 9. Export CSV
# --------------------------------------------------

tableau_player_competition_season = (
    tableau_player_competition_season
    .sort_values(["player_name", "season", "comp_name"], na_position="last")
    .reset_index(drop=True)
)

tableau_player_competition_season.to_csv(OUTPUT_FILE, index=False)

print("Export completed successfully!")
print("File:", OUTPUT_FILE)
print("Rows:", len(tableau_player_competition_season))
print("Columns:", len(tableau_player_competition_season.columns))
print("Unique players:", tableau_player_competition_season["player_id"].nunique())
print("Unique competitions:", tableau_player_competition_season["comp_id"].nunique())
print("Unique player-competition-season rows:",
      tableau_player_competition_season[KEYS[:2] + ["season"]].drop_duplicates().shape[0])
print("File size (MB):", round(OUTPUT_FILE.stat().st_size / (1024 * 1024), 2))

display(tableau_player_competition_season.head(5))
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

# ==========================================================
# 1. CONFIGURATION
# ==========================================================

OUTPUT_DIR = Path(
    "/content/drive/MyDrive/BestPlayersUnderPressure/08_Tableau_Ready"
)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "tableau_player_competition_season.csv"

# Exactly one row per player + competition + season
KEYS = ["player_id", "comp_id", "comp_name", "season"]


# ==========================================================
# 2. PREPARE SOURCE DATA
# ==========================================================

bat = batter_match_analysis.copy().rename(columns={
    "striker_id": "player_id",
    "striker_name": "player_name"
})

bowl = bowler_data.copy().rename(columns={
    "bowler_id": "player_id",
    "bowler_name": "player_name"
})

# Standardize match conditions
bat["_condition"] = bat["match_condition"].astype(str).str.upper()
bowl["_condition"] = bowl["match_condition"].astype(str).str.upper()

# Regular matches
regular_bat = bat[
    bat["_condition"].isin(["STANDARD", "REVISED"])
].copy()

regular_bowl = bowl[
    bowl["_condition"].isin(["STANDARD", "REVISED"])
].copy()

# Super Overs
super_bat = bat[
    bat["_condition"] == "SUPER_OVER"
].copy()

super_bowl = bowl[
    bowl["_condition"] == "SUPER_OVER"
].copy()


# ==========================================================
# 3. AGGREGATION FUNCTIONS
# ==========================================================

def aggregate_batting(df, prefix):

    metric_names = [
        "matches", "innings", "runs", "balls", "fours", "sixes",
        "pressure_runs", "pressure_balls",
        "pressure_fours", "pressure_sixes"
    ]

    output_columns = KEYS + [
        f"{prefix}_{col}" for col in metric_names
    ]

    if df.empty:
        return pd.DataFrame(columns=output_columns)

    result = (
        df.groupby(KEYS, dropna=False)
        .agg(
            matches=("match_id", "nunique"),
            innings=("match_id", "size"),
            runs=("overall_runs", "sum"),
            balls=("overall_balls_faced", "sum"),
            fours=("overall_fours", "sum"),
            sixes=("overall_sixes", "sum"),
            pressure_runs=("under_pressure_runs", "sum"),
            pressure_balls=("under_pressure_balls_faced", "sum"),
            pressure_fours=("under_pressure_fours", "sum"),
            pressure_sixes=("under_pressure_sixes", "sum")
        )
        .reset_index()
    )

    return result.rename(columns={
        col: f"{prefix}_{col}"
        for col in metric_names
    })


def aggregate_bowling(df, prefix):

    metric_names = [
        "matches", "innings", "legal_balls", "runs_conceded", "wickets",
        "pressure_legal_balls", "pressure_runs_conceded",
        "pressure_wickets",
        "knockout_pressure_runs", "knockout_pressure_balls",
        "knockout_pressure_wickets",
        "ordinary_pressure_runs", "ordinary_pressure_balls",
        "ordinary_pressure_wickets"
    ]

    output_columns = KEYS + [
        f"{prefix}_{col}" for col in metric_names
    ]

    if df.empty:
        return pd.DataFrame(columns=output_columns)

    result = (
        df.groupby(KEYS, dropna=False)
        .agg(
            matches=("match_id", "nunique"),
            innings=("match_id", "size"),
            legal_balls=("legal_balls", "sum"),
            runs_conceded=("total_runs_conceded", "sum"),
            wickets=("wickets", "sum"),
            pressure_legal_balls=("under_pressure_legal_balls", "sum"),
            pressure_runs_conceded=("under_pressure_runs_conceded", "sum"),
            pressure_wickets=("under_pressure_wickets", "sum"),
            knockout_pressure_runs=("knockout_under_pressure_runs", "sum"),
            knockout_pressure_balls=("knockout_under_pressure_balls", "sum"),
            knockout_pressure_wickets=("knockout_under_pressure_wickets", "sum"),
            ordinary_pressure_runs=("ordinary_under_pressure_runs", "sum"),
            ordinary_pressure_balls=("ordinary_under_pressure_balls", "sum"),
            ordinary_pressure_wickets=("ordinary_under_pressure_wickets", "sum")
        )
        .reset_index()
    )

    return result.rename(columns={
        col: f"{prefix}_{col}"
        for col in metric_names
    })


def merge_aggregate_parts(parts):

    result = parts[0]

    for part in parts[1:]:
        result = result.merge(
            part,
            on=KEYS,
            how="outer",
            validate="one_to_one"
        )

    return result


# ==========================================================
# 4. BATTING AGGREGATIONS
# ==========================================================

batting_parts = [
    aggregate_batting(regular_bat, "batting"),

    aggregate_batting(super_bat, "super_over_batting"),

    aggregate_batting(
        regular_bat[regular_bat["_condition"] == "STANDARD"],
        "standard_batting"
    ),

    aggregate_batting(
        regular_bat[regular_bat["_condition"] == "REVISED"],
        "revised_batting"
    ),

    aggregate_batting(
        regular_bat[regular_bat["is_knockout"].eq(True)],
        "knockout_batting"
    ),

    aggregate_batting(
        regular_bat[regular_bat["is_knockout"].eq(False)],
        "ordinary_batting"
    )
]

batting_stats = merge_aggregate_parts(batting_parts)


# ==========================================================
# 5. BOWLING AGGREGATIONS
# ==========================================================

bowling_parts = [
    aggregate_bowling(regular_bowl, "bowling"),

    aggregate_bowling(super_bowl, "super_over_bowling"),

    aggregate_bowling(
        regular_bowl[regular_bowl["_condition"] == "STANDARD"],
        "standard_bowling"
    ),

    aggregate_bowling(
        regular_bowl[regular_bowl["_condition"] == "REVISED"],
        "revised_bowling"
    )
]

bowling_stats = merge_aggregate_parts(bowling_parts)


# ==========================================================
# 6. BUILD PLAYER-COMPETITION-SEASON UNIVERSE
# ==========================================================

competition_meta = (
    competitions[
        ["comp_id", "season", "match_format"]
    ]
    .drop_duplicates("comp_id")
)

mapping = competition_team_player_mapping.copy()

# Use competition metadata as the source of season and format
mapping = mapping.drop(
    columns=["season", "match_format"],
    errors="ignore"
)

mapping = mapping.merge(
    competition_meta,
    on="comp_id",
    how="left",
    validate="many_to_one"
)

# One row per player, competition and season
player_competition = (
    mapping.groupby(KEYS, dropna=False)
    .agg(
        player_name=("player_name", "first"),
        match_format=("match_format", "first"),
        teams_list=(
            "team_name",
            lambda x: ", ".join(
                sorted(x.dropna().astype(str).unique())
            )
        )
    )
    .reset_index()
)


# ==========================================================
# 7. MERGE BATTING AND BOWLING
# ==========================================================

tableau_player_competition_season = (
    player_competition
    .merge(
        batting_stats,
        on=KEYS,
        how="left",
        validate="one_to_one"
    )
    .merge(
        bowling_stats,
        on=KEYS,
        how="left",
        validate="one_to_one"
    )
)


# ==========================================================
# 8. FILL MISSING AGGREGATED METRICS
# ==========================================================

metric_columns = [
    col for col in tableau_player_competition_season.columns
    if col not in KEYS + ["player_name", "match_format", "teams_list"]
]

tableau_player_competition_season[metric_columns] = (
    tableau_player_competition_season[metric_columns].fillna(0)
)


# ==========================================================
# 9. CALCULATE BATTING AND BOWLING RATES
# ==========================================================

def safe_divide(numerator, denominator, multiplier=1):

    numerator = np.asarray(numerator, dtype=float)
    denominator = np.asarray(denominator, dtype=float)

    return np.divide(
        numerator * multiplier,
        denominator,
        out=np.full(len(numerator), np.nan),
        where=denominator > 0
    )


# Batting strike rates
for prefix in [
    "batting",
    "super_over_batting",
    "standard_batting",
    "revised_batting",
    "knockout_batting",
    "ordinary_batting"
]:

    runs_col = f"{prefix}_runs"
    balls_col = f"{prefix}_balls"

    tableau_player_competition_season[
        f"{prefix}_strike_rate"
    ] = safe_divide(
        tableau_player_competition_season[runs_col],
        tableau_player_competition_season[balls_col],
        100
    )


# Bowling economy, average and strike rate
for prefix in [
    "bowling",
    "super_over_bowling",
    "standard_bowling",
    "revised_bowling"
]:

    balls = tableau_player_competition_season[
        f"{prefix}_legal_balls"
    ]

    runs = tableau_player_competition_season[
        f"{prefix}_runs_conceded"
    ]

    wickets = tableau_player_competition_season[
        f"{prefix}_wickets"
    ]

    tableau_player_competition_season[
        f"{prefix}_economy"
    ] = safe_divide(runs, balls, 6)

    tableau_player_competition_season[
        f"{prefix}_average"
    ] = safe_divide(runs, wickets)

    tableau_player_competition_season[
        f"{prefix}_strike_rate"
    ] = safe_divide(balls, wickets)


# Pressure bowling rates
for prefix in ["bowling", "super_over_bowling"]:

    balls = tableau_player_competition_season[
        f"{prefix}_pressure_legal_balls"
    ]

    runs = tableau_player_competition_season[
        f"{prefix}_pressure_runs_conceded"
    ]

    wickets = tableau_player_competition_season[
        f"{prefix}_pressure_wickets"
    ]

    tableau_player_competition_season[
        f"{prefix}_pressure_economy"
    ] = safe_divide(runs, balls, 6)

    tableau_player_competition_season[
        f"{prefix}_pressure_average"
    ] = safe_divide(runs, wickets)

    tableau_player_competition_season[
        f"{prefix}_pressure_strike_rate"
    ] = safe_divide(balls, wickets)


# ==========================================================
# 10. ADD PERFORMANCE PRESENCE FLAGS
# ==========================================================

batting_presence = (
    regular_bat.groupby(KEYS, dropna=False)
    .size()
    .rename("batting_records")
    .reset_index()
)

bowling_presence = (
    regular_bowl.groupby(KEYS, dropna=False)
    .size()
    .rename("bowling_records")
    .reset_index()
)

super_over_presence = (
    pd.concat(
        [super_bat[KEYS], super_bowl[KEYS]],
        ignore_index=True
    )
    .drop_duplicates()
)

super_over_presence["has_super_over_record"] = True

tableau_player_competition_season = (
    tableau_player_competition_season
    .merge(
        batting_presence,
        on=KEYS,
        how="left",
        validate="one_to_one"
    )
    .merge(
        bowling_presence,
        on=KEYS,
        how="left",
        validate="one_to_one"
    )
    .merge(
        super_over_presence,
        on=KEYS,
        how="left",
        validate="one_to_one"
    )
)

tableau_player_competition_season[
    ["batting_records", "bowling_records"]
] = tableau_player_competition_season[
    ["batting_records", "bowling_records"]
].fillna(0).astype(int)

tableau_player_competition_season["has_batting_record"] = (
    tableau_player_competition_season["batting_records"] > 0
)

tableau_player_competition_season["has_bowling_record"] = (
    tableau_player_competition_season["bowling_records"] > 0
)

tableau_player_competition_season["has_super_over_record"] = (
    tableau_player_competition_season["has_super_over_record"]
    .fillna(False)
    .astype(bool)
)

tableau_player_competition_season["has_match_performance_record"] = (
    tableau_player_competition_season["has_batting_record"]
    | tableau_player_competition_season["has_bowling_record"]
    | tableau_player_competition_season["has_super_over_record"]
)


# ==========================================================
# 11. FINAL ORDER AND EXPORT
# ==========================================================

tableau_player_competition_season = (
    tableau_player_competition_season
    .sort_values(
        ["player_name", "season", "comp_name"],
        na_position="last"
    )
    .reset_index(drop=True)
)

# Check duplicate player-competition-season combinations
duplicate_count = tableau_player_competition_season.duplicated(
    ["player_id", "comp_id", "season"]
).sum()

if duplicate_count > 0:
    raise ValueError(
        f"Found {duplicate_count} duplicate player-competition-season rows."
    )

# Export
tableau_player_competition_season.to_csv(
    OUTPUT_FILE,
    index=False
)


# ==========================================================
# 12. EXPORT SUMMARY
# ==========================================================

print("=" * 60)
print("TABLEAU PLAYER COMPETITION SEASON EXPORT")
print("=" * 60)

print("File:", OUTPUT_FILE)
print("Rows:", len(tableau_player_competition_season))
print("Columns:", len(tableau_player_competition_season.columns))

print(
    "Unique players:",
    tableau_player_competition_season["player_id"].nunique()
)

print(
    "Unique competitions:",
    tableau_player_competition_season["comp_id"].nunique()
)

print(
    "Unique seasons:",
    tableau_player_competition_season["season"].nunique()
)

print(
    "Players with match performance:",
    int(
        tableau_player_competition_season[
            "has_match_performance_record"
        ].sum()
    )
)

print(
    "Listed players without performance:",
    int(
        (~tableau_player_competition_season[
            "has_match_performance_record"
        ]).sum()
    )
)

print("Duplicate player-competition-season rows:", duplicate_count)

print(
    "File size (MB):",
    round(OUTPUT_FILE.stat().st_size / (1024 * 1024), 2)
)

print("\nSample records:")
display(tableau_player_competition_season.head(10))

```

```python
# Align competition-season player names with the master player dataset

master_names = (
    tableau_all_players[["player_id", "player_name"]]
    .drop_duplicates("player_id")
    .set_index("player_id")["player_name"]
)

tableau_player_competition_season["player_name"] = (
    tableau_player_competition_season["player_id"]
    .map(master_names)
    .fillna(tableau_player_competition_season["player_name"])
)

# Verify
print(
    "Remaining name mismatches:",
    (
        tableau_player_competition_season[
            ["player_id", "player_name"]
        ]
        .drop_duplicates()
        .merge(
            tableau_all_players[["player_id", "player_name"]],
            on="player_id",
            how="inner",
            suffixes=("_competition", "_master")
        )
        .query("player_name_competition != player_name_master")
        .shape[0]
    )
)

# Export corrected file
output_path = (
    "/content/drive/MyDrive/BestPlayersUnderPressure/"
    "08_Tableau_Ready/tableau_player_competition_season.csv"
)

tableau_player_competition_season.to_csv(output_path, index=False)

print("Updated CSV exported:", output_path)
```
